# 3. History of Python

## 3.1 Introduction to Python History

Python is a popular programming language created by **Guido van Rossum**.

Python development started in December 1989 at the Centrum Wiskunde & Informatica (CWI), a research institute in the Netherlands.

Python was first released publicly in February 1991.

The main idea behind creating Python was to develop a programming language that was easy to read, easy to write and powerful enough to perform complex tasks.

Python was designed to emphasize code readability and simplicity.

## 3.2 Who Created Python?

**Guido van Rossum** is the creator of Python.

- Name: Guido van Rossum
- Known as: Creator of Python
- Python development started: December 1989
- First public release: February 1991
- Development location: CWI, Netherlands

Guido wanted to create a programming language that was simple, readable and useful for software development.

Because of his contribution, he is often called the **Father of Python**.

## 3.3 Why Was Python Created?

Python was created with the following goals:

1. To make programming easier to understand.
2. To improve code readability.
3. To reduce unnecessary complexity in programming.
4. To support different types of applications.
5. To help developers write programs with fewer lines of code.

Python focuses on simplicity and readability.

## 3.4 Why is it Called Python?

The name Python was inspired by the British comedy television show **Monty Python's Flying Circus**.

The name was not taken from the snake.

Guido van Rossum wanted a short, unique and interesting name for the programming language.

## 4.1 Introduction to Python Features

Python has many features that make it popular among beginners and professional developers.

A feature is a special characteristic or capability of a programming language.

## 4.2 Simple and Easy to Learn

Python has simple and readable syntax compared with many other programming languages.

Its code is easy to understand, especially for beginners.

**Example:**

```python
print("Welcome to Python")
```

Output:

```text
Welcome to Python
```

Only one line of code is needed to display a message.

## 4.3 High-Level Programming Language

Python is a high-level programming language.

A high-level language allows programmers to write instructions using syntax that is easier for humans to understand.

Programmers do not need to manage most low-level hardware details directly.

Example:

```python
a = 10
b = 20

print(a + b)
```

Output:

```text
30
```

Python handles many lower-level operations internally.

## 4.4 Interpreted Language

Python is commonly described as an interpreted language.

In a typical Python implementation such as CPython, source code is compiled into bytecode and then executed by the Python virtual machine.

This makes running and testing Python programs convenient.

Example:

```python
print("Python is easy")
print("Python is powerful")
```

Output:

```text
Python is easy
Python is powerful
```

## 4.5 Dynamically Typed Language

Python is dynamically typed.

This means you do not need to explicitly declare a variable's data type before assigning a value.

Python determines the type at runtime.

Example:

```python
x = 10
print(x)

x = "Hello"
print(x)
```

Output:

```text
10
Hello
```

Here, the same variable is first assigned an integer and later assigned a string.

## 4.6 Object-Oriented Programming

Python supports object-oriented programming (OOP).

Object-oriented programming is a programming approach based on classes and objects.

It helps programmers organize code and reuse functionality.

Example:

```python
class Student:
    def display(self):
        print("I am a student")

s1 = Student()
s1.display()
```

Output:

```text
I am a student
```

## 4.7 Platform Independent

This makes Python useful across different computer environments.

## 5.3 Disadvantages of Python

### 1. Slower Execution Speed

Python programs can run slower than compiled languages such as C and C++ in many CPU-intensive tasks.

This is partly because of Python's dynamic features and runtime execution model.

### 2. Higher Memory Consumption

Python applications may consume more memory than equivalent programs written in languages such as C.

This can be a limitation in memory-constrained systems.

### 3. Not Ideal for Every Mobile Application

Python is not the most common choice for developing native Android and iOS applications.

Languages and frameworks designed specifically for mobile development are often preferred.

### 4. Dynamic Typing Can Cause Errors

Python does not require explicit variable type declarations.

Although this makes development easier, certain type-related errors may only appear when the program runs.

Example:

```python
age = 20
age = "Twenty"
```

Python allows the variable to refer to values of different types.

This flexibility requires careful programming.

### 5. Global Interpreter Lock in CPython

In standard GIL-enabled CPython builds, the Global Interpreter Lock (GIL) limits how multiple threads execute Python bytecode simultaneously.

This can restrict CPU-bound multithreading.

However, multiprocessing, asynchronous programming and free-threaded Python builds provide alternative approaches.

### 6. Not Suitable for Every Low-Level Task

Python is generally not the first choice for tasks requiring direct hardware control or highly optimized low-level programming.

Languages such as C and C++ are often preferred for these tasks.

# 7. Python vs C

## 7.1 Introduction

Python and C are popular programming languages, but they are designed with different priorities.

Python focuses on simplicity, readability and developer productivity.

C focuses on performance, low-level programming and efficient system resource management.

## 7.2 What is C?

C is a general-purpose, procedural programming language developed by Dennis Ritchie at Bell Laboratories.

It was developed in the early 1970s and became widely known through its use in operating systems and system software.

C is commonly used for:

- Operating systems
- Embedded systems
- Device drivers
- System programming
- Compilers

## 7.3 Difference Between Python and C

| Feature | Python | C |
|---|---|---|
| Language Level | High-level | Middle-level (commonly described) |
| Syntax | Simple and readable | More detailed syntax |
| Execution | Commonly uses bytecode and a virtual machine | Usually compiled to native machine code |
| Speed | Generally slower | Generally faster |
| Memory Management | Automatic | Often manual |
| Typing | Dynamically typed | Statically typed |
| Programming Style | Multiple paradigms | Mainly procedural |
| Pointers | No direct general-purpose pointer manipulation | Supports pointers |
| Code Length | Usually shorter | Often longer |
| Learning | Beginner-friendly | Requires understanding of lower-level concepts |
| Applications | AI, web, automation, data science | OS, embedded systems, drivers |

## 7.4 Example: Adding Two Numbers in Python

```python
a = 10
b = 20

result = a + b

print(result)
```

Output:

```text
30
```

## 7.5 Example: Adding Two Numbers in C

```c
#include <stdio.h>

int main()
{
    int a = 10;
    int b = 20;

    int result = a + b;

    printf("%d", result);

    return 0;
}
```

Output:

```text
30
```

## 7.6 Major Differences Explained

### 1. Syntax

Python has simpler syntax and requires fewer lines of code for many tasks.

C requires more explicit declarations and syntax.

### 2. Execution Speed

C programs are generally faster for CPU-intensive tasks because they are usually compiled into native machine code.

Python often requires more runtime processing.

Actual performance depends on the implementation and workload.

### 3. Memory Management

Python manages memory automatically in common situations.

C gives programmers more direct control over memory allocation and deallocation.

### 4. Applications

Python is commonly used for AI, Data Science, web development and automation.

C is commonly used for operating systems, embedded systems and low-level software.

## 7.7 Which One Should You Learn?

Choose Python if:

- You are a beginner.
- You want to learn programming easily.
- You are interested in AI and Data Science.
- You want to build applications quickly.

Choose C if:

- You want to understand memory management.
- You are interested in operating systems.
- You want to learn low-level programming.
- You are interested in embedded systems.

## 7.8 Conclusion

Python and C are both powerful programming languages.

Python is preferred for simplicity and rapid development, while C is preferred for low-level control and performance-sensitive applications.

### 3. Memory Management

### 4. Development Speed

# 11. Python Interpreter

## 11.1 Introduction

A Python interpreter is the software responsible for executing Python programs.

It processes Python code and helps the computer perform the instructions written by the programmer.

In simple words:

**A Python interpreter is a program that executes Python code.**

## 11.2 What is an Interpreter?

An interpreter is a software program that executes instructions written in a programming language.

It provides a runtime environment in which a program can run.

Python commonly uses an interpreter-based execution model.

## 11.3 How Does the Python Interpreter Work?

In the commonly used CPython implementation, Python source code goes through several stages.

### Step 1: Write Python Code

The programmer writes code in a Python file.

Example:

```python
print("Hello World")
```

### Step 2: Lexical and Syntax Processing

Python processes the source code and checks whether it follows Python's syntax rules.

### Step 3: Bytecode Compilation

CPython compiles the source code into an intermediate representation called bytecode.

### Step 4: Bytecode Execution

The Python virtual machine executes the bytecode.

### Step 5: Output

The program produces the required result.

# 12. Compiler vs Interpreter

## 12.1 Introduction

A compiler and an interpreter are tools used to process programming languages.

Both help computers execute programs written by programmers.

However, they use different execution approaches.

## 12.2 What is a Compiler?

A compiler is software that translates source code into another form, often machine code or an intermediate representation.

In a traditional compiled-language workflow, the compiler translates the program before execution.

Example languages that commonly use compilation:

- C
- C++
- Rust

## 12.3 What is an Interpreter?

An interpreter is software that executes a program using a runtime environment.

It processes program instructions and performs the required operations.

Examples of languages commonly associated with interpreter-based execution:

- Python
- JavaScript
- Ruby

Note: Modern language implementations often combine compilation and interpretation.

## 12.4 Difference Between Compiler and Interpreter

| Feature | Compiler | Interpreter |
|---|---|---|
| Main Purpose | Translates source code | Executes program instructions |
| Translation | Commonly translates a program before execution | Processes code through a runtime |
| Execution | Often produces a separate executable | Often executes through a runtime environment |
| Error Reporting | May report multiple errors during compilation | Often reports errors when execution reaches them |
| Performance | Native compiled programs are often faster | Runtime execution can have additional overhead |
| Output | May produce machine code or another target format | May execute code without a separate native executable |
| Examples | GCC, Clang | CPython interpreter, JavaScript engines |

## 12.5 Compiler Working Diagram

```text
      Source Code
           |
           v
        Compiler
           |
           v
     Machine Code
           |
           v
       Execution
           |
           v
         Output
```

## 12.6 Interpreter Working Diagram

```text
      Source Code
           |
           v
       Interpreter
           |
           v
        Execution
           |
           v
         Output
```

These diagrams show simplified traditional models.

## 12.7 Real Example: C Program

```c
#include <stdio.h>

int main()
{
    printf("Hello World");
    return 0;
}
```

A common C workflow:

1. Write the source code.
2. Compile the program using a compiler.
3. Generate an executable.
4. Run the executable.
5. Get the output.

Example command using GCC:

```bash
gcc hello.c -o hello
```

Run on Windows:

```bash
hello.exe
```

Run on Linux or macOS:

```bash
./hello
```

## 12.8 Real Example: Python Program

```python
print("Hello World")
```

Run using:

```bash
python hello.py
```

On some systems, use:

```bash
python3 hello.py
```

Python's common CPython implementation internally compiles source code into bytecode before executing it.

Therefore, it is not technically correct to say that Python never uses compilation.

## 12.9 Compiler and Interpreter in Modern Programming

Modern language implementations can combine both techniques.

Examples:

- CPython compiles source code to bytecode and executes it using a virtual machine.
- Java compiles source code into bytecode, which the JVM executes using interpretation and Just-In-Time compilation.
- Modern JavaScript engines can interpret and compile code using Just-In-Time techniques.

## 12.10 Conclusion

A compiler translates source code into another representation, often before execution.

An interpreter executes code through a runtime environment.

Both are important in programming, and modern programming languages often combine compilation and interpretation techniques.

## 13.1 Introduction

Python has evolved over time.

A Python version identifies a particular release of the Python programming language.

## 13.2 Major Python Versions

Python has two historically important major version families:

1. Python 2
2. Python 3

### Python 2

Python 2 was introduced in 2000.

It included several improvements over Python 1.

However, Python 2 reached its official End of Life on January 1, 2020.

This means it no longer receives official security updates or bug fixes from the Python core team.

### Python 3

Python 3 was released in 2008.

It introduced important language improvements and some changes that were not fully compatible with Python 2.

Python 3 is the modern version family used for current Python development.

| Year | Version | Important Information |

| 2008 | Python 3.0 | Major language update |
| 2020 | Python 2 End of Life | Official support ended |
| Present | Python 3.x | Modern Python development |

## 13.4 Python Version Numbering

Python versions generally follow this format:

```text
MAJOR.MINOR.MICRO
```

Example:

```text
3.12.4
```

Explanation:

- `3` = Major version
- `12` = Minor version
- `4` = Micro or patch version

### Major Version

A major version identifies a major generation of the language.

Example: Python 2 and Python 3.

### Minor Version

A minor version introduces new features and improvements.

Example: Python 3.11 and Python 3.12.

### Micro Version

A micro version generally includes bug fixes and security improvements.

Example: Python 3.12.1 and Python 3.12.2.

Parentheses are used to pass information to the print function.
