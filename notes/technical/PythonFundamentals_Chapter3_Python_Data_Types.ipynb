# Chapter 3: Python Data Types

# 1. What Are Data Types?

## 1.1 Introduction to Data Types

A data type is a classification that tells Python what kind of value an object represents and what operations can be performed on that value.

In programming, we work with different kinds of information, such as numbers, text, logical values and collections of data. Each kind of information has different characteristics and requires different operations.

For example:

- A person's age is generally represented using an integer.
- A product price is generally represented using a floating-point number.
- A person's name is represented using a string.
- A student's result can be represented using a Boolean value.
- A collection of student names can be represented using a list.

Python automatically determines the type of an object when a value is assigned or created.

## 1.2 Why Are Data Types Important?

Data types are important because they determine how Python stores, interprets and processes information.

The main purposes of data types are:

1. **Identifying Information:** Data types help Python identify whether a value represents a number, text, a logical value or a collection.

2. **Performing Operations:** Different data types support different operations. For example, mathematical addition can be performed on numbers, while string concatenation combines text.

3. **Managing Data:** Data types help programmers organize information in an appropriate structure.

4. **Avoiding Errors:** Understanding data types helps programmers avoid invalid operations, such as attempting to subtract a string from an integer.

5. **Memory Management:** Different object types have different internal representations and memory requirements.

6. **Data Processing:** Choosing an appropriate data type makes it easier to process, modify and retrieve information.

## 1.3 Categories of Python Data Types

Python provides several built-in data types.

They can be organized into the following categories:

### A. Numeric Data Types

Used to represent numerical values.

- `int` – Integer numbers.
- `float` – Decimal or floating-point numbers.
- `complex` – Complex numbers.

### B. Boolean Data Type

- `bool` – Represents `True` or `False`.

### C. Text Data Type

- `str` – Represents text or a sequence of characters.

### D. Sequence Data Types

Used to store multiple values in an ordered structure.

- `list`
- `tuple`
- `range`

### E. Set Data Types

Used to store collections of unique elements.

- `set`
- `frozenset`

### F. Mapping Data Type

- `dict` – Stores information as key-value pairs.

### G. None Type

- `NoneType` – Represents the absence of a value.

## 1.5 Example of Different Data Types

```python
# Numeric data types
age = 20
price = 99.50
complex_number = 3 + 4j

# Boolean data type
is_student = True

# String data type
name = "Rahul"

# List
marks = [85, 90, 95]

# Tuple
coordinates = (10, 20)

# Set
numbers = {10, 20, 30}

# Dictionary
student = {
    "name": "Rahul",
    "age": 20
}

# NoneType
result = None

# Display data types
print(type(age))
print(type(price))
print(type(complex_number))
print(type(is_student))
print(type(name))
print(type(marks))
print(type(coordinates))
print(type(numbers))
print(type(student))
print(type(result))
```

## 1.6 Conclusion

Data types are fundamental building blocks of Python programming. They help Python understand the nature of information and determine which operations can be performed on it.

Understanding data types is necessary before learning advanced topics such as operators, loops, functions, data structures and object-oriented programming.

# 2. Numeric Data Types

## 2.1 What Are Numeric Data Types?

Numeric data types are used to represent numbers in Python.

They allow programmers to perform mathematical calculations, arithmetic operations, scientific computations and numerical data processing.

Python provides three primary built-in numeric data types:

1. Integer (`int`)
2. Floating-point (`float`)
3. Complex (`complex`)

Each numeric type represents numbers in a different form and supports different mathematical operations.

## 2.2 Types of Numeric Data

### A. Integer (int)

An integer represents a whole number without a fractional or decimal part.

Examples:

```python
age = 20
temperature = -5
marks = 100
population = 500000
```

Integers can be:

- Positive numbers.
- Negative numbers.
- Zero.

### B. Float (float)

A floating-point number represents a number with a fractional component.

Examples:

```python
price = 99.50
height = 5.8
temperature = 36.5
percentage = 85.75
```

Floating-point numbers are useful when decimal precision is required, although they have finite binary precision.

### C. Complex (complex)

A complex number consists of two parts:

- Real part.
- Imaginary part.

Example:

```python
number = 3 + 4j
```

Here:

- `3` is the real part.
- `4j` is the imaginary part.

Python uses `j` to represent the imaginary unit.

| Feature | int | float | complex |

Python provides the `type()` function to identify the type of a numeric value.

```python
a = 100
b = 10.5
c = 3 + 4j

print(type(a))
print(type(b))
print(type(c))
```

Output:

```text
<class 'int'>
<class 'float'>
<class 'complex'>
```

Python provides three primary numeric data types: `int`, `float` and `complex`. They allow programmers to perform different kinds of numerical calculations, from simple counting to scientific computing.

# 3. Integer (int)

## 3.1 What Is an Integer?

An integer is a numeric data type used to represent whole numbers without a fractional or decimal component.

In Python, integers are represented using the `int` data type.

An integer can be positive, negative or zero.

Unlike some programming languages, Python integers can grow beyond fixed 32-bit or 64-bit limits, subject to available memory.

## 3.2 Examples of Integer Values

```python
a = 10
b = 200
c = -50
d = 0

print(a)
print(b)
print(c)
print(d)
```

Output:

```text
10
200
-50
0
```

All four variables refer to integer objects.

## 3.3 Characteristics of Integers

1. **Whole Numbers:** Integers do not contain a fractional component.

2. **Positive and Negative:** Integers can represent positive values, negative values and zero.

3. **Arbitrary Precision:** Python integers can represent very large numbers, limited mainly by available memory.

4. **Immutable:** Integer objects cannot be modified after creation. Reassigning a variable makes it refer to another object.

5. **Mathematical Operations:** Integers support addition, subtraction, multiplication, division, floor division, modulus and exponentiation.

6. **No Decimal Part:** An integer such as `100` is different from a floating-point number such as `100.0`.

## 3.4 Integer Literals

An integer literal is a whole-number value written directly in Python code.

Examples:

```python
a = 100
b = -250
c = 0
d = 1000000
```

Python also supports different number systems.

### Decimal Numbers (Base 10)

Decimal numbers use digits from 0 to 9.

```python
number = 100
print(number)
```

### Binary Numbers (Base 2)

Binary numbers use only 0 and 1.

They are written using the prefix `0b` or `0B`.

```python
number = 0b1010

print(number)
```

Output:

```text
10
```

### Octal Numbers (Base 8)

Octal numbers use digits from 0 to 7.

They are written using the prefix `0o` or `0O`.

```python
number = 0o12

print(number)
```

Output:

```text
10
```

### Hexadecimal Numbers (Base 16)

Hexadecimal numbers use digits 0–9 and letters A–F.

They are written using the prefix `0x` or `0X`.

```python
number = 0xA

print(number)
```

Output:

```text
10
```

## 3.5 Integer Operations

```python
a = 20
b = 10

addition = a + b
subtraction = a - b
multiplication = a * b
division = a / b
floor_division = a // b
remainder = a % b

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
print("Floor Division:", floor_division)
print("Remainder:", remainder)
```

Output:

```text
Addition: 30
Subtraction: 10
Multiplication: 200
Division: 2.0
Floor Division: 2
Remainder: 0
```

Important: The `/` operator performs true division and normally produces a float, even when both operands are integers.

## 3.6 Integer Conversion Using int()

The `int()` function converts compatible values into integers.

Example:

```python
a = int("100")
b = int(25.8)

print(a)
print(b)
```

Output:

```text
100
25
```

When converting a float to an integer, Python truncates the fractional part toward zero. It does not round to the nearest integer.

For example:

```python
print(int(-25.8))
```

Output:

```text
-25
```

## 3.7 Finding the Type of an Integer

```python
age = 20

print(type(age))
```

Output:

```text
<class 'int'>
```

## 3.8 Real-Life Applications of Integers

Integers are commonly used for:

- Age in completed years.
- Number of students.
- Product quantities.
- Counting objects.
- Number of employees.
- Page numbers.
- Database record counts.

## 3.9 Conclusion

The `int` data type represents whole numbers in Python. It is commonly used for counting, quantities, indexing and mathematical calculations.

# 4. Float (float)

## 4.1 What Is a Float?

A float, or floating-point number, is a numeric data type used to represent numbers that contain a fractional or decimal component.

Python uses the `float` data type to represent floating-point values.

Floats are useful when calculations involve measurements, prices, percentages, scientific values and other numbers that require decimal components.

Examples:

```python
price = 99.50
height = 5.8
temperature = 36.5
percentage = 85.75
```

## 4.2 Characteristics of Float

1. **Decimal Values:** Floats can represent numbers with fractional components.

2. **Positive and Negative Values:** Float values can be positive, negative or zero.

3. **Scientific Notation:** Python supports floating-point values written using scientific notation.

4. **Finite Precision:** Floating-point values have limited binary precision. Some decimal values cannot be represented exactly.

5. **Immutable:** Float objects cannot be changed after creation.

6. **Mathematical Operations:** Floats support common arithmetic operations.

## 4.3 Examples of Float Values

```python
a = 10.5
b = 25.75
c = -5.8
d = 0.0

print(a)
print(b)
print(c)
print(d)
```

Output:

```text
10.5
25.75
-5.8
0.0
```

## 4.4 Scientific Notation in Python

Python supports scientific notation for representing very large or very small numbers.

Scientific notation uses the letter `e` or `E` to represent a power of 10.

Example:

```python
a = 2.5e3
b = 3.2e-2

print(a)
print(b)
```

Output:

```text
2500.0
0.032
```

Explanation:

- `2.5e3` means 2.5 × 10³.
- `3.2e-2` means 3.2 × 10⁻².

## 4.5 Float Operations

```python
a = 10.5
b = 2.5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

Output:

```text
13.0
8.0
26.25
4.2
```

## 4.6 Floating-Point Precision

One important characteristic of floating-point numbers is that some decimal fractions cannot be represented exactly in binary.

Example:

```python
a = 0.1
b = 0.2

print(a + b)
```

Output:

```text
0.30000000000000004
```

Many beginners expect the result to be exactly `0.3`.

However, binary floating-point representation can introduce small rounding differences.

For approximate comparisons, use `math.isclose()`.

Example:

```python
import math

a = 0.1 + 0.2
b = 0.3

print(math.isclose(a, b))
```

Output:

```text
True
```

For financial calculations requiring exact decimal arithmetic, Python provides the `decimal` module.

## 4.7 Float Conversion Using float()

The `float()` function converts compatible values into floating-point numbers.

Example:

```python
a = float("10.5")
b = float(20)

print(a)
print(b)
```

Output:

```text
10.5
20.0
```

## 4.8 Finding the Type of Float

```python
price = 99.50

print(type(price))
```

Output:

```text
<class 'float'>
```

## 4.9 Real-Life Applications of Float

Floats are commonly used for:

- Product prices.
- Temperature measurements.
- Height and weight measurements.
- Scientific calculations.
- Percentages.
- Distance calculations.
- Financial estimates.

## 4.10 Conclusion

The `float` data type represents floating-point numbers. It is useful for calculations involving fractional values and decimal measurements.

# 5. Complex (complex)

## 5.1 What Is a Complex Number?

A complex number is a number that contains two components: a real part and an imaginary part.

Complex numbers are commonly used in advanced mathematics, engineering, electronics and scientific computing.

In Python, complex numbers are represented using the `complex` data type.

A complex number is generally written in the following mathematical form:

**z = a + bj**

Where:

- `a` represents the real part.
- `b` represents the imaginary coefficient.
- `j` represents the imaginary unit.

The imaginary unit satisfies the mathematical relationship j² = -1.

Python uses `j` instead of the mathematical symbol i.

## 5.2 Syntax of Complex Numbers

```python
variable = real_part + imaginary_part j
```

Example:

```python
number = 3 + 4j

print(number)
```

Output:

```text
(3+4j)
```

## 5.3 Examples of Complex Numbers

```python
a = 3 + 4j
b = 5 + 2j
c = 7j
d = 10 + 0j

print(a)
print(b)
print(c)
print(d)
```

Output:

```text
(3+4j)
(5+2j)
7j
(10+0j)
```

## 5.4 Characteristics of Complex Numbers

1. Complex numbers contain real and imaginary components.

2. Python uses the letter `j` to represent the imaginary component.

3. Complex numbers support addition, subtraction, multiplication and division.

4. Complex numbers are immutable.

5. Complex numbers cannot be directly ordered using comparison operators such as `<` and `>`.

6. Python provides methods and attributes to access the real and imaginary components.

## 5.5 Accessing Real and Imaginary Parts

Python provides the `real` and `imag` attributes.

Example:

```python
number = 3 + 4j

print("Real Part:", number.real)
print("Imaginary Part:", number.imag)
```

Output:

```text
Real Part: 3.0
Imaginary Part: 4.0
```

Both components are returned as floating-point values.

## 5.6 Complex Number Operations

Example:

```python
a = 3 + 4j
b = 2 + 1j

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

Output:

```text
Addition: (5+5j)
Subtraction: (1+3j)
Multiplication: (2+11j)
Division: (2+1j)
```

## 5.7 Creating Complex Numbers Using complex()

Python provides the built-in `complex()` function to create complex numbers.

Syntax:

```python
complex(real, imaginary)
```

Example:

```python
number = complex(3, 4)

print(number)
```

Output:

```text
(3+4j)
```

## 5.8 Conjugate of a Complex Number

The conjugate of a complex number is obtained by changing the sign of its imaginary component.

Python provides the `conjugate()` method.

Example:

```python
number = 3 + 4j

print(number.conjugate())
```

Output:

```text
(3-4j)
```

## 5.9 Finding the Type of Complex

```python
number = 3 + 4j

print(type(number))
```

Output:

```text
<class 'complex'>
```

## 5.10 Real-Life Applications of Complex Numbers

Complex numbers are used in:

- Electrical engineering.
- Signal processing.
- Quantum mechanics.
- Control systems.
- Circuit analysis.
- Advanced mathematical calculations.

## 5.11 Conclusion

The `complex` data type represents numbers with real and imaginary components. It is especially useful in scientific, mathematical and engineering applications.

# 6. Boolean (bool)

## 6.1 What Is a Boolean Data Type?

A Boolean is a built-in Python data type used to represent one of two logical values:

- `True`
- `False`

Boolean values are mainly used in decision-making, logical operations, conditional statements and comparison operations.

For example, when checking whether a person is eligible to vote, a program can evaluate an age condition and produce either `True` or `False`.

## 6.2 Boolean Values in Python

Python has exactly two Boolean constants:

```python
True
False
```

Important: The first letter must be uppercase.

Correct:

```python
is_active = True
is_completed = False
```

Incorrect:

```python
is_active = true
is_completed = false
```

Python treats `true` and `false` as ordinary names rather than Boolean constants.

## 6.3 Example of Boolean Variables

```python
is_student = True
is_employee = False

print(is_student)
print(is_employee)
```

Output:

```text
True
False
```

## 6.4 Boolean Values from Comparison Operations

Comparison operators evaluate conditions and produce Boolean results.

Example:

```python
a = 10
b = 20

print(a > b)
print(a < b)
print(a == b)
print(a != b)
```

Output:

```text
False
True
False
True
```

Explanation:

- `a > b` checks whether a is greater than b.
- `a < b` checks whether a is smaller than b.
- `a == b` checks whether both values are equal.
- `a != b` checks whether both values are different.

## 6.5 Boolean Operators

Python provides three main logical operators.

### A. AND Operator

The `and` operator returns `True` only when both conditions are true.

```python
print(True and True)
print(True and False)
print(False and True)
print(False and False)
```

Output:

```text
True
False
False
False
```

### B. OR Operator

The `or` operator returns `True` when at least one condition is true.

```python
print(True or True)
print(True or False)
print(False or True)
print(False or False)
```

Output:

```text
True
True
True
False
```

### C. NOT Operator

The `not` operator reverses a Boolean value.

```python
print(not True)
print(not False)
```

Output:

```text
False
True
```

## 6.6 Truth Values

Python allows different objects to be evaluated in a Boolean context.

Some common false-like values include:

- `False`
- `None`
- `0`
- `0.0`
- `0j`
- `""`
- `[]`
- `()`
- `{}`
- `set()`

Most other objects are considered true-like.

Example:

```python
print(bool(0))
print(bool(10))
print(bool(""))
print(bool("Python"))
print(bool([]))
print(bool([1, 2, 3]))
```

Output:

```text
False
True
False
True
False
True
```

## 6.7 Boolean in Conditional Statements

Boolean values are commonly used with `if` and `else`.

Example:

```python
age = 20

if age >= 18:
    print("Eligible")
else:
    print("Not eligible")
```

Output:

```text
Eligible
```

The comparison produces `True`, so the first block executes.

## 6.8 Boolean Conversion Using bool()

The `bool()` function converts an object into a Boolean value.

Example:

```python
a = bool(1)
b = bool(0)

print(a)
print(b)
```

Output:

```text
True
False
```

## 6.9 Finding the Type of Boolean

```python
result = True

print(type(result))
```

Output:

```text
<class 'bool'>
```

## 6.10 Real-Life Applications of Boolean

Boolean values are used in:

- Login authentication.
- Student eligibility checks.
- Payment status.
- User account activation.
- Form validation.
- Permission checks.
- Decision-making systems.

## 6.11 Conclusion

The Boolean data type represents logical values using `True` and `False`. It is essential for conditions, comparisons and decision-making in Python programs.

# 7. String (str)

## 7.1 What Is a String?

A string is a built-in Python data type used to represent text.

A string is an ordered sequence of Unicode characters enclosed within quotation marks.

Strings can contain:

- Letters.
- Numbers represented as text.
- Spaces.
- Special characters.
- Symbols.
- Unicode characters.

For example, a person's name, address, email address and a message can all be represented using strings.

Python uses the `str` data type to represent strings.

## 7.2 Creating Strings

Python supports three common ways of writing string literals.

### A. Single Quotes

```python
name = 'Rahul'

print(name)
```

Output:

```text
Rahul
```

### B. Double Quotes

```python
name = "Rahul"

print(name)
```

Output:

```text
Rahul
```

### C. Triple Quotes

Triple quotes can be used for strings that span multiple lines.

```python
message = """Welcome to Python.
Python is easy to learn.
Python is powerful."""

print(message)
```

Output:

```text
Welcome to Python.
Python is easy to learn.
Python is powerful.
```

## 7.3 Characteristics of Strings

1. **Ordered:** Characters have a defined sequence and can be accessed using indexes.

2. **Immutable:** Once a string is created, its individual characters cannot be changed.

3. **Unicode Support:** Strings can contain characters from many writing systems.

4. **Indexing:** Individual characters can be accessed using their positions.

5. **Slicing:** A portion of a string can be extracted using slicing.

6. **Concatenation:** Two or more strings can be combined using the `+` operator.

7. **Repetition:** A string can be repeated using the `*` operator.

## 7.4 String Indexing

Indexing means accessing individual characters using their positions.

Python uses zero-based indexing, meaning the first character has index `0`.

Example:

```python
name = "PYTHON"

print(name[0])
print(name[1])
print(name[2])
print(name[5])
```

Output:

```text
P
Y
T
N
```

### Negative Indexing

Python also supports negative indexing.

Negative indexing starts from the end of the string.

```python
name = "PYTHON"

print(name[-1])
print(name[-2])
print(name[-3])
```

Output:

```text
N
O
H
```

## 7.5 String Slicing

Slicing is used to extract a portion of a string.

Syntax:

```python
string[start:stop:step]
```

Where:

- `start` is the starting index.
- `stop` is the ending boundary, which is excluded.
- `step` specifies the increment between selected positions.

Example:

```python
name = "PYTHON"

print(name[0:3])
print(name[2:5])
print(name[:4])
print(name[2:])
```

Output:

```text
PYT
THO
PYTH
THON
```

## 7.6 String Concatenation

Concatenation means joining two or more strings.

The `+` operator is used to concatenate strings.

Example:

```python
first_name = "Rahul"
last_name = "Kumar"

full_name = first_name + " " + last_name

print(full_name)
```

Output:

```text
Rahul Kumar
```

## 7.7 String Repetition

The `*` operator repeats a string a specified number of times.

Example:

```python
word = "Python "

print(word * 3)
```

Output:

```text
Python Python Python
```

## 7.8 Important String Methods

Python provides many built-in string methods.

| Method | Purpose |
|---|---|
| `upper()` | Converts letters to uppercase |
| `lower()` | Converts letters to lowercase |
| `title()` | Converts words to title case |
| `capitalize()` | Capitalizes the first character |
| `strip()` | Removes leading and trailing whitespace |
| `replace()` | Replaces specified text |
| `split()` | Splits a string into a list |
| `join()` | Joins iterable strings |
| `find()` | Finds the first occurrence of a substring |
| `count()` | Counts occurrences of a substring |
| `startswith()` | Checks the beginning of a string |
| `endswith()` | Checks the ending of a string |
| `isalpha()` | Checks whether all characters are alphabetic |
| `isdigit()` | Checks whether all characters are digits |
| `isalnum()` | Checks whether all characters are alphanumeric |

Example:

```python
text = "python programming"

print(text.upper())
print(text.title())
print(text.replace("python", "Java"))
```

Output:

```text
PYTHON PROGRAMMING
Python Programming
Java programming
```

## 7.9 Strings Are Immutable

Strings cannot be modified directly after creation.

Incorrect:

```python
name = "Python"

name[0] = "J"
```

This generates a `TypeError`.

Correct:

```python
name = "Python"

name = "J" + name[1:]

print(name)
```

Output:

```text
Jython
```

A new string is created and assigned to the variable.

## 7.10 Finding String Length

The `len()` function returns the number of characters in a string.

Example:

```python
name = "Python"

print(len(name))
```

Output:

```text
6
```

## 7.11 Escape Characters

Escape characters are special sequences used inside strings to represent certain characters.

Examples:

| Escape Sequence | Meaning |
|---|---|
| `\n` | New line |
| `\t` | Horizontal tab |
| `\\` | Backslash |
| `\'` | Single quote |
| `\"` | Double quote |

Example:

```python
print("Hello\nWorld")
print("Python\tProgramming")
```

Output:

```text
Hello
World
Python  Programming
```

## 7.12 Real-Life Applications of Strings

Strings are used for:

- Names.
- Addresses.
- Email addresses.
- Password text handling.
- Messages.
- Website content.
- Product descriptions.
- File names.
- Text processing.

## 7.13 Conclusion

The `str` data type represents text in Python. Strings support indexing, slicing, concatenation, repetition and many useful methods. Since strings are immutable, modifications create new string objects rather than changing existing characters.

# 8. List

## 8.1 What Is a List?

A list is a built-in Python collection data type used to store multiple items in a single variable.

A list is an ordered and mutable collection that can contain elements of different data types.

Lists are useful when a program needs to store and manage multiple values together, such as student marks, product names, employee records or shopping items.

Python represents lists using square brackets `[]`.

## 8.2 Creating a List

Syntax:

```python
variable_name = [item1, item2, item3]
```

Example:

```python
numbers = [10, 20, 30, 40]

print(numbers)
```

Output:

```text
[10, 20, 30, 40]
```

## 8.3 Characteristics of Lists

1. **Ordered:** List elements maintain their sequence.

# 10. Set

## 10.1 What Is a Set?

A set is a built-in Python collection data type used to store unique elements.

A set is an unordered collection, meaning it does not provide a guaranteed positional order for its elements.

Sets automatically eliminate duplicate values.

They are particularly useful for removing duplicates, performing membership checks and carrying out mathematical set operations such as union, intersection and difference.

Python represents sets using curly braces `{}` or the `set()` constructor.

## 10.2 Creating a Set

Syntax:

```python
variable_name = {item1, item2, item3}
```

Example:

```python
numbers = {10, 20, 30, 40}

print(numbers)
```

The elements may be displayed in an order different from the order in which they were entered.

## 10.3 Characteristics of Sets

1. **Unique Elements:** A set does not retain duplicate elements.

2. **Unordered:** Sets do not guarantee a particular element order.

3. **Mutable:** Elements can be added or removed.

4. **No Indexing:** Elements cannot be accessed using positional indexes.

5. **Hashable Elements:** Set elements must be hashable.

6. **Mixed Data Types:** A set can contain different hashable data types.

7. **Mathematical Operations:** Sets support union, intersection, difference and symmetric difference.

## 10.4 Creating Different Sets

### A. Set of Integers

```python
numbers = {10, 20, 30, 40}

print(numbers)
```

### B. Set of Strings

```python
names = {"Rahul", "Ravi", "Rohan"}

print(names)
```

### C. Set with Duplicate Values

```python
numbers = {10, 20, 10, 30, 20}

print(numbers)
```

Output contains only unique elements:

```text
{10, 20, 30}
```

The displayed order is not guaranteed.

Removes the specified element.

```python
numbers = {10, 20, 30}

numbers.remove(20)

print(numbers)
```

Removes the specified element if it exists.

Unlike `remove()`, it does not raise an error when the element is absent.

```python
numbers = {10, 20, 30}

numbers.discard(50)

print(numbers)
```

## 10.7 Mathematical Set Operations

Set operations are one of the most important features of Python sets.

Consider two sets:

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
```

### A. Union

Union combines all unique elements from both sets.

Symbol: `A ∪ B`

Python operator: `|`

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A | B)
```

Output:

```text
{1, 2, 3, 4, 5, 6}
```

### B. Intersection

Intersection returns only the common elements.

Symbol: `A ∩ B`

Python operator: `&`

```python
print(A & B)
```

Output:

```text
{3, 4}
```

### C. Difference

Difference returns the elements present in the first set but not in the second.

Symbol: `A - B`

Python operator: `-`

```python
print(A - B)
```

Output:

```text
{1, 2}
```

### D. Symmetric Difference

Symmetric difference returns elements that are present in either set but not in both.

Python operator: `^`

```python
print(A ^ B)
```

Output:

```text
{1, 2, 5, 6}
```

# 12. NoneType

## 12.1 What Is NoneType?

`NoneType` is a built-in Python data type that represents the absence of a value.

The only value of this type is `None`.

`None` is used when a variable or function does not currently have a meaningful value, or when a function intentionally returns no result.

It is different from zero, an empty string, an empty list or the Boolean value `False`.

## 12.2 What Is None?

`None` is a special constant in Python.

Example:

```python
result = None

print(result)
print(type(result))
```

Output:

```text
None
<class 'NoneType'>
```

## 12.3 Characteristics of NoneType

1. `None` represents the absence of a value.

2. It is a singleton object, meaning there is only one `None` object in a Python process.

3. It is immutable.

4. It is commonly used as a default value for function parameters.

5. Functions that do not explicitly return a value return `None`.

6. It is commonly used to indicate missing or unavailable information.

## 12.4 Example: Variable Assigned None

```python
name = None

print(name)
```

Output:

```text
None
```

This indicates that the variable does not currently refer to a meaningful name value.

## 12.5 Example: Function Without Return

```python
def greet():
    print("Hello")

result = greet()

print(result)
```

Output:

```text
Hello
None
```

The function displays a message but does not explicitly return a value.

Therefore, its return value is `None`.

## 12.6 Checking for None

Python recommends using the `is` operator to check whether a value is `None`.

Example:

```python
result = None

if result is None:
    print("No value available")
else:
    print("Value is available")
```

Output:

```text
No value available
```

## 12.7 None vs Zero vs False vs Empty String

These values are not interchangeable.

| Value | Type | Meaning |
|---|---|---|
| `None` | NoneType | Absence of a value |
| `0` | int | Integer zero |
| `0.0` | float | Floating-point zero |
| `False` | bool | Logical false |
| `""` | str | Empty string |
| `[]` | list | Empty list |
| `{}` | dict | Empty dictionary |

All these values are false-like in Boolean contexts, but they represent different things.

## 12.8 Example

```python
a = None
b = 0
c = False
d = ""

print(a is None)
print(bool(a))
print(bool(b))
print(bool(c))
print(bool(d))
```

Output:

```text
True
False
False
False
False
```

## 12.9 Real-Life Applications of None

`None` is used for:

- Missing database values.
- Optional function parameters.
- Unavailable search results.
- Functions without an explicit return value.
- Uninitialized application data.
- Representing the absence of a result.

## 12.10 Conclusion

`NoneType` represents the absence of a value in Python. Its only value is `None`, which is commonly used for missing information and functions that do not explicitly return a result.

# 13. Mutable Data Types

## 13.1 What Are Mutable Data Types?

Mutable data types are data types whose objects can be modified after they have been created.

This means that their contents can be changed without necessarily creating an entirely new object.

Python provides several mutable built-in data types, including:

- `list`
- `dict`
- `set`
- `bytearray`

Mutable objects are useful when programs need to add, remove or update information during execution.

## 13.2 Characteristics of Mutable Data Types

1. **Modifiable:** Their contents can be changed after creation.

2. **Dynamic:** Their size or contents can change during program execution.

3. **In-Place Operations:** Many operations modify the existing object instead of creating a new object.

4. **Identity:** Modifying an object generally preserves its identity.

5. **Flexible:** They are useful for managing collections that change over time.

## 13.3 Examples of Mutable Data Types

### A. List

Lists allow elements to be added, removed and modified.

```python
numbers = [10, 20, 30]

numbers[1] = 100

print(numbers)
```

Output:

```text
[10, 100, 30]
```

### B. Dictionary

Dictionaries allow key-value pairs to be added, modified and removed.

```python
student = {
    "name": "Rahul",
    "age": 20
}

student["age"] = 21

print(student)
```

Output:

```text
{'name': 'Rahul', 'age': 21}
```

### C. Set

Sets allow unique elements to be added and removed.

```python
numbers = {10, 20, 30}

numbers.add(40)

print(numbers)
```

### D. Bytearray

A bytearray is a mutable sequence of integers representing bytes, with values from 0 to 255.

Example:

```python
data = bytearray([65, 66, 67])

data[0] = 90

print(data)
```

Output:

```text
bytearray(b'ZBC')
```

## 13.4 Mutable Object Identity

When a mutable object is modified, its identity generally remains the same.

Example:

```python
numbers = [10, 20, 30]

print(id(numbers))

numbers.append(40)

print(id(numbers))

print(numbers)
```

The two identity values will normally be the same because the existing list object was modified.

The exact identity number is not fixed.

## 13.5 Mutable Objects and References

When one variable is assigned to another, both variables may refer to the same mutable object.

Example:

```python
a = [10, 20, 30]

b = a

b.append(40)

print(a)
print(b)
```

Output:

```text
[10, 20, 30, 40]
[10, 20, 30, 40]
```

Explanation:

- `a` refers to a list object.
- `b = a` makes `b` refer to the same list.
- Modifying the list using `b` is visible through `a` as well.

## 13.6 Creating an Independent Copy

If we want a separate list, we can create a copy.

Example:

```python
a = [10, 20, 30]

b = a.copy()

b.append(40)

print(a)
print(b)
```

Output:

```text
[10, 20, 30]
[10, 20, 30, 40]
```

This creates a shallow copy. For nested mutable objects, changes to inner objects may still be shared.

## 13.7 Advantages of Mutable Data Types

- Easy to update information.
- Useful for dynamic collections.
- Support in-place modification.
- Suitable for changing application data.
- Reduce the need to create replacement collections for every change.

## 13.8 Disadvantages of Mutable Data Types

- Multiple references may modify the same object.
- Unexpected changes can occur through shared references.
- Mutable objects require careful handling in complex programs.
- Mutable objects are generally not hashable and therefore cannot normally be used as dictionary keys or set elements.

## 13.9 Conclusion

Mutable data types allow objects to be modified after creation. Lists, dictionaries, sets and bytearrays are important examples of mutable Python data types.

# 14. Immutable Data Types

## 14.1 What Are Immutable Data Types?

Immutable data types are data types whose objects cannot be modified after they have been created.

Once an immutable object is created, its contents remain unchanged.

If we perform an operation that appears to modify an immutable object, Python creates or returns another object instead of changing the original object's contents.

Python provides several immutable built-in data types, including:

- `int`
- `float`
- `complex`
- `bool`
- `str`
- `tuple`
- `frozenset`
- `bytes`
- `range`

## 14.2 Characteristics of Immutable Data Types

1. **Cannot Be Modified:** The object's contents cannot be changed after creation.

2. **Stable Values:** The object's value remains unchanged.

3. **New Objects:** Operations that produce changed values return new objects where appropriate.

4. **Safe Sharing:** Immutable objects can often be shared without worrying about their contents being changed.

5. **Hashability:** Many immutable types are hashable, but not every immutable object is necessarily hashable. For example, a tuple containing an unhashable list is itself unhashable.

## 14.3 Examples of Immutable Data Types

### A. Integer

```python
a = 10

a = 20

print(a)
```

Output:

```text
20
```

The integer object `10` was not modified. The variable was reassigned to another integer object.

### B. Float

```python
price = 99.50

price = 100.50

print(price)
```

### C. String

```python
name = "Python"

new_name = name.upper()

print(name)
print(new_name)
```

Output:

```text
Python
PYTHON
```

The original string remains unchanged.

### D. Tuple

```python
numbers = (10, 20, 30)

print(numbers)
```

Individual tuple elements cannot be reassigned.

### E. Frozenset

```python
numbers = frozenset([10, 20, 30])

print(numbers)
```

The elements cannot be added or removed.

## 14.4 Example: Attempting to Modify an Immutable String

Incorrect:

```python
name = "Python"

name[0] = "J"
```

Output:

```text
TypeError: 'str' object does not support item assignment
```

Correct:

```python
name = "Python"

name = "J" + name[1:]

print(name)
```

Output:

```text
Jython
```

The new string is assigned to the variable.

## 14.5 Advantages of Immutable Data Types

1. They prevent accidental modification of existing values.
2. They are useful for representing fixed information.
3. They can safely be shared between different parts of a program.
4. Hashable immutable objects can be used as dictionary keys and set elements.
5. They can make program behavior easier to reason about.

## 14.6 Mutable vs Immutable Data Types

| Feature | Mutable | Immutable |
|---|---|---|
| Modification | Can be modified | Cannot be modified |
| Contents | Can change | Remain unchanged |
| Examples | List, dictionary, set | Integer, float, string, tuple |
| In-place updates | Supported | Not supported for object contents |
| Shared References | Changes can be visible through other references | Contents cannot be changed through references |
| Hashability | Usually unhashable | Depends on the type and its contents |

## 14.7 Important Difference Between Reassignment and Modification

These two operations are not the same.

Reassignment:

```python
x = 10

x = 20
```

Modification of a mutable object:

```python
numbers = [10, 20]

numbers.append(30)
```

In the first example, the variable `x` is reassigned.

In the second example, the existing list object is modified.

## 14.8 Conclusion

Immutable data types cannot be modified after creation. They are useful for fixed values and predictable data handling. Understanding immutability is important for learning Python references, memory behavior and object identity.

# 16. Implicit Type Conversion

## 16.1 What Is Implicit Type Conversion?

Implicit type conversion is the process in which Python automatically converts a value from one compatible data type to another during an operation.

The programmer does not need to explicitly call a conversion function.

Python performs certain automatic conversions to make operations between compatible numeric types possible.

This is also commonly called automatic type conversion.

## 16.2 Why Does Implicit Conversion Happen?

Implicit conversion happens when Python needs to combine compatible types during an operation.

For example, when an integer and a floating-point number are added together, Python produces a floating-point result.

This allows mathematical operations to work naturally without requiring programmers to manually convert every operand.

## 16.3 Example: Integer and Float

```python
a = 10
b = 5.5

result = a + b

print(result)
print(type(result))
```

Output:

```text
15.5
<class 'float'>
```

Explanation:

- `a` is an integer.
- `b` is a floating-point number.
- Python performs the operation using a compatible floating-point representation.
- The result is a float.

## 16.4 Example: Integer and Complex

```python
a = 10
b = 3 + 4j

result = a + b

print(result)
print(type(result))
```

Output:

```text
(13+4j)
<class 'complex'>
```

Python converts the integer into a compatible complex representation for the operation.

## 16.5 Numeric Conversion Hierarchy

For many mixed numeric operations, Python uses a general progression:

```text
       int
        |
        v
      float
        |
        v
     complex
```

This diagram represents a common direction of numeric promotion.

For example:

- `int + float` produces a float.
- `int + complex` produces a complex number.
- `float + complex` produces a complex number.

## 16.6 Example: Integer Division

```python
a = 10
b = 2

result = a / b

print(result)
print(type(result))
```

Output:

```text
5.0
<class 'float'>
```

Even though both operands are integers, true division (`/`) produces a float.

## 16.7 Limitations of Implicit Conversion

Python does not automatically convert every combination of data types.

Example:

```python
a = 10
b = "20"

print(a + b)
```

This generates a `TypeError` because Python does not automatically convert a string into an integer for addition.

Correct:

```python
a = 10
b = "20"

result = a + int(b)

print(result)
```

Output:

```text
30
```

## 16.8 Advantages of Implicit Conversion

1. Reduces the need for manual conversions.
2. Makes compatible numerical calculations easier.
3. Reduces repetitive code.
4. Helps combine compatible numeric values.

## 16.9 Conclusion

Implicit type conversion occurs automatically when Python performs compatible operations involving different numeric types. It makes mathematical calculations more convenient.

# 19. isinstance() Function

## 19.1 What Is the isinstance() Function?

The `isinstance()` function is a built-in Python function used to check whether an object belongs to a specified class or one of its subclasses.

It returns a Boolean result:

- `True` if the object is an instance of the specified type or a subclass.
- `False` otherwise.

It is especially useful for validating data before performing operations.

## 19.2 Syntax

```python
isinstance(object, classinfo)
```

Where:

- `object` is the value being checked.
- `classinfo` is a type, class or tuple of types and classes.

## 19.3 Example: Checking Integer

```python
age = 20

print(isinstance(age, int))
```

Output:

```text
True
```

The value `20` is an integer.

## 19.4 Example: Checking String

```python
name = "Rahul"

print(isinstance(name, str))
```

Output:

```text
True
```

## 19.5 Example: Checking Multiple Data Types

The second argument can be a tuple containing multiple types.

Example:

```python
value = 10.5

print(isinstance(value, (int, float)))
```

Output:

```text
True
```

The object is a float, and `float` is one of the accepted types.

## 19.6 Example: Checking a List

```python
numbers = [10, 20, 30]

if isinstance(numbers, list):
    print("The object is a list")
else:
    print("The object is not a list")
```

Output:

```text
The object is a list
```

## 19.7 isinstance() and Inheritance

One important feature of `isinstance()` is that it supports inheritance.

This means it returns `True` when an object belongs to a subclass of the specified class.

Example:

```python
class Animal:
    pass

class Dog(Animal):
    pass

dog = Dog()

print(isinstance(dog, Dog))
print(isinstance(dog, Animal))
```

Output:

```text
True
True
```

Explanation:

- `dog` is an object of the `Dog` class.
- `Dog` inherits from `Animal`.
- Therefore, the object is also considered an instance of `Animal`.

## 19.8 Difference Between type() and isinstance()

| Feature | type() | isinstance() |
|---|---|---|
| Purpose | Identifies exact object type | Checks type membership |
| Return Value | Type object | Boolean |
| Inheritance Support | Exact-type comparison does not consider subclasses | Supports subclasses |
| Multiple Types | Not directly as a single check | Accepts a tuple of types |
| Common Use | Inspecting object types | Type validation |

Example:

```python
class Animal:
    pass

class Dog(Animal):
    pass

dog = Dog()

print(type(dog) == Animal)
print(isinstance(dog, Animal))
```

Output:

```text
False
True
```

## 19.9 Real-Life Applications

`isinstance()` is used for:

- Validating user input.
- Checking function arguments.
- Processing different data types.
- Writing reusable functions.
- Working with inheritance.
- Avoiding incompatible operations.

## 19.10 Conclusion

The `isinstance()` function checks whether an object belongs to a particular type or class hierarchy. It is a useful tool for validation and writing flexible Python programs.

# 20. Object Identity and id()

## 20.1 What Is Object Identity?

In Python, every object has an identity that distinguishes it from other objects during its lifetime.

Object identity represents the identity of an object, not simply the value it contains.

Two variables may contain equal values but refer to different objects.

Alternatively, two variables may refer to the same object.

Understanding object identity is important for learning Python references, memory behavior, mutable objects and immutable objects.

## 20.2 What Is the id() Function?

The `id()` function is a built-in Python function that returns an integer representing the identity of an object.

The identity remains constant throughout that object's lifetime.

Syntax:

```python
id(object)
```

The exact number returned by `id()` is implementation-dependent and should not be treated as a permanent memory address.

In CPython, an object's identity is commonly associated with its memory address, but Python does not guarantee that interpretation for all implementations.

## 20.3 Example of id()

```python
a = 10

print(id(a))
```

Output:

```text
The output is an implementation-dependent integer.
```

The actual identity number may differ between executions.

## 20.4 Two Variables Referring to the Same Object

Example:

```python
a = [10, 20, 30]

b = a

print(id(a))
print(id(b))
```

The two identity values will be the same because both variables refer to the same list object.

Diagram:

```text
       a --------+
                 |
                 v
          [10, 20, 30]
                 ^
                 |
       b --------+

       Same Object
```

## 20.5 Two Variables Referring to Different Objects

Example:

```python
a = [10, 20, 30]
b = [10, 20, 30]

print(id(a))
print(id(b))
```

The two identity values will normally be different because two separate list objects were created.

Diagram:

```text
       a
       |
       v
   [10, 20, 30]


       b
       |
       v
   [10, 20, 30]

    Different Objects
```

Both lists contain equal values, but they are separate objects.

## 20.6 Difference Between == and is

Python provides two different operators for comparing objects.

### A. Equality Operator (==)

The `==` operator checks whether two objects have equal values.

Example:

```python
a = [10, 20, 30]
b = [10, 20, 30]

print(a == b)
```

Output:

```text
True
```

Both lists contain the same elements in the same order.

### B. Identity Operator (is)

The `is` operator checks whether two references point to the same object.

Example:

```python
a = [10, 20, 30]
b = [10, 20, 30]

print(a is b)
```

Output:

```text
False
```

They are different list objects.

### C. Both Operators Together

```python
a = [10, 20, 30]
b = a

print(a == b)
print(a is b)
```

Output:

```text
True
True
```

Both variables refer to the same object, so their values are equal and their identities are the same.

## 20.7 Equality vs Identity Comparison

| Feature | == | is |
|---|---|---|
| Meaning | Checks value equality | Checks object identity |
| Compares | Values | Whether references identify the same object |
| Returns | True or False | True or False |
| Example | `a == b` | `a is b` |
| Common Use | Comparing values | Checking identity |

## 20.8 Object Identity with Mutable Objects

Mutable objects can be modified while keeping their identity.

Example:

```python
numbers = [10, 20, 30]

before = id(numbers)

numbers.append(40)

after = id(numbers)

print(before == after)
print(numbers)
```

Output:

```text
True
[10, 20, 30, 40]
```

The list was modified in place, so its identity remained unchanged.

## 20.9 Object Identity with Immutable Objects

Immutable objects cannot be modified.

When a variable is reassigned, it may begin referring to a different object.

Example:

```python
name = "Python"

before = id(name)

name = name + " Programming"

after = id(name)

print(before == after)
print(name)
```

Output:

```text
False
Python Programming
```

In this example, concatenation creates a new string object, and the variable is reassigned to it.

Note: Python may reuse objects or identities in some situations. Never depend on identity numbers being different merely because two values are written separately.

## 20.10 Checking for None Using is

The `is` operator is commonly used to check whether a value is `None`.

Example:

```python
result = None

if result is None:
    print("No result available")
```

Output:

```text
No result available
```

This is the recommended way to compare a value with `None`.

## 20.11 Real-Life Importance of Object Identity

Object identity is useful for:

- Understanding references.
- Debugging shared mutable objects.
- Avoiding unexpected changes.
- Understanding shallow and deep copies.
- Working with object-oriented programming.
- Checking whether two references point to the same object.

## 20.12 Conclusion

Object identity represents the identity of a Python object. The `id()` function returns an integer identifying the object during its lifetime, while the `is` operator checks whether two references point to the same object.

Understanding object identity helps programmers understand how Python handles variables, references and objects.

| Main Use | Changing collections | Fixed collections | Unique values | Structured records |

```python
name = "Rahul"
age = 20
height = 5.8
is_student = True
numbers = [10, 20, 30]
coordinates = (10, 20)
unique_numbers = {10, 20, 30}
student = {"name": "Rahul", "age": 20}
result = None

print("Name:", name, type(name))
print("Age:", age, type(age))
print("Height:", height, type(height))
print("Student:", is_student, type(is_student))
print("Numbers:", numbers, type(numbers))
print("Coordinates:", coordinates, type(coordinates))
print("Set:", unique_numbers, type(unique_numbers))
print("Dictionary:", student, type(student))
print("Result:", result, type(result))
```
