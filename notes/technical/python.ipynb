## 5. Operators

### Arithmetic Operators

```python
a = 10
b = 3

print(a + b)  # Addition
print(a - b)  # Subtraction
print(a * b)  # Multiplication
print(a / b)  # Division
print(a // b) # Floor division
print(a % b)  # Modulus
print(a ** b) # Power
```

### Comparison Operators

```python
print(10 == 10)
print(10 != 5)
print(10 > 5)
print(10 < 5)
print(10 >= 10)
print(5 <= 10)
```

### Logical Operators

```python
x = 10

print(x > 5 and x < 20)
print(x > 15 or x < 20)
print(not(x > 5))
```

## 6. Conditional Statements

```python
age = 18

if age >= 18:
    print("Adult")
elif age >= 13:
    print("Teenager")
else:
    print("Child")
```

## 7. Loops

### For Loop

```python
for i in range(1, 6):
    print(i)
```

### While Loop

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

### Break Statement

```python
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```

### Continue Statement

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

## 8. Strings

```python
text = "Python Programming"

print(text.upper())
print(text.lower())
print(text.capitalize())
print(text.replace("Python", "Java"))
print(len(text))
print(text[0])
print(text[0:6])
```

## 9. Lists

Lists store multiple values.

```python
fruits = ["Apple", "Banana", "Mango"]

print(fruits[0])

fruits.append("Orange")
fruits.remove("Banana")

print(fruits)
```

## 10. Tuples

Tuples are ordered and immutable.

```python
numbers = (10, 20, 30)

print(numbers[0])
print(len(numbers))
```

## 11. Dictionaries

Dictionaries store key-value pairs.

```python
student = {
    "name": "Rahul",
    "age": 16,
    "grade": "10th"
}

print(student["name"])

student["age"] = 17
student["city"] = "Hyderabad"

print(student)
```

## 12. Sets

Sets store unique values.

```python
numbers = {1, 2, 3, 3, 4}

numbers.add(5)
numbers.remove(2)

print(numbers)
```

## 13. Functions

Functions are reusable blocks of code.

```python
def greet():
    print("Hello!")

greet()
```

### Functions with Parameters

```python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

## 14. Type Conversion

```python
x = "100"

print(int(x))
print(float(x))
print(str(100))
print(bool(1))
```

## 15. Exception Handling

```python
try:
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))

    print(a / b)

except ZeroDivisionError:
    print("Cannot divide by zero!")

except ValueError:
    print("Enter valid numbers!")

finally:
    print("Program completed")
```

## 16. Classes and Objects

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(self.name, self.age)

s1 = Student("Rahul", 16)
s1.display()
```

## 17. Modules

```python
import math

print(math.sqrt(25))
print(math.pi)
print(math.factorial(5))
```

## 18. File Handling

```python
# Writing to a file
with open("sample.txt", "w") as file:
    file.write("Hello Python!")

# Reading a file
with open("sample.txt", "r") as file:
    print(file.read())
```
