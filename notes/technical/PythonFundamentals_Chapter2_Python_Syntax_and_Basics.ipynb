# Chapter 2: Python Syntax and Basics

## 1. Python Syntax

### 1.1 What is Python Syntax?

Python syntax refers to the set of rules and guidelines that determine how Python programs must be written. Just like every human language has grammar rules, Python has its own rules for writing instructions that the Python interpreter can understand and execute.

Python syntax defines how to write statements, expressions, variables, functions, conditions, loops and other programming elements. If the syntax rules are not followed correctly, Python may generate a syntax error and stop the program from executing.

Python is known for its simple and readable syntax. It uses indentation to organize code blocks instead of relying on curly braces.

### 1.2 Important Rules of Python Syntax

1. **Case Sensitivity:** Python is a case-sensitive programming language. This means uppercase and lowercase letters are treated differently. For example, `name`, `Name` and `NAME` are three different identifiers.

2. **Indentation:** Python uses spaces or tabs at the beginning of a line to define blocks of code. Indentation is compulsory for statements inside functions, conditions and loops.

3. **New Lines:** Python generally treats a new line as the end of a statement. Multiple statements can also be written on one line using semicolons, although this is usually discouraged for readability.

4. **Colons:** A colon (`:`) is used at the end of statements that introduce an indented block, such as `if`, `for`, `while`, `def` and `class`.

5. **Comments:** Comments are written using the `#` symbol. They explain code and are ignored during normal execution.

6. **Variables:** Variables are created by assigning values to names. Python does not require an explicit variable declaration.

### 1.3 Example of Python Syntax

```python
name = "Rahul"
age = 20

if age >= 18:
    print("Name:", name)
    print("Eligible")
```

**Output:**

```text
Name: Rahul
Eligible
```

**Explanation:**

- `name = "Rahul"` assigns a string value to a variable.
- `age = 20` assigns an integer value to another variable.
- `if age >= 18:` checks a condition.
- The colon indicates the beginning of a code block.
- The indented `print()` statements execute when the condition is true.

### 1.4 Incorrect Python Syntax

```python
if age >= 18
    print("Eligible")
```

**Error:**

```text
SyntaxError: expected ':'
```

The colon is missing after the condition.

### 1.5 Conclusion

Python syntax provides the rules for writing valid Python programs. Understanding syntax is important because every Python program must follow these rules to execute correctly.

These words are used to define the structure and behavior of Python programs. Since keywords have special meanings, they cannot be used as ordinary identifiers or variable names.

For example, `if`, `else`, `for`, `while`, `class`, `def` and `return` are Python keywords.

Keywords help Python understand whether a programmer wants to perform a condition check, create a loop, define a function, create a class or perform another language operation.

Python provides a module named `keyword` that contains information about reserved keywords.

Example:

```python
import keyword

print(keyword.kwlist)
```

To check whether a word is a keyword:

```python
import keyword

print(keyword.iskeyword("if"))
print(keyword.iskeyword("name"))
```

**Output:**

```text
True
False
```

```python
if_value = 10

print(if_value)
```

Keywords are reserved words that have special meanings in Python. They are essential for defining program structures and controlling program execution.

## 3. Identifiers

### 3.1 What are Identifiers?

Identifiers are the names given to different programming elements in Python.

They are used to identify variables, functions, classes, modules and other objects in a program.

Whenever a programmer creates a variable, function or class, a suitable identifier is used to refer to it.

For example, in `student_name = "Rahul"`, the name `student_name` is an identifier because it identifies the variable that refers to the assigned value.

Identifiers help programmers organize their programs and make their code easier to understand.

### 3.2 Examples of Identifiers

```python
name = "Rahul"
age = 20
total_marks = 450

def calculate_sum():
    return 10 + 20

class Student:
    pass
```

In this example:

- `name` is a variable identifier.
- `age` is a variable identifier.
- `total_marks` is a variable identifier.

Identifiers can be used to name different programming elements.

Used to identify variables.

## 4. Variables

### 4.1 What are Variables?

A variable is a name that refers to an object or value stored in a Python program.

Variables allow programmers to store, access and manipulate data while a program is running.

For example, when we write `age = 20`, the name `age` refers to the integer object `20`.

Variables are important because they allow a program to work with changing information instead of using fixed values everywhere.

### 4.2 Why are Variables Used?

Variables are used to:

1. Store information that a program needs.
2. Reuse values in different parts of a program.
3. Perform mathematical calculations.
4. Store user input.
5. Update information during program execution.
6. Make programs easier to organize and maintain.

### 4.3 Syntax of Variables

```python
variable_name = value
```

Here:

- `variable_name` is the identifier.
- `=` is the assignment operator.
- `value` is the object being assigned to the variable.

### 4.4 Example of Variables

```python
name = "Rahul"
age = 20
marks = 85.5

print(name)
print(age)
print(marks)
```

**Output:**

```text
Rahul
20
85.5
```

**Explanation:**

- `name` refers to a string.
- `age` refers to an integer.
- `marks` refers to a floating-point number.
- The `print()` function displays the values.

### 4.5 Variables Can Store Different Types of Data

Python variables can refer to different types of objects.

```python
name = "Rahul"       # String
age = 20             # Integer
price = 99.50        # Float
is_student = True    # Boolean

print(name)
print(age)
print(price)
print(is_student)
```

### 4.6 Variables Can Be Reassigned

A variable can be assigned a new value during program execution.

Example:

```python
age = 18

print(age)

age = 20

print(age)
```

**Output:**

```text
18
20
```

The variable `age` initially refers to `18`. Later, it refers to `20`.

### 4.7 Python Variables Are Dynamically Typed

Python is a dynamically typed language.

This means that programmers do not need to declare a variable's data type explicitly. Python determines the type of the object at runtime.

A variable can also refer to objects of different types at different times.

Example:

```python
value = 100

print(value)

value = "Python"

print(value)
```

**Output:**

```text
100
Python
```

### 4.8 Variables Refer to Objects

In Python, variables are names that refer to objects rather than traditional fixed memory boxes.

Example:

```python
a = 10
b = a

print(a)
print(b)
```

**Output:**

```text
10
10
```

Both names initially refer to the same integer object.

### 4.9 Checking the Type of a Variable

Python provides the `type()` function to identify the type of an object.

```python
name = "Rahul"
age = 20
price = 99.50

print(type(name))
print(type(age))
print(type(price))
```

**Output:**

```text
<class 'str'>
<class 'int'>
<class 'float'>
```

### 4.10 Conclusion

Variables are fundamental building blocks of Python programming. They allow programmers to store and work with information throughout the execution of a program.

## 6. Comments

### 6.1 What are Comments?

Comments are explanatory notes written inside a Python program to describe the purpose or functionality of the code.

Comments help programmers understand what a particular statement, function or block of code does.

Python ignores ordinary comments during normal program execution.

Comments are mainly used to improve code readability, document program logic and make maintenance easier.

### 6.2 Why are Comments Used?

Comments are useful for several reasons:

1. They explain the purpose of the program.
2. They describe complex logic.
3. They help other developers understand the code.
4. They provide useful information during debugging.
5. They help document important program details.
6. They can temporarily disable individual lines of code during testing.

### 6.3 Example of Comments

```text
30
```

The lines beginning with `#` are comments. Python ignores these comments during normal execution.

Commonly used forms include:

## 7. Single-Line Comments

### 7.1 What are Single-Line Comments?

A single-line comment is a comment written on one line of a Python program.

Python uses the hash symbol (`#`) to begin a comment.

Everything after the `#` symbol on that physical line is treated as comment text, unless the symbol occurs inside a string literal.

Single-line comments are useful for explaining a particular instruction or adding a short note to the program.

### 7.2 Syntax

```python
# This is a single-line comment
```

### 7.3 Example

```python
# Store two numbers
a = 10
b = 20

# Calculate their sum
result = a + b

# Display the result
print(result)
```

**Output:**

```text
30
```

### 7.4 Inline Comments

An inline comment is written on the same line as a Python statement.

Example:

```python
age = 20  # Store the student's age

price = 100  # Store the product price
```

### 7.5 Commenting Out Code

Comments can temporarily disable a line of code.

Example:

```python
print("Hello")

# print("This line will not execute")

print("Welcome")
```

**Output:**

```text
Hello
Welcome
```

### 7.6 Best Practices

- Write comments that explain why something is done.
- Keep comments short and meaningful.
- Avoid unnecessary comments.
- Update comments whenever the related code changes.
- Do not use comments to explain obvious instructions.

### 7.7 Conclusion

Single-line comments help programmers explain individual statements and make Python code easier to understand and maintain.

## 8. Multi-Line Documentation

### 8.1 What is Multi-Line Documentation?

Multi-line documentation is used to provide detailed explanations about Python modules, classes and functions.

Python uses a special form of string called a documentation string, commonly known as a docstring.

A docstring is generally written using triple double quotes (`"""`) or triple single quotes (`'''`).

Unlike ordinary comments, docstrings are stored as part of the documented object's metadata and can be accessed at runtime.

### 8.2 What is a Docstring?

A docstring is a string literal placed as the first statement inside a module, function, class or method.

It describes what that programming element does.

Docstrings are useful for:

1. Explaining the purpose of a function.
2. Describing the behavior of a class.
3. Documenting module functionality.
4. Explaining function parameters.
5. Describing return values.
6. Helping developers understand how to use code.

### 8.3 Syntax

```python
"""
This is a multi-line documentation string.

It can contain several lines
of explanatory information.
"""
```

### 8.4 Example: Function Docstring

```python
def add(a, b):
    """
    This function adds two numbers.

    Parameters:
        a: First number.
        b: Second number.

    Returns:
        The sum of both numbers.
    """
    return a + b


result = add(10, 20)

print(result)
```

**Output:**

```text
30
```

### 8.5 Accessing a Docstring

Python provides the `__doc__` attribute to access an object's documentation string.

Example:

```python
def greet():
    """Displays a welcome message."""
    print("Welcome to Python")


print(greet.__doc__)
```

**Output:**

```text
Displays a welcome message.
```

### 8.6 Difference Between Comments and Docstrings

| Comments | Docstrings |
|---|---|
| Begin with `#` | Usually written using triple quotes |
| Explain code to readers | Document modules, classes and functions |
| Ignored during normal execution | Stored as documentation metadata |
| Cannot normally be accessed as an object attribute | Can be accessed using `__doc__` |
| Can be written almost anywhere | Must appear as the first statement of the documented object |

### 8.7 Important Note

Triple quotes do not automatically make every string a docstring.

A triple-quoted string is recognized as a docstring when it appears in the correct documentation position, such as the first statement in a function or class.

### 8.8 Conclusion

Docstrings provide structured documentation for Python modules, classes and functions. They make programs easier to understand, use and maintain.

## 9. Indentation

### 9.1 What is Indentation?

Indentation refers to the spaces or tabs placed at the beginning of a line of code.

Python uses indentation to identify blocks of code and determine which statements belong together.

Unlike languages such as C, C++ and Java, Python does not use curly braces to define code blocks.

Indentation is an essential part of Python syntax. Incorrect indentation can cause an `IndentationError` or other syntax-related errors.

### 9.2 Why is Indentation Important?

Indentation is important because:

1. It defines code blocks.
2. It improves code readability.
3. It identifies statements belonging to conditions and loops.
4. It identifies statements belonging to functions and classes.
5. It helps organize program structure.

### 9.3 Indentation Rules

- Python commonly uses four spaces for each indentation level.
- Statements inside a block must have consistent indentation.
- Statements belonging to the same block should normally have the same indentation level.
- Nested blocks require additional indentation.
- Mixing tabs and spaces inconsistently should be avoided.

### 9.4 Example of Correct Indentation

```python
age = 20

if age >= 18:
    print("Eligible")
    print("You are an adult")

print("Program completed")
```

**Output:**

```text
Eligible
You are an adult
Program completed
```

**Explanation:**

The two indented print statements belong to the `if` block.

The last print statement is not indented, so it executes outside the `if` block.

### 9.5 Example of Incorrect Indentation

```python
age = 20

if age >= 18:
print("Eligible")
```

**Error:**

```text
IndentationError: expected an indented block
```

The statement inside the `if` block must be indented.

### 9.6 Understanding Nested Indentation

When one block is placed inside another block, additional indentation is required.

Example:

```python
age = 20

if age >= 18:
    print("Adult")

    if age >= 21:
        print("Age is 21 or above")
```

**Output:**

```text
Adult
```

The second condition is nested inside the first condition.

### 9.7 Indentation Diagram

```text
if condition:
    Statement 1
    Statement 2

    if another_condition:
        Statement 3
        Statement 4

Statement 5
```

Explanation:

- `Statement 1` and `Statement 2` belong to the first block.
- `Statement 3` and `Statement 4` belong to the nested block.
- `Statement 5` is outside both blocks.

### 9.8 Conclusion

Indentation is one of the most important features of Python syntax. It defines code blocks and helps maintain a clean and readable program structure.

## 12. Multiple Assignments

### 12.1 What are Multiple Assignments?

Multiple assignment is a feature in Python that allows programmers to assign values to multiple variables in a single statement.

It reduces repetitive code and makes variable assignment more convenient.

Python supports multiple forms of assignment, including chained assignment and unpacking assignment.

### 12.2 Multiple Variable Assignment

We can assign different values to multiple variables in one statement.

Example:

```python
a, b, c = 10, 20, 30

print(a)
print(b)
print(c)
```

**Output:**

```text
10
20
30
```

Explanation:

- `a` receives `10`.
- `b` receives `20`.
- `c` receives `30`.

### 12.3 Assigning the Same Value to Multiple Variables

Python allows the same value to be assigned to multiple variables using chained assignment.

Example:

```python
a = b = c = 100

print(a)
print(b)
print(c)
```

**Output:**

```text
100
100
100
```

All three names refer to the same assigned object.

### 12.4 Swapping Two Variables

Python allows programmers to exchange the values of two variables without using a temporary variable.

Example:

```python
a = 10
b = 20

a, b = b, a

print(a)
print(b)
```

**Output:**

```text
20
10
```

### 12.5 Unpacking a Collection

Multiple assignment can also be used to unpack values from a list or tuple.

Example:

```python
numbers = [10, 20, 30]

a, b, c = numbers

print(a)
print(b)
print(c)
```

**Output:**

```text
10
20
30
```

The number of unpacked variables must match the number of elements unless extended unpacking is used.

### 12.6 Extended Unpacking

Python also supports the starred expression during unpacking.

Example:

```python
numbers = [10, 20, 30, 40, 50]

first, *middle, last = numbers

print(first)
print(middle)
print(last)
```

**Output:**

```text
10
[20, 30, 40]
50
```

### 12.7 Advantages of Multiple Assignment

1. Reduces repetitive assignment statements.
2. Makes code shorter.
3. Makes swapping variables easier.
4. Helps unpack lists and tuples.

### 12.8 Conclusion

Multiple assignment allows programmers to assign values to several variables in a single statement. It is a convenient feature that makes Python code shorter and more readable.

## 13. Constants and Naming Conventions

### 13.1 What are Constants?

A constant is a value that is intended to remain unchanged throughout a program.

Constants are useful when a program repeatedly uses fixed values such as mathematical values, configuration settings, tax rates or maximum limits.

Unlike some programming languages, Python does not have a general built-in mechanism that makes an ordinary variable completely immutable simply because it is declared as a constant.

Instead, Python follows naming conventions to indicate that a variable should not be changed.

### 13.2 Naming Convention for Constants

Python recommends writing constant names using uppercase letters.

If the name contains multiple words, underscores are used to separate them.

Examples:

```python
PI = 3.14159
MAX_LIMIT = 100
TAX_RATE = 0.18
COMPANY_NAME = "ABC Technologies"
```

These names indicate that the values are intended to remain unchanged.

### 13.3 Example of Using Constants

```python
PI = 3.14159

radius = 5

area = PI * radius * radius

print("Area:", area)
```

**Output:**

```text
Area: 78.53975
```

Here, `PI` represents a fixed mathematical value.

### 13.4 Can Constants Be Changed in Python?

Yes. Python does not automatically prevent reassignment of uppercase variable names.

Example:

```python
MAX_LIMIT = 100

MAX_LIMIT = 200

print(MAX_LIMIT)
```

**Output:**

```text
200
```

This is allowed, but changing a value intended to be a constant is generally discouraged.

### 13.5 Final Constants Using Final

Python provides the `Final` annotation through the `typing` module to indicate that a name should not be reassigned.

Example:

```python
from typing import Final

PI: Final = 3.14159
```

This annotation is primarily checked by static type checkers. Python does not enforce it automatically at runtime.

### 13.6 Common Python Naming Conventions

Python follows the PEP 8 style guide.

| Programming Element | Recommended Naming Style | Example |
|---|---|---|
| Variable | snake_case | `student_name` |
| Function | snake_case | `calculate_total()` |
| Class | PascalCase | `StudentDetails` |
| Constant | UPPER_CASE | `MAX_LIMIT` |
| Module | lowercase | `calculator` |
| Package | lowercase | `mypackage` |
| Internal variable | Leading underscore | `_internal_value` |

### 13.7 What is PEP 8?

PEP 8 stands for Python Enhancement Proposal 8.

It is the official Python style guide that provides recommendations for writing clean, readable and consistent Python code.

It covers:

- Naming conventions.
- Indentation.
- Line length.
- Whitespace.
- Imports.
- Code formatting.

### 13.8 Conclusion

Constants represent values that are intended to remain unchanged. Python programmers commonly use uppercase naming conventions to identify constants. Following PEP 8 makes programs more consistent and readable.

## 18. Type Conversion

### 18.1 What is Type Conversion?

Type conversion is the process of changing a value from one data type to another.

Python supports type conversion so that programmers can work with different types of data and perform the required operations.

For example, if a user enters a number using `input()`, Python receives it as a string. To perform mathematical calculations, we can convert that string into an integer or floating-point number.

Python supports two major types of conversion:

1. Implicit Type Conversion.
2. Explicit Type Conversion.

### 18.2 Implicit Type Conversion

Implicit type conversion occurs when Python automatically converts a value from one type to another during an operation.

The programmer does not need to perform the conversion manually.

Example:

```python
a = 10
b = 2.5

result = a + b

print(result)
print(type(result))
```

**Output:**

```text
12.5
<class 'float'>
```

**Explanation:**

- `a` is an integer.
- `b` is a floating-point number.
- Python converts the integer to a compatible numeric type for the operation.
- The result is a floating-point number.

### 18.3 Explicit Type Conversion

Explicit type conversion occurs when the programmer manually converts a value from one data type to another.

Python provides built-in conversion functions such as:

- `int()`
- `float()`
- `str()`
- `bool()`
- `list()`
- `tuple()`
- `set()`
- `dict()`

**Output:**

## 19. Type Checking

### 19.1 What is Type Checking?

Type checking is the process of identifying or verifying the data type of a value or object in a program.

Python provides built-in functions such as `type()` and `isinstance()` to check object types.

Type checking helps programmers understand what kind of data is being processed and avoid operations that are not appropriate for a particular type.

For example, before performing a numerical calculation, a programmer may check whether the supplied value is an integer or floating-point number.

### 19.2 Why is Type Checking Important?

Type checking is useful because:

1. It identifies the type of an object.
2. It helps prevent incompatible operations.
3. It supports input validation.
4. It helps debug programs.
5. It helps developers understand how data is stored and processed.
6. It supports writing more reliable programs.

### 19.3 Using the type() Function

The `type()` function returns the type of an object.

Syntax:

```python
type(object)
```

Example:

```python
name = "Rahul"
age = 20
price = 99.50
is_student = True

print(type(name))
print(type(age))
print(type(price))
print(type(is_student))
```

**Output:**

```text
<class 'str'>
<class 'int'>
<class 'float'>
<class 'bool'>
```

### 19.4 Using isinstance()

The `isinstance()` function checks whether an object belongs to a particular class or one of its subclasses.

Syntax:

```python
isinstance(object, classinfo)
```

It returns either `True` or `False`.

Example:

```python
age = 20

print(isinstance(age, int))
print(isinstance(age, str))
```

**Output:**

```text
True
False
```

### 19.5 Checking Multiple Types

We can also check whether an object belongs to one of several types by passing a tuple of types.

Example:

```python
value = 10.5

print(isinstance(value, (int, float)))
```

**Output:**

```text
True
```

### 19.6 Difference Between type() and isinstance()

| type() | isinstance() |
|---|---|
| Returns the exact type of an object | Checks whether an object is an instance of a specified type or subclass |
| Useful for identifying object types | Useful for type validation |
| Returns a type object | Returns True or False |
| Does not consider subclass relationships when comparing exact types | Supports subclass relationships |

Example:

```python
class Animal:
    pass

class Dog(Animal):
    pass

dog = Dog()

print(type(dog) == Animal)
print(isinstance(dog, Animal))
```

**Output:**

```text
False
True
```

### 19.7 Type Checking with User Input

Example:

```python
age = input("Enter your age: ")

print(type(age))
```

**Sample execution:**

```text
Enter your age: 20
<class 'str'>
```

Even though the user entered a number, the input is a string.

### 19.8 Type Checking with Conditions

Example:

```python
value = 100

if isinstance(value, int):
    print("The value is an integer")
else:
    print("The value is not an integer")
```

**Output:**

```text
The value is an integer
```

### 19.9 Conclusion

Type checking helps programmers identify and validate the types of objects in Python. The `type()` and `isinstance()` functions are commonly used for this purpose.

```python
celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9/5) + 32

print("Temperature in Fahrenheit:", fahrenheit)
```

## Program 5: Calculate the Area of a Rectangle

`Area = Length * Breadth`

```python
length = float(input("Enter length: "))
breadth = float(input("Enter breadth: "))

area = length * breadth

print("Area of Rectangle:", area)
```
