### **Q1. What does a Business Analyst do?**

Very important – high probability

**1. Understand the concept**

A Business Analyst (BA) acts as a bridge between business stakeholders and the technical team.

The primary responsibility of a BA is to understand a business problem, identify what the business needs, translate those needs into clear requirements, and help the technical team deliver the right solution.

Think of a simple example:

A company receives hundreds of customer complaints because customers cannot track their orders.

- Business team: Wants to reduce customer complaints.
- Customers: Want to know where their orders are.
- Development team: Needs clear instructions to build the feature.
- Business Analyst: Understands everyone's requirements and translates them into a solution.

The BA ensures that the final solution addresses the actual business problem.

**2. What are the major responsibilities of a Business Analyst?**

There are six important responsibilities.

**A. Requirement Gathering**

The BA communicates with stakeholders to understand their business needs.

Common techniques:

- Interviews
- Meetings and workshops
- Questionnaires
- Observation
- Document analysis

Example: Interviewing the sales manager to understand what information is required in a sales dashboard.

**B. Requirement Analysis**

After gathering requirements, the BA analyses them to identify gaps, inconsistencies, dependencies and priorities.

Example: The sales manager requests a dashboard showing revenue, but the BA discovers that the manager also needs region-wise performance and monthly comparisons.

**C. Requirement Documentation**

The BA documents requirements so that developers and testers understand what needs to be delivered.

Important documents include:

- BRD – Business Requirements Document.
- FRD – Functional Requirements Document.
- SRS – Software Requirements Specification.
- User Stories.
- Acceptance Criteria.

**D. Communication with Stakeholders**

A BA coordinates between different teams, such as:

- Business teams
- Developers
- Testers
- Project Managers
- Product Owners
- Management

The BA ensures everyone has a common understanding of the requirements.

**E. Data Analysis and Reporting**

A BA uses tools such as Excel, SQL and Power BI to analyse business data and identify trends.

Example: Analysing sales data to identify why revenue declined in a particular region.

**F. Testing and UAT Support**

A BA supports User Acceptance Testing (UAT) to ensure the developed solution meets business requirements.

The BA may prepare test scenarios, coordinate with business users, track defects and obtain business sign-off.

**3. Detailed interview-ready answer**

This is how you can answer the interviewer:

A Business Analyst acts as a bridge between business stakeholders and the technical team.

The main responsibility of a Business Analyst is to understand business problems, gather requirements from stakeholders, analyse those requirements, and translate them into clear documentation so that the development team can build the right solution.

A Business Analyst is involved in different stages of a project.

First, we gather requirements through stakeholder meetings, interviews and workshops. Then we analyse the requirements, identify gaps and document them using BRDs, FRDs or user stories.

We also work closely with developers and testers to clarify requirements and ensure that the solution is developed according to business expectations.

Apart from requirement management, a Business Analyst may also use tools like Excel, SQL and Power BI to analyse data, prepare reports and provide business insights.

Finally, we support User Acceptance Testing to make sure the delivered solution meets the business requirements.

So, overall, a Business Analyst helps organisations solve business problems by connecting business needs with technical solutions.

**4. Practical real-world example**

Let's understand this through an e-commerce company.

As a Business Analyst, what would you do?

This is how a BA contributes to business improvement.

## **Q2. What is the difference between BRD and FRD?**

Very important interview question

This is one of the most frequently asked questions in Business Analyst interviews. The interviewer wants to know whether you understand how business requirements are converted into technical requirements.

Let's understand it from the basics.

### **1. What is BRD?**

**BRD stands for Business Requirements Document.**

A BRD is a document that explains what the business wants to achieve and why the requirement is needed.

It focuses on the business objectives, problems, expectations and high-level requirements.

In simple words:

- What does the business need?
- Why does the business need it?
- What problem are we trying to solve?
- What is the expected business outcome?

**Example:**

Imagine an e-commerce company wants to introduce an online order-tracking feature.

The business problem is that customers frequently contact customer support to know the status of their orders.

The business requirement could be:

"The company wants customers to track their orders online so that customer support calls can be reduced and customer satisfaction can be improved."

This is a BRD-level requirement because it explains the business need and objective without specifying the technical implementation.

### **2. What is FRD?**

**FRD stands for Functional Requirements Document.**

An FRD explains how the system should function to fulfil the business requirements mentioned in the BRD.

It provides detailed information about system functionality, inputs, outputs, validations, workflows and expected system behaviour.

In simple words:

- How should the system work?
- What features should be available?
- What happens when a user enters information?
- What validations should the system perform?
- What output should the system display?

**Example:**

Continuing with our e-commerce order-tracking example:

The FRD may specify:

1. The customer should be able to access the Track Order page.
2. The customer should enter a 10-digit Order ID.
3. The system should validate the entered Order ID.
4. If the Order ID is valid, the system should display the order status.
5. The system should display the order date and courier tracking link.
6. If the Order ID is invalid, the system should display an appropriate error message.

This is an FRD because it explains the detailed functionality of the system.

### **3. BRD vs FRD – Important differences**

This comparison is particularly important for your interview.

| Basis          | BRD                                                | FRD                                                  |
| -------------- | -------------------------------------------------- | ---------------------------------------------------- |
| Full form      | Business Requirements Document                     | Functional Requirements Document                     |
| Focus          | What and Why                                       | How                                                  |
| Purpose        | Explains business needs                            | Explains system functionality                        |
| Audience       | Business stakeholders, sponsors                    | Developers, testers, designers                       |
| Level          | High-level                                         | Detailed                                             |
| Content        | Objectives, scope, business rules, success metrics | Functions, screens, inputs, outputs, validations     |
| Created during | Requirement gathering and analysis                 | Functional analysis and specification                |
| Example        | Customers should track orders online               | System accepts Order ID and displays tracking status |

**Remember this simple relationship**

**BRD – Business Requirement**

What does the business need and why?

BA analyses and translates requirements

**FRD – Functional Requirement**

How should the system fulfil the requirement?

**Development and Testing**

Developers build the functionality and testers validate it.

### **4. Practical real-time example**

Let's take a banking application.

**Business problem:**

Customers visit bank branches to check their account balance. The bank wants to reduce branch visits by introducing an online balance enquiry feature.

**A. BRD – Business Requirements**

The BRD would contain:

- Business objective: Enable customers to check their account balances online.
- Business reason: Reduce branch visits and improve customer convenience.
- Expected benefit: Faster access to account information.
- Scope: Customers should be able to view their available account balance through the banking application.

Notice that we are explaining the business requirement, not the technical implementation.

**B. FRD – Functional Requirements**

The FRD would contain:

- The customer should log in using valid credentials.
- The system should authenticate the customer.
- The customer should navigate to the Account Summary page.
- The system should retrieve the account balance.
- The system should display the available balance.
- If the session expires, the system should request the customer to log in again.
- If the service is unavailable, the system should display an appropriate error message.

Here, we are explaining exactly how the application should behave.

### **5. What are the responsibilities of a BA while preparing BRD and FRD?**

This is an important follow-up question.

**While preparing BRD:**

1. Identify business stakeholders.
2. Understand the existing business problem.
3. Conduct stakeholder interviews and requirement-gathering sessions.
4. Understand business objectives and project scope.
5. Document high-level business requirements.
6. Identify assumptions, constraints and risks.
7. Get confirmation and sign-off from relevant stakeholders.

**While preparing FRD:**

1. Analyse the approved business requirements.
2. Convert business requirements into detailed functional requirements.
3. Define system behaviour and workflows.
4. Specify inputs, outputs and validations.
5. Identify alternate scenarios and exception handling.
6. Review requirements with developers and testers.
7. Ensure requirements are testable and traceable.

### **6. Interview-ready answer (60–90 seconds)**

This is the answer I recommend practising.

BRD stands for Business Requirements Document, and FRD stands for Functional Requirements Document.

The main difference between them is that BRD explains what the business needs and why it needs it, whereas FRD explains how the system should function to fulfil those business requirements.

A BRD contains high-level business objectives, project scope, business requirements and expected outcomes. It is mainly prepared for business stakeholders and management.

On the other hand, an FRD contains detailed functional specifications, including system behaviour, inputs, outputs, validations, workflows and use cases. It is mainly used by developers and testers.

For example, consider an e-commerce application.

The BRD requirement could be that customers should be able to track their orders online to reduce customer support calls.

The FRD would explain how this functionality works, such as allowing customers to enter their Order ID, validating it, displaying the current order status and showing an error message for an invalid Order ID.

As a Business Analyst, my responsibility is to understand the business requirements, document them clearly and translate them into functional specifications so that the technical team can implement the correct solution.

### **7. Possible follow-up questions**

Prepare these questions because the interviewer may go deeper.

**Q1. Who prepares the BRD and FRD?**

A Business Analyst generally coordinates and prepares these documents in consultation with stakeholders and the technical team. The exact ownership can vary between organisations.

**Q2. Who approves the BRD?**

Typically, the business sponsor or authorised business stakeholders review and approve the business requirements.

**Q3. Who uses the FRD?**

Primarily developers, testers, designers and other technical team members.

**Q4. Can BRD and FRD be combined into one document?**

Yes. Depending on the organisation and project methodology, both may be maintained in one document or separate documents.

**Q5. What happens if the FRD does not match the BRD?**

BRD focuses on business needs and objectives. SRS (Software Requirements Specification) provides a detailed specification of software requirements, which can include both functional and non-functional requirements.



## **Q3. What are the main sections of a BRD?**

This question is important because the interviewer wants to know whether you understand how to prepare a Business Requirements Document in a real project.

Let's understand each section with a practical example.

## **1. What is a BRD?**

BRD stands for **Business Requirements Document**.

It is a formal document that describes the business problem, business objectives, project scope, stakeholder expectations and high-level requirements.

A BRD helps all stakeholders understand:

- What problem are we solving?
- Why are we solving it?
- What does the business expect?
- What is included and excluded from the project?
- What results should the project achieve?

**Practical example:**

Imagine you are working as a Business Analyst for an e-commerce company.

The company receives too many customer-support calls because customers cannot track their orders.

Management decides to develop an online order-tracking feature.

As a BA, you need to prepare a BRD explaining the business requirements for this project.

## **2. Main sections of a BRD**

According to your interview preparation guide (page 3), a BRD generally contains the following sections:

1. Executive Summary
2. Business Objectives
3. Project Scope (In Scope and Out of Scope)
4. Stakeholders
5. Current State (AS-IS) and Future State (TO-BE)
6. Business Requirements
7. Assumptions and Constraints
8. Risks
9. Success Criteria / KPIs
10. Sign-off

Let's understand each one.

### **1. Executive Summary**

**Meaning:** A brief overview of the entire project.

It explains what the project is about, why it is needed and what business problem it addresses.

It helps management understand the project without reading the entire document.

**Example – E-commerce project:**

"The purpose of this project is to develop an online order-tracking feature that allows customers to view their order status and delivery information. This initiative aims to improve customer satisfaction and reduce the number of order-related support calls."

**As a BA:** You prepare a concise summary based on discussions with business stakeholders.

### **2. Business Objectives**

**Meaning:** This section explains what the organisation wants to achieve through the project.

Objectives should ideally be measurable.

**Example:**

- Reduce order-related customer-support calls.
- Improve customer satisfaction.
- Provide transparent delivery information.
- Reduce manual work for the customer-support team.

For example, management might set a target of reducing order-status enquiries by 30%.

**As a BA:** You understand management expectations and translate them into measurable business objectives.

### **3. Project Scope**

**Meaning:** Project scope defines what activities and functionalities are included and excluded from the project.

It has two important parts.

**A. In Scope**

Features and activities that are included in the project.

Example:

- Customers can track their orders.
- Customers can view delivery status.
- Customers can view estimated delivery dates.
- Customers can access courier tracking information.

**B. Out of Scope**

Features and activities that are not included in the current project.

Example:

- Online payment integration.
- Order cancellation functionality.
- Product returns and refunds.
- Changing delivery addresses.

**Why is this important?**

It prevents scope creep, which means uncontrolled additions to project requirements.

### **4. Stakeholders**

**Meaning:** Stakeholders are individuals or teams who are involved in, affected by, or have an interest in the project.

For our e-commerce project:

| Stakeholder      | Responsibility                     |
| ---------------- | ---------------------------------- |
| Business Manager | Defines business objectives        |
| Business Analyst | Gathers and documents requirements |
| Development Team | Develops the application           |
| Testing Team     | Tests the functionality            |
| Customer Support | Provides customer-related insights |
| Customers        | Use the order-tracking feature     |
| Project Manager  | Manages project delivery           |

**As a BA:** You identify stakeholders, understand their expectations and coordinate with them throughout the project.

### **5. AS-IS and TO-BE Process**

This is a particularly important concept for a Business Analyst.

**AS-IS means the current process.**

It explains how the business process works today.

Example:

Currently, customers contact customer support to ask about their order status. Support employees manually check the order details and respond.

**AS-IS – Current process**

Customer places an order

Customer contacts support

Support employee checks order status

Support employee updates customer

**TO-BE means the desired future process.**

It explains how the business process should work after implementing the proposed solution.

Example:

Customers can independently check their order status through the application.

**TO-BE – Future process**

Customer places an order

Customer opens Track Order

System retrieves order information

Customer views delivery status

**As a BA:** You study the current process, identify problems and recommend improvements to achieve the desired future state.

### **6. Business Requirements**

**Meaning:** This section contains the high-level requirements requested by the business.

Example:

| Requirement ID | Business requirement                                       |
| -------------- | ---------------------------------------------------------- |
| BR-01          | Customers should be able to track orders online.           |
| BR-02          | Customers should be able to view delivery status.          |
| BR-03          | Customers should be able to view estimated delivery dates. |
| BR-04          | The system should provide courier tracking information.    |

Each requirement should have a unique ID so it can be tracked throughout the project.

### **7. Assumptions and Constraints**

These are two different concepts.

**Assumptions:** Things considered to be true while planning the project, but which have not necessarily been confirmed.

Examples:

- Customers have access to the internet.
- The logistics team will provide updated delivery information.
- Customers have valid order IDs.

**Constraints:** Limitations or restrictions affecting the project.

Examples:

- The project must be completed within three months.
- The project has a limited development budget.
- The feature must work with the existing e-commerce platform.

### **8. Risks**

**Meaning:** Risks are potential problems that could negatively affect the project.

Examples:

| Risk                         | Possible impact                |
| ---------------------------- | ------------------------------ |
| Delayed logistics data       | Incorrect tracking information |
| Technical integration issues | Project delays                 |
| Incomplete requirements      | Rework and additional cost     |
| Data security issues         | Customer information exposure  |

**As a BA:** You identify potential risks, communicate them to stakeholders and coordinate with the relevant teams to address them.

### **9. Success Criteria / KPIs**

**Meaning:** This section defines how the organisation will measure whether the project has achieved its objectives.

KPI stands for Key Performance Indicator.

Example:

| KPI                                      | Example target |
| ---------------------------------------- | -------------- |
| Reduction in order-related support calls | 30%            |
| Customer satisfaction                    | 90%            |
| Tracking feature availability            | 99.5%          |
| Successful tracking requests             | 98%            |

These are illustrative project targets, not actual company figures.

**As a BA:** You work with stakeholders to define measurable success criteria and validate results after implementation.

### **10. Sign-off**

**Meaning:** Sign-off is the formal approval of the documented requirements by authorised stakeholders.

Before development begins, relevant stakeholders review the BRD and confirm that the documented business requirements accurately represent their expectations.

Sign-off helps establish an agreed baseline for the project.

**As a BA:**

- Share the BRD with stakeholders.
- Collect feedback.
- Make necessary changes.
- Obtain formal approval.
- Maintain the approved version.

## **5. Possible follow-up interview questions**

### **1. What is the difference between AS-IS and TO-BE?**

**Answer:**

AS-IS represents the current business process, explaining how things are working at present. TO-BE represents the desired future process, explaining how the business wants the process to work after implementing improvements.

As a Business Analyst, I analyse the AS-IS process to identify existing problems, bottlenecks and inefficiencies. Then I define the TO-BE process by suggesting improvements or solutions.

For example, currently customers contact customer support to check their order status. In the TO-BE process, customers can independently track their orders through an online application.

### **2. What is the difference between assumptions and constraints?**

**Answer:**

Assumptions are conditions that we consider to be true while planning or executing a project, even though they may not be fully confirmed.

Constraints are limitations or restrictions that affect the project.

For example, assuming that the logistics team will provide updated delivery information is an assumption.

However, having a fixed project budget or a deadline of three months is a constraint.

As a Business Analyst, I document both assumptions and constraints because they can affect project planning, requirements and delivery.

### **3. What is scope creep, and how do you prevent it?**

**Answer:**

Scope creep refers to the uncontrolled addition of new requirements or features after the project scope has been agreed upon.

It can lead to project delays, increased costs and additional workload.

As a Business Analyst, I prevent scope creep by clearly defining the project scope, documenting requirements, obtaining stakeholder sign-off and following a formal change request process.

Whenever a stakeholder requests a new feature, I analyse its impact on time, cost and resources, discuss it with the relevant stakeholders and obtain approval before including it in the project.

### **4. Who is responsible for approving a BRD?**

**Answer:**

The Business Analyst is generally responsible for preparing and coordinating the BRD, but the final approval usually comes from the business sponsor or authorised business stakeholders.

The BA gathers requirements, documents them, reviews the document with stakeholders, incorporates feedback and submits it for approval.

Once the stakeholders confirm that the requirements accurately represent their business needs, they provide formal sign-off.

### **5. What is a business requirement?**

**Answer:**

A business requirement describes a high-level need or objective that an organisation wants to achieve through a project or solution.

It explains what the business needs and why that need is important, without going into detailed technical implementation.

For example, in an e-commerce company, a business requirement could be that customers should be able to track their orders online to improve customer satisfaction and reduce customer support calls.

As a Business Analyst, my responsibility is to gather, analyse, document and validate business requirements with stakeholders.

### **6. What is the difference between a business objective and a business requirement?**

**Answer:**

A business objective describes the overall goal that an organisation wants to achieve, whereas a business requirement describes what the business needs to fulfil that objective.

For example, the business objective of an e-commerce company could be to improve customer satisfaction and reduce customer support calls by 30%.

To achieve this objective, a business requirement could be to introduce an online order-tracking feature that allows customers to check their delivery status independently.

So, the business objective defines the desired outcome, while the business requirement defines the business need that helps achieve that outcome.

### **7. What happens if a stakeholder wants to change a requirement after BRD sign-off?**

**Answer:**

If a stakeholder requests a change after BRD sign-off, I would not immediately accept or reject the request.

First, I would understand the reason behind the change and document it as a formal change request.

Then I would perform an impact analysis with the technical team to understand how the change affects the project timeline, cost, resources and existing requirements.

After that, I would discuss the impact and available options with the relevant stakeholders and obtain approval from the authorised decision-maker.

Once the change is approved, I would update the BRD, requirements traceability matrix and other relevant documents, and communicate the changes to the development and testing teams.

This ensures that changes are properly controlled and do not create unnecessary scope creep.

## **Q4. What is the difference between Functional and Non-Functional Requirements?**

Very important interview question

This is a fundamental BA question. The interviewer wants to understand whether you can identify what a system should do and how well it should perform.

### **1. What are Functional Requirements?**

**Functional Requirements describe what a system should do.**

They define the specific features, operations and functionalities that a system must provide to users.

In simple words, they explain the expected behaviour of the application.

**Examples:**

Consider an online shopping application.

1. Users should be able to register and log in.
2. Users should be able to search for products.
3. Users should be able to add products to their shopping cart.
4. Users should be able to place orders.
5. Users should be able to make payments.
6. Users should be able to track their orders.

These are functional requirements because they describe what the application should do.

### **2. What are Non-Functional Requirements?**

**Non-Functional Requirements describe how well a system should perform.**

They define the quality attributes, performance standards and operational conditions that the system must satisfy.

In simple words, they explain how efficiently, securely and reliably the application should work.

**Examples:**

Using the same shopping application:

1. Performance: The application should load within 3 seconds under specified conditions.
2. Security: Customer information should be protected through appropriate security controls.
3. Usability: The application should be easy to navigate.
4. Availability: The application should maintain 99.9% availability.
5. Scalability: The application should support increased users during peak shopping periods.

These are non-functional requirements because they describe the quality and performance of the system.

## **3. Functional vs Non-Functional Requirements**

| Basis            | Functional Requirements      | Non-Functional Requirements              |
| ---------------- | ---------------------------- | ---------------------------------------- |
| Meaning          | What the system should do    | How well the system should perform       |
| Focus            | Features and functionality   | Quality and performance                  |
| Purpose          | Define system behaviour      | Define system quality                    |
| Examples         | Login, registration, payment | Security, speed, availability            |
| Testing          | Functional testing           | Performance, security, usability testing |

## **Q6. Which requirement elicitation techniques do you know?**

### **1. What is Requirement Elicitation?**

Requirement Elicitation is the process of collecting, discovering and understanding business requirements from stakeholders to identify what they need from a project or system.

In simple words, it means understanding the business requirements by communicating with stakeholders, observing their work, analysing existing documents and discussing possible solutions.

**Example:**

Suppose a company wants to develop a new employee management system.

As a Business Analyst, you need to understand:

- What problems employees are currently facing.
- How HR manages employee information.
- What features HR expects in the new system.
- What reports management needs.
- What improvements are required.

To gather this information, you use different requirement elicitation techniques.

### **2. What are the different Requirement Elicitation Techniques?**

There are eight important techniques.

**1. Interviews**

Interviews involve directly communicating with stakeholders to understand their requirements, expectations and problems.

The BA asks questions and collects detailed information.

Example:

If you are developing an employee management system, you interview the HR manager to understand how employee attendance, leave management and payroll currently work.

When to use:

- When detailed information is required.
- When requirements are unclear.
- When you need to understand individual stakeholder expectations.

**2. Workshops (JAD Sessions)**

Workshops are structured meetings where multiple stakeholders come together to discuss, identify and finalise requirements.

JAD stands for Joint Application Development.

Example:

You organise a workshop with HR managers, employees, developers and project managers to discuss the features required in an employee management system.

When to use:

- When multiple stakeholders are involved.
- When requirements need to be discussed collectively.
- When you need to resolve conflicting requirements.

**3. Questionnaires and Surveys**

Questionnaires and surveys are used to collect information from a large number of stakeholders through predefined questions.

Example:

A company wants to understand employee satisfaction with its existing HR portal.

As a BA, you prepare a questionnaire asking employees about login issues, leave application problems and missing features.

When to use:

- When there are many stakeholders.
- When stakeholders are located in different places.
- When you need structured feedback quickly.

**4. Observation (Job Shadowing)**

Observation involves watching users perform their daily activities to understand the existing business process.

Example:

You observe an HR executive processing employee leave requests to understand the actual workflow, manual activities and challenges.

Sometimes users may not explain their complete process during interviews, so observation helps identify hidden problems.

When to use:

- When users cannot clearly explain their processes.
- When you need to understand actual day-to-day activities.
- When identifying process inefficiencies.

**5. Document Analysis**

Document analysis involves reviewing existing business documents, reports, policies and system records to understand the current process and requirements.

Example:

For an employee management system, you review:

- Existing HR policies.
- Employee attendance reports.
- Leave application forms.
- Payroll reports.
- Existing process documentation.

This helps you understand existing business rules and identify missing information.

When to use:

- When an existing system or process is available.
- When historical information is required.
- When understanding business rules and workflows.

**6. Brainstorming**

Brainstorming is a technique where stakeholders come together to generate ideas and identify possible solutions to a business problem.

Example:

A company wants to improve customer satisfaction.

You conduct a brainstorming session with the customer support team to identify possible improvements, such as automated notifications, self-service options and faster complaint resolution.

When to use:

- When exploring new ideas.
- When identifying possible solutions.
- When solving complex business problems.

**7. Prototyping / Wireframing**

Prototyping involves creating a sample or preliminary model of a proposed system to help stakeholders visualise and validate requirements before actual development.

Example:

A company wants to develop an online banking application.

As a BA, you create a simple wireframe showing:

- Login screen.
- Account summary page.
- Fund transfer page.
- Transaction history page.

You share the wireframe with stakeholders and collect their feedback.

When to use:

- When requirements are unclear.
- When developing a new application.
- When stakeholders need to visualise the proposed solution.

**8. Focus Groups**

A focus group involves gathering a selected group of users to discuss their opinions, experiences and expectations about a product or system.

Example:

An e-commerce company wants to improve its mobile application.

You invite a group of customers to discuss their shopping experience, payment issues and expectations for new features.

When to use:

- When collecting user opinions.
- When understanding customer expectations.
- When evaluating a proposed product or feature.

## **Q7. What is a User Story and what are Acceptance Criteria?**

### **1. What is a User Story?**

A User Story is a short description of a requirement written from the perspective of the end user.

It explains what the user wants, what functionality they need and why they need it.

User Stories are commonly used in Agile methodology to capture requirements in a simple and understandable format.

The standard format is:

**User Story Format**

As a **[type of user]**,

I want **[some functionality]**,

So that **[business benefit or reason]**.

**Example:**

Consider an e-commerce application.

"As a customer, I want to track my order online so that I can know the current delivery status without contacting customer support."

Let's understand the three parts:

- **As a customer:** Who wants the functionality?
- **I want to track my order online:** What functionality is required?
- **So that I can know the current delivery status:** Why is the functionality needed?

### **2. What are Acceptance Criteria?**

Acceptance Criteria are the specific conditions that a feature or user story must satisfy to be considered complete and acceptable.

They help developers understand what needs to be implemented and testers understand what needs to be verified.

In simple words, Acceptance Criteria define the expected conditions and outcomes of a feature.

**Example:**

For the above order-tracking User Story, the acceptance criteria could be:

1. The customer should be able to enter an Order ID.
2. The system should validate the entered Order ID.
3. If the Order ID is valid, the system should display the current order status.
4. The system should display the estimated delivery date.
5. If the Order ID is invalid, the system should display an error message.

These conditions help confirm whether the feature meets the user's expectations.

### **3. What is Given–When–Then?**

Given–When–Then is a commonly used format for writing acceptance criteria, particularly in Behaviour-Driven Development (BDD).

It consists of three parts:

| Component | Meaning                            |
| --------- | ---------------------------------- |
| Given     | The initial condition or situation |
| When      | The action performed by the user   |
| Then      | The expected result                |

**Practical example: Order Tracking**

- **Given:** The customer has a valid Order ID.
- **When:** The customer enters the Order ID and clicks the Track Order button.
- **Then:** The system should display the current order status and estimated delivery date.

Another example – Invalid Order ID:

- **Given:** The customer enters an invalid Order ID.
- **When:** The customer clicks the Track Order button.
- **Then:** The system should display an appropriate error message.

### **4. What is the difference between User Story and Acceptance Criteria?**

| Basis   | User Story                           | Acceptance Criteria                                      |
| ------- | ------------------------------------ | -------------------------------------------------------- |
| Meaning | Describes the user's requirement     | Defines conditions for accepting the requirement         |
| Purpose | Explains what the user wants and why | Explains what must happen for the feature to be accepted |
| Format  | As a... I want... So that...         | Given... When... Then...                                 |
| Detail  | High-level                           | Specific and detailed                                    |
| Used by | BA, Product Owner, development team  | Developers, testers, BA                                  |
| Example | Customer wants to track orders       | Valid Order ID displays tracking status                  |

**Remember:** A User Story explains what the user wants, while Acceptance Criteria explain how we know that the requirement has been fulfilled.

4. Defining acceptance criteria.

7. Updating stories based on stakeholder feedback.

## **Q8. What makes a good requirement? Explain INVEST and SMART.**

### **1. What makes a good requirement?**

A good requirement clearly describes what the business or user needs from a system or solution.

It should be understandable, complete, realistic and testable so that developers can implement it and testers can verify it.

A good requirement should have the following characteristics:

**1. Clear and Unambiguous**

The requirement should be easy to understand and should not have multiple interpretations.

Example:

Incorrect: "The application should load quickly."

Correct: "The application should load the dashboard within 3 seconds under the agreed operating conditions."

**2. Complete**

The requirement should contain all the necessary information without missing important details.

Example:

Instead of saying "Users should log in", specify the expected login inputs, validation rules and behaviour for invalid credentials.

**3. Consistent**

The requirement should not conflict with other documented requirements.

Example:

One requirement should not state that customers can cancel orders at any time while another prohibits cancellation after dispatch, unless the distinction is clearly defined.

**4. Testable**

The requirement should be possible to verify through testing.

Example:

"The system should display the order status within 2 seconds" can be tested by measuring response time under defined conditions.

**5. Feasible**

The requirement should be achievable within the available technology, budget, resources and timeline.

**6. Traceable**

Every requirement should have a unique ID and should be traceable throughout the project, from documentation to development and testing.

Example: BR-001 can be linked to its corresponding functional requirement and test case.

### **2. What is INVEST?**

INVEST is a set of six principles used to evaluate the quality of User Stories in Agile projects.

| Letter | Principle   | Meaning                                                     |
| ------ | ----------- | ----------------------------------------------------------- |
| I      | Independent | A story should have minimal dependencies on other stories.  |
| N      | Negotiable  | Details can be discussed and refined with the team.         |
| V      | Valuable    | It should provide value to the user or business.            |
| E      | Estimable   | The development team should be able to estimate the effort. |
| S      | Small       | It should be small enough to complete within a sprint.      |
| T      | Testable    | It should have clear conditions to verify completion.       |

**Practical example:**

Consider this User Story:

"As a customer, I want to track my order online so that I can know the delivery status."

Let's apply INVEST:

- Independent: The story should be deliverable with minimal dependency on unrelated features.
- Negotiable: The tracking details and display format can be discussed with stakeholders.
- Valuable: Customers can independently check delivery information.
- Estimable: Developers can estimate the effort required.
- Small: The story can be divided into manageable development tasks.
- Testable: Testers can verify whether order tracking works correctly.

### **3. What is SMART?**

SMART is a framework used to define clear and measurable objectives.

| Letter | Principle  | Meaning                                    |
| ------ | ---------- | ------------------------------------------ |
| S      | Specific   | Clearly defines what needs to be achieved. |
| M      | Measurable | Progress can be measured.                  |
| A      | Achievable | The target is realistic.                   |
| R      | Relevant   | It aligns with business objectives.        |
| T      | Time-bound | It has a defined deadline.                 |

**Practical example:**

A company wants to reduce customer-support calls related to order tracking.

A SMART objective could be:

"Reduce order-status-related customer support calls by 30% within six months by implementing an online order-tracking feature."

Let's understand:

- Specific: Reduce order-status-related support calls.
- Measurable: 30% reduction.
- Achievable: Based on available resources and feasibility.
- Relevant: Improves customer experience and reduces support workload.
- Time-bound: Within six months.

### **4. Difference between INVEST and SMART**

| Basis          | INVEST                                   | SMART                                     |
| -------------- | ---------------------------------------- | ----------------------------------------- |
| Purpose        | Evaluates User Story quality             | Defines effective objectives              |
| Mainly used in | Agile requirement management             | Business planning and goal setting        |
| Focus          | User Stories                             | Business objectives and targets           |
| Key benefit    | Makes stories easier to develop and test | Makes goals measurable and achievable     |
| Example        | Customer can track orders                | Reduce support calls by 30% in six months |

## **Q9. What is Gap Analysis?**

### **1. What is Gap Analysis?**

Gap Analysis is a technique used by a Business Analyst to compare the **current state (AS-IS)** of a business process with the **desired future state (TO-BE)** to identify what is missing and what improvements are required to achieve the business objectives.

In simple words, Gap Analysis helps us understand:

- Where are we currently?
- Where do we want to reach?
- What is missing?
- What actions are required to close the gap?

### **2. Practical Real-Time Example**

Consider an e-commerce company.

**Business problem:**

The company currently processes customer orders manually, which causes delays and errors.

**Step 1: Identify AS-IS (Current State)**

Currently:

- Employees manually check customer orders.
- Order details are updated in Excel.
- Delivery information is shared manually.
- Customers frequently contact customer support.

## **Q10. What is SWOT Analysis and Root Cause Analysis?**

### **1. What is SWOT Analysis?**

SWOT Analysis is a strategic analysis technique used to evaluate a business, project or organisation by identifying four important factors:

- **S – Strengths**
- **W – Weaknesses**
- **O – Opportunities**
- **T – Threats**

It helps organisations understand their internal capabilities and external business environment before making decisions.

### **2. Explain the four components of SWOT Analysis**

**1. Strengths**

Strengths are the internal positive factors that give an organisation an advantage over competitors.

Examples:

- Strong brand reputation.
- Experienced employees.
- Good customer relationships.
- Advanced technology.
- Strong financial position.

**2. Weaknesses**

Weaknesses are internal limitations that may negatively affect business performance.

Examples:

- Outdated technology.
- Poor customer service.
- Limited resources.
- Manual business processes.
- Lack of skilled employees.

**3. Opportunities**

Opportunities are external factors that a business can use to improve its performance and growth.

Examples:

- Entering new markets.
- Increasing demand for online services.
- Introducing new products.
- Adopting new technologies.
- Expanding the customer base.

**4. Threats**

Threats are external factors that may negatively affect the organisation.

Examples:

- Increasing competition.
- Changes in government regulations.
- Economic slowdown.
- Changing customer preferences.
- New competitors entering the market.

### **3. Practical Example of SWOT Analysis**

Consider an e-commerce company that wants to expand its online business.

| SWOT          | Example                                                |
| ------------- | ------------------------------------------------------ |
| Strengths     | Strong brand reputation and loyal customers            |
| Weaknesses    | Slow delivery process                                  |
| Opportunities | Growing demand for online shopping                     |
| Threats       | Increasing competition from other e-commerce companies |

**As a Business Analyst:**

I can use SWOT Analysis to understand the company's current position, identify business challenges and opportunities, and provide insights that support management decision-making.

### **4. What is Root Cause Analysis (RCA)?**

Root Cause Analysis is a systematic technique used to identify the underlying reason behind a business problem rather than simply addressing its symptoms.

In simple words, RCA helps us understand **why a problem is happening**, so that we can implement a solution that prevents it from happening again.

**Example:**

Suppose an e-commerce company is experiencing an increase in delayed deliveries.

The immediate problem is that customers are not receiving their orders on time.

However, instead of immediately hiring more delivery employees, a Business Analyst investigates the actual reason behind the delays.

Possible causes:

- Incorrect delivery addresses.
- Delays in warehouse processing.
- Poor coordination with courier partners.
- Inefficient order-management processes.

After investigating, the BA identifies the actual root cause and recommends an appropriate solution.

### **5. Common Root Cause Analysis Techniques**

There are two important techniques you should know for your interview.

**A. 5 Whys Technique**

The 5 Whys technique involves repeatedly asking "Why?" to identify the underlying cause of a problem.

Practical example:

**Problem:** Customer orders are getting delayed.

| Question                           | Answer                                                        |
| ---------------------------------- | ------------------------------------------------------------- |
| Why are orders delayed?            | Orders are dispatched late.                                   |
| Why are they dispatched late?      | Warehouse processing takes too long.                          |
| Why does processing take too long? | Employees manually verify orders.                             |
| Why is verification manual?        | There is no automated verification system.                    |
| Why is there no automated system?  | The company has not implemented an order-management solution. |

**Identified root cause:** Lack of an automated order-management process.

**Recommended solution:** Introduce an automated order-management system to reduce manual verification and processing time.

**B. Fishbone Diagram (Ishikawa Diagram)**

A Fishbone Diagram is a visual technique used to identify and organise possible causes of a particular problem.

It is also called a Cause-and-Effect Diagram.



For example, if customer complaints are increasing, possible causes can be categorised into:

- **People:** Insufficient employee training.
- **Process:** Inefficient complaint-handling process.
- **Technology:** Outdated customer-support software.
- **Data:** Incorrect customer information.
- **Management:** Lack of proper monitoring.

The BA investigates these possible causes to identify the actual underlying problem.

### **6. Difference between SWOT Analysis and Root Cause Analysis**

| Basis      | SWOT Analysis                    | Root Cause Analysis                    |
| ---------- | -------------------------------- | -------------------------------------- |
| Purpose    | Evaluate business position       | Identify the actual cause of a problem |
| Focus      | Internal and external factors    | Underlying problems                    |
| Used for   | Strategic planning               | Problem-solving                        |
| Techniques | SWOT matrix                      | 5 Whys, Fishbone Diagram               |
| Example    | Analysing business opportunities | Investigating delayed deliveries       |

## **Q11. What is MoSCoW Prioritisation?**

**MoSCoW prioritisation** is a requirement prioritisation technique used by Business Analysts to classify requirements based on their importance to the project.

It is particularly useful when a project has limited time, budget, or resources and the team cannot deliver every requirement at once.

MoSCoW stands for:

### **1. Must Have**

These are critical requirements. Without them, the solution cannot meet its primary business objectives or be considered usable.

**Example:** In an e-commerce application, customers must be able to add products to their cart and complete payments.

### **2. Should Have**

These are important requirements, but the project can still function without them temporarily. They should ideally be delivered if time and resources permit.

**Example:** Customers should receive an order confirmation email after making a purchase.

### **3. Could Have**

These are desirable features that improve the user experience but have relatively low priority.

**Example:** Customers could have the option to save products to a wishlist.

### **4. Won’t Have (This Time)**

These are requirements that stakeholders agree will not be included in the current release. They may be considered for a future release.

**Example:** Adding AI-based product recommendations may be postponed to a future version.

## **How does a Business Analyst use MoSCoW?**

A Business Analyst typically:

1. Collects requirements from stakeholders.
2. Understands the business value and urgency of each requirement.
3. Discusses dependencies, risks, and implementation effort.
4. Classifies requirements into the four MoSCoW categories.
5. Gets agreement from stakeholders and the project team.
6. Documents the priorities and reviews them when project circumstances change.

**Example: Prioritising an e-commerce application**

| Requirement                 | Priority               | Reason                               |
| --------------------------- | ---------------------- | ------------------------------------ |
| User registration and login | Must Have              | Required for customer accounts       |
| Online payment              | Must Have              | Essential for completing purchases   |
| Order confirmation email    | Should Have            | Improves customer communication      |
| Wishlist                    | Could Have             | Enhances user experience             |
| AI product recommendations  | Won’t Have (This Time) | Can be considered in a later release |

## **Interview-ready answer**

MoSCoW is a requirement prioritisation technique used by Business Analysts to classify requirements into four categories: Must Have, Should Have, Could Have, and Won’t Have for this release.

Must Have requirements are essential for project success. Should Have requirements are important but not critical. Could Have requirements are optional enhancements, and Won’t Have requirements are excluded from the current release but may be considered later.

For example, in an e-commerce project, online payment would be a Must Have, order confirmation emails could be a Should Have, a wishlist could be a Could Have, and AI-based recommendations could be deferred to a future release.

As a Business Analyst, I would use MoSCoW to understand stakeholder priorities, consider business value and implementation constraints, facilitate prioritisation discussions, and ensure the team focuses on delivering the most important requirements first.

MoSCoW classifies requirements into four priority categories. Business value prioritisation ranks requirements based on the value they deliver to the organisation or its customers. Business value can be one of the factors used when assigning MoSCoW categories.

## **Q12. What is a Requirements Traceability Matrix (RTM)?**

A **Requirements Traceability Matrix (RTM)** is a document or table used by a Business Analyst to track requirements throughout the project lifecycle. It links each requirement to its related design elements, development work, and test cases.

Its main purpose is to ensure that every approved requirement is implemented and tested, and that no requirement is missed.

### **Example of an RTM**

Suppose you are working on an e-commerce application with the following requirements:

| Requirement ID | Requirement              | Design/Feature       | Test Case ID | Status      |
| -------------- | ------------------------ | -------------------- | ------------ | ----------- |
| REQ001         | User login               | Login page           | TC001        | Passed      |
| REQ002         | Add products to cart     | Shopping cart        | TC002        | Passed      |
| REQ003         | Online payment           | Payment gateway      | TC003        | In Progress |
| REQ004         | Order confirmation email | Notification service | TC004        | Not Started |

The RTM helps the BA check whether each requirement has been designed, implemented, and tested.

### **Why is an RTM important?**

1. **Requirement coverage:** Ensures every requirement is accounted for.
2. **Testing validation:** Confirms that each requirement has a corresponding test case.
3. **Change impact analysis:** Helps identify which designs, features, and test cases may be affected when a requirement changes.
4. **Avoiding missed requirements:** Reduces the risk of requirements being overlooked.
5. **Project tracking:** Helps stakeholders monitor requirement implementation and testing status.

### **What are the common types of traceability?**

- **Forward traceability:** Tracks requirements forward to design, development, and test cases. It checks whether all requirements are implemented.
- **Backward traceability:** Tracks test cases and implementation back to the original requirements. It helps identify work that was not supported by an approved requirement.
- **Bidirectional traceability:** Combines both directions and provides complete traceability.

### **How does a Business Analyst maintain an RTM?**

A BA typically:

1. Assigns a unique ID to each requirement.
2. Records the requirement description and source.
3. Links it to relevant design documents and development features.
4. Maps it to test cases.
5. Updates implementation and testing status.
6. Revises the matrix when requirements change.

### **Interview-ready answer**

An RTM helps ensure that all requirements are implemented and tested. It also supports change impact analysis, prevents requirements from being missed, and helps track project progress.

Forward traceability checks whether every requirement is carried through to implementation and testing. Backward traceability checks whether every implemented feature or test case can be linked to an approved requirement.

## **Q14. What is the difference between Waterfall and Agile methodologies?**

**Waterfall and Agile are two different software development methodologies used to manage and deliver projects.**

The main difference is that Waterfall follows a sequential approach, where each phase is generally completed before the next phase begins. Agile follows an iterative and incremental approach, where software is developed and delivered in smaller cycles, with continuous feedback and improvements.

A Business Analyst works in both methodologies, but the nature of the BA's responsibilities and documentation can differ.

### **1. Waterfall Methodology**

Waterfall is a traditional project management and software development approach in which the project progresses through predefined phases in sequence.

The typical phases are:

1. Requirement gathering
2. Requirement analysis
3. System design
4. Development
5. Testing
6. Deployment
7. Maintenance

Each phase is generally completed and reviewed before the next phase starts.

**How does Waterfall work?**

Suppose an organisation wants to develop an employee management system.

Before development begins, the BA gathers detailed requirements from HR, payroll, management, and IT teams.

The BA prepares requirement documents such as:

- Business Requirements Document (BRD)
- Functional Requirements Document (FRD)
- Process flow diagrams
- Use cases
- Requirement specifications

After the requirements are reviewed and approved, the development team starts building the system.

Once development is completed, the testing team verifies the application against the documented requirements.

If the organisation wants to introduce a new feature after requirements have been approved, the change usually goes through a formal change-request and impact-analysis process.

**Role of a BA in Waterfall:**

- Conduct detailed requirement-gathering sessions.
- Prepare and maintain BRD and FRD documents.
- Define business rules and functional requirements.
- Obtain stakeholder approval and sign-off.
- Communicate requirements to developers and testers.
- Manage requirement changes through formal change control.
- Support testing and UAT.
- Ensure the final application matches the approved requirements.

**Key characteristics of Waterfall:**

- Sequential development phases.
- Detailed documentation.
- Requirements are generally defined upfront.
- Formal approvals and sign-offs.
- Changes can be more difficult and expensive once development has progressed.
- Testing generally follows development.

### **2. Agile Methodology**

Agile is an iterative and incremental approach in which a project is divided into smaller development cycles called **sprints**.

A sprint is a fixed period during which the team plans, develops, tests, and reviews a selected set of requirements.

A common sprint duration is one to four weeks, although the duration depends on the team's working approach.

Instead of defining every detail before development starts, Agile teams work on prioritised requirements and refine them continuously based on feedback.

**How does Agile work?**

Consider the same employee management system.

Instead of waiting until the entire application is completed, the team divides the project into smaller features.

For example:

- Sprint 1: Employee registration and login.
- Sprint 2: Employee profile management.
- Sprint 3: Attendance management.
- Sprint 4: Leave management.
- Sprint 5: Payroll reporting.

At the end of each sprint, the team demonstrates the completed functionality to stakeholders, collects feedback, and uses that feedback to plan future work.

Requirements are generally maintained in a **Product Backlog**, which contains features, user stories, enhancements, and defects.

**Role of a BA in Agile:**

- Work closely with the Product Owner and stakeholders.
- Gather and clarify business requirements.
- Convert requirements into user stories.
- Define acceptance criteria for each user story.
- Maintain and refine the product backlog.
- Participate in backlog refinement sessions.
- Support sprint planning by clarifying requirements.
- Attend daily stand-ups when relevant to the BA's work.
- Participate in sprint reviews and gather stakeholder feedback.
- Support testing and UAT.
- Update user stories when requirements evolve.

**Example of a user story in Agile:**

**User story: Employee login**

As an employee, I want to log in using my registered email and password so that I can access my employee dashboard.

**Acceptance criteria:**

1. The employee should be able to log in using valid credentials.
2. The system should display an error message for invalid credentials.
3. The employee should be redirected to the dashboard after successful login.
4. The system should provide a password-reset option.

**Key characteristics of Agile:**

- Iterative and incremental development.
- Work is divided into sprints.
- Requirements can evolve.
- Continuous stakeholder feedback.
- Frequent delivery of working software.
- Collaboration between business and technical teams.
- User stories and acceptance criteria are commonly used.

### **4. What is the BA's role in Agile ceremonies?**

An interviewer may ask this to check whether you understand how an Agile team works.

**Sprint Planning**

The team decides which prioritised user stories will be taken into the upcoming sprint. The BA clarifies requirements, business rules, dependencies, and acceptance criteria.

**Daily Stand-up**

A short daily team meeting where members share progress, planned work, and blockers. The BA may attend to clarify requirements or communicate issues affecting delivery.

**Backlog Refinement**

The team reviews upcoming backlog items, clarifies requirements, breaks down large stories, identifies dependencies, and estimates effort. The BA plays an important role in ensuring stories are sufficiently clear before development.

**Sprint Review**

The team demonstrates completed work to stakeholders and gathers feedback. The BA helps validate whether the delivered functionality meets business expectations.

**Sprint Retrospective**

The team discusses what went well, what did not go well, and what can be improved in the next sprint. The BA can contribute suggestions to improve requirement gathering, communication, and collaboration.

### **5. Can Agile and Waterfall be used together?**

Yes. Some organisations use a **Hybrid approach**, combining elements of both methodologies.

For example, an organisation may use Waterfall for upfront project approvals, budget planning, and high-level documentation, while using Agile sprints for software development and incremental delivery.

A BA working in a hybrid environment may prepare a high-level BRD while also creating detailed user stories for individual development sprints.

## **Q15. What is User Acceptance Testing (UAT), and what is the role of a Business Analyst in UAT?**

**User Acceptance Testing (UAT)** is the final testing phase in which actual business users or designated stakeholders verify whether the developed application meets their business requirements and is suitable for real-world use.

The main objective of UAT is to confirm that the software solves the business problem it was designed to address and supports the expected business processes.

Unlike technical testing, which focuses on whether the software works correctly from a technical perspective, UAT focuses on whether the software works correctly from the business user's perspective.

### **1. Why is UAT important?**

UAT is important because even if an application passes technical testing, it may still fail to meet actual business expectations.

For example, developers may build an order-tracking application that correctly displays shipment statuses. However, the customer-support team may discover that the application does not display the delivery partner's contact information, which they need to resolve customer complaints.

The application may be technically functional, but it does not completely satisfy the business requirement.

UAT helps identify such gaps before the application is released to production.

Its main purposes are:

1. **Business requirement validation:** Confirms that the application meets approved business requirements.
2. **Business process validation:** Ensures that actual business workflows can be completed.
3. **Defect identification:** Identifies issues that affect business operations.
4. **User satisfaction:** Confirms that the solution is usable and meets stakeholder expectations.
5. **Release readiness:** Helps stakeholders decide whether the application is ready for production deployment.

### **2. When is UAT performed?**

UAT is generally performed after the development team has completed the required functionality and the testing team has completed the relevant system and functional testing.

A typical sequence is:

**Requirement Gathering and Analysis**

**Development**

**System and Functional Testing**

**User Acceptance Testing (UAT)**

**Business Approval and Production Deployment**

If UAT identifies defects, they are logged, investigated, corrected, and retested. Depending on their severity and the project's release criteria, UAT may need to be repeated before approval.

### **3. What is the role of a Business Analyst in UAT?**

The BA plays an important coordination and validation role during UAT. The BA ensures that business users understand what needs to be tested and that their feedback is correctly communicated to the technical team.

**A. Preparing UAT test scenarios**

The BA reviews approved business requirements and identifies the real-world business scenarios that users should validate.

Example for an e-commerce order-tracking application:

- Customer should be able to view order status.
- Customer should receive a notification when an order is shipped.
- Customer should see the expected delivery date.
- Customer-support agents should be able to search for an order.
- Cancelled orders should display the correct status.

**B. Preparing UAT test cases**

The BA may prepare or review test cases that explain the steps users need to follow and the expected results.

For example:

**UAT test case: Customer order tracking**

TEST CASE ID

**UAT001**

**Objective**

Verify that a customer can view the current status of an order.

**Precondition**

The customer has an account and has placed an order.

**Test steps**

1. Log in to the application.
2. Navigate to My Orders.
3. Select an existing order.
4. Open the order-tracking section.

**Expected result**

The application should display the correct order status and expected delivery information.

**Actual result**

To be recorded by the UAT tester after execution.

**Status**

Not yet executed

**C. Coordinating with business users**

The BA identifies the appropriate business users who should participate in UAT and coordinates the testing activities.

Responsibilities include:

- Scheduling UAT sessions.
- Explaining the scope of testing.
- Providing test scenarios and instructions.
- Ensuring users have the necessary access and test data.
- Answering business-related questions during testing.

**D. Explaining requirements to testers**

Sometimes users may not understand the expected behaviour of a feature.

The BA clarifies the approved requirements, business rules, and acceptance criteria so that testers can evaluate the functionality correctly.

**E. Logging and tracking defects**

If users identify an issue, the BA helps document it clearly.

A defect report typically contains:

- Defect ID.
- Requirement or test case reference.
- Defect description.
- Steps to reproduce.
- Expected result.
- Actual result.
- Severity and priority.
- Screenshots or supporting evidence, where appropriate.
- Current status and assigned owner.

The BA coordinates with developers and testers to ensure that defects are understood and addressed.

**F. Retesting and validating fixes**

After a defect is fixed, the relevant scenario is tested again.

The BA may coordinate with business users to confirm that the corrected functionality now meets the expected business requirement.

**G. Obtaining UAT sign-off**

Once the required scenarios have passed and outstanding issues have been handled according to the project's acceptance criteria, the BA coordinates business approval.

UAT sign-off indicates that the authorised business stakeholders accept the solution for the agreed scope. It is an important input into the production release decision.

### **5. What happens if UAT fails?**

If UAT fails, the BA should not treat it as an automatic rejection of the entire application. The issue needs to be investigated.

The typical process is:

1. Record the failed scenario and actual result.
2. Compare the result with the approved requirement and acceptance criteria.
3. Determine whether it is a defect, a requirement gap, or a new change request.
4. Discuss the issue with developers, testers, and relevant stakeholders.
5. Get the issue corrected or the proposed change formally assessed.
6. Retest the affected functionality.
7. Obtain approval once the agreed acceptance criteria are satisfied.

**Important distinction:**

- If the system does not behave according to an approved requirement, it is generally a defect.
- If the stakeholder requests functionality that was not part of the approved requirement, it may be a change request.

This distinction is particularly important for a BA because it prevents new scope from being incorrectly treated as a software defect.

## **Possible follow-up interview questions**

**1. Who performs UAT?**

UAT is generally performed by business users, end users, or authorised representatives who understand the actual business processes. The BA coordinates and supports the activity, while QA and development teams provide technical assistance.

**2. What is UAT sign-off?**

UAT sign-off is formal confirmation from authorised business stakeholders that the solution meets the agreed acceptance criteria for the tested scope. It is generally used as an input for the production release decision.

**3. What is the difference between a defect and a change request during UAT?**

A defect occurs when the application does not behave as described in an approved requirement. A change request is raised when a stakeholder wants new or modified functionality that was not included in the approved scope.

**4. What is the difference between a test scenario and a test case?**

A test scenario describes *what needs to be tested* at a high level.

A test case describes *how to test it*, including preconditions, steps, test data, and expected results.

For example:

- Test scenario: Verify customer login.
- Test case: Enter valid email and password, click Login, and verify that the customer reaches the dashboard.

**7. What is the difference between UAT and regression testing?**

UAT checks whether the solution meets business requirements and user expectations. Regression testing checks whether changes or bug fixes have affected previously working functionality.

**8. What is a UAT environment?**

A UAT environment is a separate testing environment that is configured to resemble the production environment as closely as practical. Business users use it to test the application without affecting live production data or operations.

## **Q16. What are the different types of diagrams used by a Business Analyst?**

A Business Analyst uses diagrams and visual models to represent business processes, system behaviour, data movement, and user interactions.

Diagrams help convert complex requirements into a visual format that business stakeholders, developers, testers, and other project members can understand.

They are particularly useful when written requirements alone are not sufficient to explain how a business process works or how a system should behave.

For example, suppose an organisation wants to develop an e-commerce order-management system. A written requirement might say:

"The system should allow customers to place orders, make payments, and track deliveries."

However, this does not fully explain the sequence of activities, the people involved, the data exchanged, or what happens when a payment fails.

A BA can use different diagrams to explain these aspects.

The important diagrams mentioned in the interview guide are:

### **1. Flowchart**

A **flowchart** is a graphical representation of a process using symbols and arrows to show the sequence of activities and decisions.

It helps a BA understand and communicate how a process works from beginning to end.

**Common flowchart symbols:**



**When does a BA use flowcharts?**

- To understand existing business processes.
- To identify unnecessary steps and bottlenecks.
- To explain process changes to stakeholders.
- To identify decision points and exception scenarios.
- To communicate workflows to developers and testers.

### **2. BPMN (Business Process Model and Notation)**

**BPMN** is a standardised graphical notation used to model business processes.

It is more structured than a basic flowchart and is particularly useful for complex processes involving multiple departments, participants, decisions, and parallel activities.

BPMN has four important categories of elements:

- **Flow objects:** Events, activities, and gateways.
- **Connecting objects:** Sequence flows, message flows, and associations.
- **Swimlanes:** Pools and lanes representing participants or responsibilities.
- **Artifacts:** Data objects, groups, and annotations.

**Important BPMN symbols:**

| Symbol            | Meaning                             |
| ----------------- | ----------------------------------- |
| Circle            | Event, such as process start or end |
| Rounded rectangle | Activity or task                    |
| Diamond           | Gateway or decision                 |
| Solid arrow       | Sequence flow                       |
| Dashed arrow      | Message flow                        |

**Example: Order fulfilment process**

A BPMN model can show how an order moves between the customer, payment service, warehouse, and delivery partner.

For example:

1. Customer places an order.
2. Payment service processes the payment.
3. Warehouse receives the confirmed order.
4. Warehouse packs the product.
5. Delivery partner collects the package.
6. Customer receives the delivery.

If payment fails, the process follows an alternative path.

**Why is BPMN useful?**

It helps the BA document complex workflows, clarify department responsibilities, identify process inefficiencies, and communicate standardised process models.

### **3. Use Case Diagram**

A **Use Case Diagram** is a UML diagram that represents the interactions between external users or systems (called actors) and the system being developed.

It focuses on what users can do with the system, rather than explaining the internal technical implementation.

**Main components:**

- **Actor:** A person, role, or external system interacting with the application.
- **Use case:** A function or service provided by the system.
- **System boundary:** Defines what is inside the system.
- **Association:** Shows interaction between an actor and a use case.

**Example: E-commerce application**

Actors:

- Customer
- Admin
- Payment gateway

Use cases:

- Register account
- Log in
- Search products
- Add products to cart
- Place order
- Make payment
- Track order
- Manage products



**When does a BA use it?**

- To identify system users and their interactions.
- To define the functional scope of an application.
- To identify missing functionality.
- To explain system capabilities to stakeholders.
- To support functional requirement documentation.

### **4. Activity Diagram**

An **Activity Diagram** is a UML behavioural diagram that represents the workflow of activities within a system or business process.

It shows the sequence of activities, decisions, parallel actions, and the flow from start to completion.

Although it can look similar to a flowchart, an activity diagram is part of UML and supports modelling more detailed system behaviour.

**Example: Customer login**

1. Customer opens the application.
2. System displays the login page.
3. Customer enters credentials.
4. System validates the credentials.
5. If valid, the system displays the dashboard.
6. If invalid, the system displays an error message.



**When does a BA use it?**

- To explain detailed workflows.
- To document alternative and exception paths.
- To model business activities and system behaviour.
- To help developers and testers understand process logic.

### **5. Data Flow Diagram (DFD)**

A **Data Flow Diagram** represents how data moves through a system.

It focuses on data inputs, processing, storage, and outputs rather than the sequence of business activities.

**Main components:**

1. **External Entity:** A person, organisation, or external system that sends or receives data.
2. **Process:** An activity that transforms or processes data.
3. **Data Store:** A location where information is stored.
4. **Data Flow:** Shows the movement of data between entities, processes, and data stores.

**Example: Online order system**

- Customer submits order details.
- Order processing system receives the information.
- Payment service processes payment details.
- Order database stores order information.
- Notification service sends confirmation to the customer.



**Types of DFD:**

- **Level 0:** Context diagram showing the system as one high-level process and its external interactions.
- **Level 1:** Breaks the main process into major sub-processes.
- **Level 2:** Provides more detailed decomposition of individual processes.

**When does a BA use DFDs?**

- To understand data movement.
- To identify where data originates and where it is stored.
- To understand system integrations.
- To identify missing data inputs or outputs.
- To communicate data requirements to technical teams.

### **6. Entity Relationship Diagram (ERD)**

An **Entity Relationship Diagram** represents the structure of a database and the relationships between different data entities.

It helps a BA understand what information needs to be stored and how different data objects are connected.

**Main components:**

- **Entity:** An object about which data is stored, such as Customer or Order.
- **Attribute:** A property of an entity, such as Customer ID or Order Date.
- **Primary Key:** A field that uniquely identifies each record.
- **Foreign Key:** A field that links one table to another.
- **Relationship:** Describes how entities are associated.

**Example: E-commerce database**

Consider three entities:

| Customer         | Order            | Product         |
| ---------------- | ---------------- | --------------- |
| Customer_ID (PK) | Order_ID (PK)    | Product_ID (PK) |
| Customer_Name    | Customer_ID (FK) | Product_Name    |
| Email            | Order_Date       | Price           |
| Phone            | Total_Amount     | Category        |

Relationships:

- One customer can place multiple orders.
- Each order belongs to a customer.
- An order can contain multiple products, and a product can appear in multiple orders. This is usually represented through an additional Order_Details table.



**When does a BA use ERDs?**

- To understand database requirements.
- To discuss data structures with developers.
- To identify relationships between business entities.
- To support data migration and reporting requirements.
- To identify duplicate or unnecessary data storage.

### **7. Wireframes and Mockups**

Wireframes and mockups are visual representations of a proposed user interface.

They help stakeholders understand what an application screen may look like before developers build it.

**Wireframe:**

A wireframe is a basic structural layout of a screen. It focuses on placement and functionality rather than detailed visual styling.

For example, an e-commerce product page wireframe may contain:

- Product image placeholder.
- Product name.
- Price.
- Quantity selector.
- Add to Cart button.
- Product description.

**Mockup:**

A mockup is a more detailed visual representation that includes design elements such as colours, fonts, images, spacing, and branding.


**Wireframe**

Basic layout and structure.

Detailed visual design.

**When does a BA use them?**

- To clarify user-interface requirements.
- To collect feedback before development.
- To identify missing fields or buttons.
- To align stakeholders and designers.
- To reduce misunderstandings between business and development teams.

### **8. Swimlane Diagram**

A **Swimlane Diagram** is a process flow diagram that separates activities according to the person, team, or department responsible for performing them.

It helps identify responsibilities, handoffs, delays, and communication gaps between teams.

**Example: Order fulfilment process**

An e-commerce order may involve:

- Customer
- Sales team
- Finance team
- Warehouse team
- Delivery team

Each department has its own lane showing the activities it performs.



**When does a BA use swimlane diagrams?**

- When a process involves multiple departments.
- To identify who is responsible for each activity.
- To identify delays during handoffs.
- To clarify accountability.
- To redesign inefficient business processes.

5. Data Flow Diagrams: Used to show how data moves between external entities, processes, and data stores.
6. Entity Relationship Diagrams: Used to represent database entities, attributes, and relationships between tables.
7. Wireframes and Mockups: Used to represent proposed user-interface layouts and gather stakeholder feedback before development.
8. Swimlane Diagrams: Used to represent process activities according to the responsible departments or individuals.

For example, if I am working on an e-commerce order-management project, I may use a flowchart to explain the order process, a swimlane diagram to show responsibilities across sales, finance, warehouse, and delivery teams, a use case diagram to identify customer and administrator interactions, and an ERD to understand the relationship between customers, orders, and products.

I have also come across tools such as Microsoft Visio, Lucidchart, draw\.io, and Balsamiq for creating these diagrams.

The choice of diagram depends on the business problem, the type of requirement, and the audience I need to communicate with.

## **Possible follow-up interview questions**

**1. What is the difference between a flowchart and a BPMN diagram?**

A flowchart is a general-purpose diagram used to represent process steps and decisions. BPMN is a standardised notation specifically designed for business-process modelling. BPMN provides more formal elements for representing events, activities, gateways, participants, and process interactions.

**2. What is the difference between a use case diagram and an activity diagram?**

A use case diagram shows what functionality the system provides and which actors interact with it. An activity diagram shows the sequence of activities and decisions involved in completing a workflow.

**3. What is the difference between a DFD and an ERD?**

A DFD explains how data moves through a system, including inputs, processing, and outputs. An ERD explains how data is structured and how different entities are related in a database.

**4. What is the difference between a wireframe and a mockup?**

A wireframe represents the basic structure and placement of interface elements. A mockup represents a more detailed visual design, including colours, typography, images, and styling.

**5. Which diagram would you use to explain a process involving multiple departments?**

I would use a swimlane diagram because it separates activities according to the responsible department or role. It makes handoffs, responsibilities, and process delays easier to identify.

**6. Which diagram would you use to explain how data moves from a customer to a database?**

I would use a Data Flow Diagram because it shows the movement of data between external entities, system processes, and data stores.

**7. Which diagram would you use to represent a database structure?**

I would use an Entity Relationship Diagram because it represents entities, attributes, primary keys, foreign keys, and relationships between tables.

**8. Does a Business Analyst need to create all these diagrams in every project?**

No. The BA selects diagrams based on project complexity and stakeholder requirements. A simple process may only need a flowchart, whereas a complex system may require BPMN, use case diagrams, DFDs, ERDs, and wireframes.

**9. What is AS-IS and TO-BE process modelling?**

An AS-IS diagram represents how a business process currently operates. A TO-BE diagram represents the proposed future process after improvements or changes.

For example, the AS-IS process may involve employees manually entering customer orders into a spreadsheet. The TO-BE process may introduce an automated order-management system.

**10. What would you do if stakeholders cannot understand a technical diagram?**

I would simplify the diagram, remove unnecessary technical details, use familiar business terminology, and explain the process using a practical example. The purpose of a diagram is to improve understanding, not make the requirements more complicated.

## **Q17. What is a Use Case?**

A **Use Case** is a description of how an actor (a user or another system) interacts with an application to achieve a specific goal.

It explains what a user wants to accomplish using the system and how the system should respond to the user's actions.

A Business Analyst uses use cases to understand and document system functionality from the user's perspective. They help developers understand expected system behaviour and help testers identify scenarios that need to be validated.

**Example: Online Shopping Application**

Consider an e-commerce application where a customer wants to place an order.

The customer interacts with the system to:

1. Log in.
2. Search for a product.
3. Add the product to the cart.
4. Enter delivery details.
5. Make payment.
6. Receive order confirmation.

This interaction can be documented as a use case called **Place an Order**.

### **1. Main components of a Use Case**

According to the interview guide, a use case contains six important components:

#### **A. Actor**

An actor is a person, user role, or external system that interacts with the application to achieve a goal.

Actors are outside the system boundary.

There are two common categories:

- **Primary actor:** Initiates the use case to achieve a goal.
- **Secondary actor:** Supports the system in completing the use case.

**Example: Online shopping application**

| Actor            | Responsibility                          |
| ---------------- | --------------------------------------- |
| Customer         | Searches for products and places orders |
| Admin            | Manages products and customer orders    |
| Payment Gateway  | Processes online payments               |
| Delivery Partner | Updates shipment and delivery status    |

In the "Place an Order" use case, the customer is the primary actor, while the payment gateway is a supporting actor.

#### **B. Preconditions**

Preconditions are the conditions that must be satisfied before a use case can begin.

They define what must already be true for the system to execute the requested functionality.

**Example: Place an Order**

Possible preconditions:

- The customer must be registered.
- The customer must be logged in.
- The product must be available in stock.
- The customer must have a valid delivery address.

If these conditions are not satisfied, the system may prevent the customer from proceeding or guide them to complete the required steps.

#### **C. Main Flow (Basic Flow)**

The main flow describes the normal sequence of interactions between the actor and the system when everything works as expected.

It is also called the *happy path*.

**Example: Customer places an order**

| Step | Actor action                      | System response                   |
| ---- | --------------------------------- | --------------------------------- |
| 1    | Customer logs in                  | System validates credentials      |
| 2    | Customer searches for a product   | System displays matching products |
| 3    | Customer selects a product        | System displays product details   |
| 4    | Customer adds product to cart     | System updates the cart           |
| 5    | Customer enters delivery details  | System validates the address      |
| 6    | Customer selects payment method   | System initiates payment          |
| 7    | Customer completes payment        | System confirms the transaction   |
| 8    | Customer views order confirmation | System generates an order ID      |

This is the expected successful sequence of events.

#### **D. Alternate Flows**

Alternate flows describe different valid paths through a use case when the actor makes a different choice or when an optional condition occurs.

They are not necessarily errors. They are alternative ways of completing or continuing the process.

**Example: Place an Order**

- Customer chooses cash on delivery instead of online payment.
- Customer applies a valid discount coupon.
- Customer changes the delivery address before confirming the order.
- Customer removes a product from the cart before checkout.

These scenarios should be documented because the system must handle them correctly.

#### **E. Exceptions**

Exceptions describe situations where the normal process cannot continue because something goes wrong or a required condition is not met.

**Example: Place an Order**

- Payment fails because the bank declines the transaction.
- Product becomes out of stock during checkout.
- The delivery address is invalid.
- The payment gateway does not respond.
- The system encounters a technical error.

The BA documents the expected system response for each exception.

For example, if payment fails, the system should display an appropriate error message, avoid creating a confirmed paid order, and allow the customer to retry or choose another payment method, according to the business rules.

#### **F. Postconditions**

Postconditions describe the expected state of the system after the use case finishes, whether successfully or unsuccessfully.

They explain what should be true at the end of the interaction.

**Example: Successful order placement**

- A unique order ID is generated.
- The order is saved in the database.
- Payment status is recorded.
- Order status is updated to "Order Placed".
- Confirmation is displayed to the customer.

**Example: Failed payment**

- The payment status is recorded as failed or pending, depending on the payment response.
- The order is not treated as successfully paid.
- The customer receives an appropriate message.

### **3. Use Case vs Use Case Diagram**

This is an important distinction for interviews.

| Use Case                                                                           | Use Case Diagram                                                        |
| ---------------------------------------------------------------------------------- | ----------------------------------------------------------------------- |
| Describes how an actor interacts with a system to achieve a goal                   | Visually represents actors and their interactions with system functions |
| Contains preconditions, main flow, alternate flows, exceptions, and postconditions | Contains actors, use cases, system boundary, and relationships          |
| Provides detailed functional behaviour                                             | Provides a high-level overview of system functionality                  |
| Usually documented in text or a structured template                                | Represented graphically using UML notation                              |

For example, "Place an Order" is a use case. A diagram showing Customer connected to Place Order, Make Payment, and Track Order is a use case diagram.

### **4. Use Case vs User Story**

These two concepts are closely related but serve different purposes.

| Use Case                                              | User Story                                                |
| ----------------------------------------------------- | --------------------------------------------------------- |
| Describes interaction between an actor and the system | Describes a requirement from the user's perspective       |
| Usually provides detailed interaction flows           | Usually concise                                           |
| Includes main, alternate, and exception flows         | Includes acceptance criteria to define expected behaviour |
| Common in detailed functional analysis                | Common in Agile development                               |

**Example of a user story:**

"As a customer, I want to place an order online so that I can purchase products without visiting a physical store."

**Example of the corresponding use case:**

"Place an Order" with its actor, preconditions, main flow, alternate flows, exceptions, and postconditions.

A user story expresses the need, while a use case describes the interaction in greater detail.

### **5. How does a Business Analyst create a Use Case?**

A BA generally follows these steps:

1. **Identify the business goal:** Understand what the user wants to achieve.
2. **Identify actors:** Determine which users and external systems interact with the application.
3. **Define the system boundary:** Clarify which functionality belongs to the system being analysed.
4. **Define preconditions:** Identify what must be true before the interaction begins.
5. **Document the main flow:** Describe the normal sequence of actor and system interactions.
6. **Identify alternate flows:** Consider other valid ways to complete the process.
7. **Identify exceptions:** Document errors, failures, and unusual conditions.
8. **Define postconditions:** Specify the expected outcome after the interaction.
9. **Review with stakeholders:** Validate the use case with business users and the technical team.

### **6. Why are Use Cases important?**

Use cases provide several benefits:

- **Clear requirements:** They explain system functionality from the user's perspective.
- **Better stakeholder communication:** Business and technical teams can discuss the same expected behaviour.
- **Identification of missing scenarios:** Alternate flows and exceptions help uncover requirements that might otherwise be overlooked.
- **Testing support:** QA teams can derive test scenarios and test cases from use case flows.
- **Scope definition:** They help identify which functions are included in the system.
- **Reduced ambiguity:** Detailed interactions and expected outcomes make requirements easier to understand.

## **Possible follow-up interview questions**

**1. What is the difference between an actor and a user?**

A user is generally a human interacting with a system. An actor is a broader UML concept that can represent a human user, an external application, or another system interacting with the system being modelled.

For example, a customer is an actor, and an external payment gateway is also an actor.

**2. Can a use case have multiple actors?**

Yes. A use case can involve multiple actors. For example, in an online payment use case, the customer initiates the payment and the payment gateway processes the transaction.

**3. What is the difference between a main flow and an alternate flow?**

The main flow describes the normal successful sequence of events. An alternate flow describes another valid path through the process, such as choosing cash on delivery instead of online payment.

An alternate flow represents a valid variation in the process. An exception flow represents an error or an unexpected condition that prevents the normal process from continuing.

- Alternate flow: Customer chooses cash on delivery.
- Exception flow: Payment gateway fails to process a transaction.

**5. Can a use case be used to prepare test cases?**

Yes. Testers can derive test cases from the main flow, alternate flows, and exception flows. Each important interaction and expected outcome can be tested to verify that the application behaves correctly.

No. A functional requirement states what the system must do. A use case describes how an actor interacts with the system to achieve a goal. A single use case can help elaborate several functional requirements.




## **Q18. What is Stakeholder Analysis?**

**Stakeholder analysis** is the process of identifying all the people, groups, or organisations that can affect a project or are affected by its outcome, and understanding their interests, influence, expectations, and involvement.

A Business Analyst uses stakeholder analysis to determine:

- Who needs to be involved in the project.
- Who can make or influence important decisions.
- Who needs regular updates.
- Whose requirements and expectations must be understood.
- How to manage different stakeholder interests and potential conflicts.

It is important because a project can have technically correct requirements but still fail if important stakeholders are ignored or their expectations are not managed.

### **1. Why is Stakeholder Analysis important?**

Every project involves people with different responsibilities, expectations, and levels of decision-making authority.

For example, when developing an e-commerce order-tracking system:

- Customers want accurate order status and delivery updates.
- Customer-support teams want quick access to order information.
- Logistics teams want accurate shipment information.
- Developers want clear and technically understandable requirements.
- Management wants reduced customer complaints and improved operational efficiency.

If the BA gathers requirements only from management, important operational requirements from customer-support or logistics teams may be missed.

Stakeholder analysis helps prevent this by identifying relevant people early and planning how to involve each of them.

### **2. How does a Business Analyst perform Stakeholder Analysis?**

A BA generally follows these steps:

**Step 1: Identify stakeholders**

Identify everyone who can influence the project or be affected by its outcome.

Stakeholders may be internal or external.

| Internal stakeholders | External stakeholders      |
| --------------------- | -------------------------- |
| Project Manager       | Customers                  |
| Business users        | Vendors                    |
| Developers            | Suppliers                  |
| QA testers            | External service providers |
| Senior management     | Regulatory authorities     |
| Product Owner         | Business partners          |

For an e-commerce project, the BA may identify customers, customer-support representatives, logistics managers, finance teams, developers, QA testers, and senior management.

**Step 2: Understand stakeholder roles and expectations**

After identifying stakeholders, the BA understands:

- What is their responsibility?
- What do they expect from the project?
- What information or functionality do they need?
- How will the project affect their work?
- Can they approve requirements or make decisions?
- How frequently should they be involved?

For example:

| Stakeholder           | Main expectation                              |
| --------------------- | --------------------------------------------- |
| Customer              | Accurate delivery updates                     |
| Customer-support team | Access to order and shipment details          |
| Logistics team        | Ability to update shipment status             |
| Management            | Reduced support costs and complaints          |
| Development team      | Clear and feasible requirements               |
| QA team               | Testable requirements and acceptance criteria |

**Step 3: Analyse stakeholders using the Power-Interest Matrix**

This is one of the most important concepts in the interview guide.

The Power-Interest Matrix classifies stakeholders based on two dimensions:

- **Power/Influence:** How much authority or influence the stakeholder has over project decisions.
- **Interest:** How much the stakeholder cares about the project and its outcome.

It creates four categories.

## **Power–Interest Matrix**

Illustrative stakeholder classification

**Manage Closely**

High power · High interest

Project sponsor, project manager

Involve in key decisions and maintain regular communication.

**Keep Satisfied**

High power · Low interest

Senior management with limited day-to-day involvement

Provide important updates without unnecessary detail.

**Keep Informed**

Low power · High interest

Customer-support representatives, end users

Share progress and collect feedback regularly.

**Monitor**

Low power · Low interest

People with limited involvement in the project

Monitor their involvement and provide updates when necessary.

Power increases vertically; interest increases horizontally. Actual stakeholder placement depends on the project.

Let's understand each category in more detail.

**A. Manage Closely — High Power, High Interest**

These stakeholders have significant decision-making authority and are highly interested in the project.

Examples:

- Project sponsor.
- Project manager.
- Product Owner.

BA's approach:

- Involve them in important discussions.
- Conduct regular status meetings.
- Obtain approvals and decisions.
- Escalate significant risks and issues.
- Keep them informed about progress and changes.

**B. Keep Satisfied — High Power, Low Interest**

These stakeholders have considerable authority but may not be involved in daily project activities.

Examples:

- Senior executives.
- Department heads who are not directly involved in implementation.

BA's approach:

- Provide concise project updates.
- Communicate major risks and business impacts.
- Seek decisions when required.
- Avoid overwhelming them with unnecessary operational details.

**C. Keep Informed — Low Power, High Interest**

These stakeholders may not have major decision-making authority, but they are directly affected by the project.

Examples:

- Customer-support representatives.
- Operational employees.
- End users.

BA's approach:

- Conduct requirement-gathering sessions.
- Collect feedback regularly.
- Explain process changes.
- Share relevant progress updates.
- Involve them in UAT when appropriate.

**D. Monitor — Low Power, Low Interest**

These stakeholders have limited influence and limited day-to-day interest in the project.

BA's approach:

- Monitor their involvement.
- Provide relevant updates when necessary.
- Avoid spending excessive time on unnecessary communication.

**Step 4: Plan stakeholder communication**

After classifying stakeholders, the BA creates a communication approach.

For example:

| Stakeholder       | Communication method                  | Frequency          |
| ----------------- | ------------------------------------- | ------------------ |
| Project sponsor   | Status report and review meeting      | Weekly             |
| Project manager   | Project coordination meeting          | Daily or as needed |
| Business users    | Workshops and requirement reviews     | As required        |
| Developers        | Requirement clarification sessions    | During development |
| QA testers        | Requirement and test-case discussions | During testing     |
| Senior management | Summary report                        | At key milestones  |

The frequency should be based on the project's needs and stakeholder expectations.

**Step 5: Review and update stakeholder analysis**

Stakeholder involvement can change during a project.

For example, a finance manager may have limited involvement during initial requirements but become highly involved when payment reconciliation functionality is being developed.

Therefore, the BA should review stakeholder roles, influence, interest, and communication needs throughout the project.

### **3. What is a RACI Matrix?**

The interview guide also mentions RACI as a tool used to clarify stakeholder roles and responsibilities.

**RACI stands for:**

- **R – Responsible:** The person who performs the actual task.
- **A – Accountable:** The person ultimately answerable for the task's completion and outcome.
- **C – Consulted:** People whose opinions or expertise are required before a decision or action.
- **I – Informed:** People who need to receive updates about progress or decisions.

A RACI matrix prevents confusion about who is doing the work, who owns the decision, who needs to provide input, and who should receive updates.

#### **Example: RACI Matrix for an e-commerce project**

| Project activity              | BA  | Project Manager | Developer | QA Tester | Business Owner |
| ----------------------------- | --- | --------------- | --------- | --------- | -------------- |
| Gather requirements           | R   | I               | C         | I         | A              |
| Document requirements         | R/A | I               | C         | C         | C              |
| Develop application           | C   | I               | R/A       | I         | I              |
| Perform testing               | C   | I               | C         | R/A       | I              |
| Approve business requirements | R   | I               | I         | C         | A              |
| UAT sign-off                  | R   | I               | C         | C         | A              |

This is an illustrative allocation. Actual RACI assignments depend on the organisation and project governance.

**Important RACI distinction:**

- Responsible = Does the work.
- Accountable = Owns the outcome.
- Consulted = Provides input.
- Informed = Receives information.

A task can have multiple responsible people, but it is generally best to have one clearly identified accountable owner.

### **4. Difference between Stakeholder Analysis and RACI**

| Stakeholder Analysis                                     | RACI Matrix                                                        |
| -------------------------------------------------------- | ------------------------------------------------------------------ |
| Identifies stakeholders and their influence and interest | Defines responsibilities for specific activities                   |
| Focuses on stakeholder engagement                        | Focuses on task ownership and accountability                       |
| Helps decide how to communicate with stakeholders        | Helps decide who performs, approves, advises, and receives updates |
| Uses Power-Interest Matrix                               | Uses Responsible, Accountable, Consulted, Informed                 |

**Example:**

Stakeholder analysis tells the BA that the project sponsor has high power and high interest, so the sponsor should be managed closely.

The RACI matrix clarifies whether that sponsor is accountable for approving the project requirements or simply needs to be informed.

They complement each other but solve different problems.

### **5. What challenges can occur during Stakeholder Analysis?**

A BA may encounter several challenges:

**Conflicting expectations:** Different stakeholders may request different solutions. The BA should understand the reasons behind each request, compare business value and impact, facilitate discussion, and use prioritisation techniques such as MoSCoW.

**Unresponsive stakeholders:** The BA should understand why the stakeholder is unavailable, arrange a convenient meeting, send specific questions, or identify an appropriate delegate. If the delay affects the project, the BA should escalate it to the project manager.

**Missing stakeholders:** Important users may be overlooked during initial planning. The BA should review business processes and consult existing stakeholders to identify other affected teams.

**Changing stakeholder influence:** A stakeholder's importance may increase or decrease as the project progresses. The BA should periodically update the Power-Interest Matrix.

**Communication gaps:** Different stakeholders may need different levels of detail. The BA should adapt communication to the audience, using detailed requirement documents for technical teams and concise business summaries for management.

## **Q19. What is Scope Creep, and how do you control it?**

**Scope creep** refers to the uncontrolled expansion of a project's requirements or deliverables after the project scope has been agreed upon, usually without properly evaluating or approving the impact of those changes.

It happens when new features, functionalities, or activities are continuously added to a project without corresponding adjustments to the timeline, budget, resources, or priorities.

A Business Analyst must manage scope carefully because uncontrolled changes can delay delivery, increase costs, create confusion, and affect the quality of the final solution.

### **1. Example of Scope Creep**

Suppose an organisation starts developing an e-commerce application.

The initially approved project scope includes:

- Customer registration and login.
- Product search.
- Shopping cart.
- Online payment.
- Order tracking.

The project team estimates that these features can be delivered within three months.

However, during development, stakeholders begin requesting additional features:

- Wishlist functionality.
- Product recommendations.
- Live chat support.
- Loyalty and reward points.
- Multiple payment gateways.
- AI-based product suggestions.

If the team keeps accepting these requests without evaluating their impact on the original timeline, budget, resources, and priorities, the project experiences scope creep.

**Original approved scope**

Login · Product Search · Cart · Payment · Order Tracking

Estimated timeline: 3 months

**Uncontrolled additional requests**

Wishlist · Live Chat · Loyalty Points · AI Recommendations

Added without formal impact assessment or approval

**Potential project consequences**

- Delivery delays
- Increased development costs
- Resource constraints
- Testing complexity
- Risk of missing the original deadline

The important point is that adding a new feature is not automatically scope creep. It becomes scope creep when requirements expand without appropriate evaluation, approval, and control.

### **2. What are the main causes of Scope Creep?**

Understanding the causes helps a BA prevent scope creep rather than just react to it.

**A. Unclear initial requirements**

If requirements are not properly understood or documented, stakeholders may keep requesting additional functionality during development.

Example: A stakeholder initially asks for a sales dashboard but does not specify filters, export functionality, or reporting frequency. Later, these become additional requests.

**B. Frequent stakeholder changes**

Different stakeholders may continuously introduce new expectations after the project has started.

Example: The sales manager requests additional sales reports, while the finance manager requests new financial metrics.

**C. Lack of proper scope definition**

If the project does not clearly define what is included and excluded, stakeholders may assume that additional features are part of the original agreement.

**D. No formal change-control process**

When developers or project teams accept new requests informally, the project can expand without anyone assessing the impact.

**E. Poor stakeholder communication**

If stakeholders do not understand project limitations, timelines, or priorities, they may continue requesting features that cannot realistically be delivered within the agreed constraints.

**F. Gold plating**

Gold plating means adding extra features or improvements that were not requested or approved, often because the team believes they will make the product better.

Example: A developer adds an advanced reporting feature even though the approved requirement only asks for a basic monthly report.

### **3. How does a Business Analyst control Scope Creep?**

This is the most important part of the interview question.

A BA uses a structured scope-management and change-control process.

#### **Step 1: Define and document the project scope**

At the beginning of the project, the BA works with stakeholders to define:

- Project objectives.
- Project deliverables.
- In-scope requirements.
- Out-of-scope requirements.
- Assumptions and constraints.
- Acceptance criteria.

The scope should be reviewed and formally approved by the authorised stakeholders.

**Example:**

For an e-commerce application:

| In scope       | Out of scope               |
| -------------- | -------------------------- |
| Customer login | AI product recommendations |
| Product search | Loyalty programme          |
| Shopping cart  | Live chat support          |
| Online payment | Advanced personalisation   |
| Order tracking | Voice-based shopping       |

This creates a clear reference point when new requests arise.

#### **Step 2: Establish a formal Change Request Process**

If a stakeholder requests a new feature, the BA should not immediately accept or reject it.

Instead, the BA records it as a **Change Request (CR)**.

A change request typically contains:

- Change Request ID.
- Description of the requested change.
- Reason or business justification.
- Requesting stakeholder.
- Expected business benefit.
- Priority.
- Potential impact.
- Approval status.

**Example:**

The sales manager requests a wishlist feature after the e-commerce application has entered development.

The BA records the request instead of directly asking developers to implement it.

#### **Step 3: Perform Impact Analysis**

The BA works with the project manager, developers, testers, and other relevant stakeholders to understand how the requested change affects the project.

The main areas to assess are:

**1. Time impact**

How much additional time is required?

Example: Adding wishlist functionality may require an additional two weeks.

**2. Cost impact**

Will the change increase development, licensing, infrastructure, or operational costs?

**3. Effort and resource impact**

Will additional developers, testers, designers, or business users be needed?

**4. Technical impact**

Will the change affect existing system functionality, architecture, integrations, or databases?

**5. Dependency impact**

Does the new feature depend on another feature or requirement being completed first?

**6. Testing impact**

Will additional test cases, regression testing, or UAT activities be required?

The BA documents these findings so stakeholders can make an informed decision.

#### **Step 4: Prioritise the Change Request**

Not every requested change needs to be implemented immediately.

The BA can use MoSCoW prioritisation:

- Must Have
- Should Have
- Could Have
- Won't Have (This Time)

For example, if the wishlist feature is useful but not essential for the initial release, it may be moved to a future release.

The BA should consider business value, urgency, effort, and project constraints when facilitating prioritisation.

#### **Step 5: Obtain Formal Approval**

After impact analysis and prioritisation, the BA presents the available options to the authorised decision-maker.

For example:

| Option                               | Possible outcome                            |
| ------------------------------------ | ------------------------------------------- |
| Add the feature now                  | Delivery date may move from 3 to 3.5 months |
| Replace a lower-priority feature     | Original delivery date may be maintained    |
| Move the feature to the next release | Current release remains unchanged           |

The project sponsor, Product Owner, or designated change authority decides how to proceed, according to project governance.

The BA should not independently approve changes outside their authority.

#### **Step 6: Update Project Documentation**

Once the change is approved, the BA updates the relevant documents.

These may include:

- BRD or FRD.
- User stories and acceptance criteria.
- Project scope documentation.
- Requirements Traceability Matrix (RTM).
- Process flow diagrams.
- Test scenarios and test cases.

The BA also communicates the approved change to the affected stakeholders.

#### **Step 7: Monitor Scope Throughout the Project**

The BA regularly compares ongoing work against the approved scope.

If new requests arise, they should follow the same change-control process.

This helps prevent informal additions and ensures that everyone works toward the agreed project objectives.

### **4. Difference between Scope Creep and a Change Request**

These concepts are related but not the same.

| Scope Creep                                    | Change Request                                |
| ---------------------------------------------- | --------------------------------------------- |
| Uncontrolled expansion of project scope        | Formal request to modify the approved scope   |
| Often happens without proper approval          | Recorded and evaluated through change control |
| Can cause unexpected delays and cost increases | Helps assess and manage the impact of changes |
| Indicates weak scope management                | Is a normal project-management activity       |

**Example:**

A stakeholder asks for an additional payment option.

- If the development team adds it informally without evaluating the impact, it contributes to scope creep.
- If the BA records the request, assesses the impact, obtains approval, and updates the documentation, it is a controlled scope change.

### **5. Is every scope change bad?**

No. Scope changes can be beneficial and sometimes necessary.

For example:

- A new regulatory requirement may need to be included.
- Customer feedback may reveal an important missing feature.
- A business priority may change.
- A security requirement may need to be strengthened.

The goal of scope management is not to reject every new requirement. It is to ensure that changes are properly evaluated, prioritised, approved, and documented.

## **Q20. What is a KPI? Give examples.**

**KPI stands for Key Performance Indicator.** A KPI is a measurable value used to evaluate how effectively an individual, team, department, or organisation is achieving a specific business objective.

In simple terms, a KPI helps an organisation understand whether it is achieving its goals and where improvement is needed.

For a Business Analyst, KPIs are important because they help measure business performance, identify problems, evaluate process improvements, and determine whether a project has delivered the expected business value.

### **1. Why are KPIs important?**

Organisations set business objectives, but objectives alone do not show whether progress is being made.

KPIs convert those objectives into measurable indicators.

For example:

**Business objective:** Improve customer satisfaction.

A company cannot evaluate this objective simply by saying that customer satisfaction should improve. It needs measurable indicators, such as:

- Customer satisfaction score.
- Customer complaint rate.
- Average ticket resolution time.
- Customer retention rate.

By monitoring these indicators over time, the organisation can evaluate whether its actions are producing the desired results.

The main purposes of KPIs are:

1. **Measure performance:** Evaluate how well a business, team, or process is performing.
2. **Track progress:** Determine whether business objectives are being achieved.
3. **Identify problems:** Highlight areas where performance is below expectations.
4. **Support decision-making:** Help management make decisions using measurable information.
5. **Evaluate improvements:** Measure whether a new system or process has delivered the expected benefits.

### **2. Important types of KPIs with examples**

Your interview guide mentions five important business KPIs: monthly revenue, customer churn rate, average order value, on-time delivery percentage, and ticket resolution time.

Let's understand each one in detail.

#### **A. Monthly Revenue**

**Meaning:** Monthly revenue measures the total income generated by a business through its sales during a particular month.

It helps management understand sales performance and identify revenue trends.

**Formula:**

\\[ \text{Monthly Revenue}=\sum \text{Sales Amount} \\]

**Example:**

Suppose an e-commerce company records the following monthly sales:

| Month    | Revenue    |
| -------- | ---------- |
| January  | ₹10,00,000 |
| February | ₹12,00,000 |
| March    | ₹9,00,000  |
| April    | ₹14,00,000 |

Management can use this KPI to identify sales growth or decline.

For example, revenue increased from ₹10 lakh in January to ₹14 lakh in April.

A BA may investigate which products, regions, sales channels, or customer segments contributed to this increase.

**Business use:** Sales performance monitoring, revenue forecasting, and business planning.

#### **B. Customer Churn Rate**

**Meaning:** Customer churn rate measures the percentage of customers who stop using a company's product or service during a particular period.

It is especially important for subscription-based businesses, telecom companies, banking services, and online platforms.

**Formula:**

\\[ \text{Churn Rate} = \frac{\text{Customers Lost During Period}} {\text{Customers at Start of Period}}\times100 \\]

**Example:**

Suppose a subscription company starts the month with 1,000 customers.

During the month, 50 customers cancel their subscriptions.

\\[ \text{Churn Rate}=\frac{50}{1000}\times100=5\\% \\]

5% churn rate

This means the company lost 5% of its starting customer base during that month.

**Business use:** Understanding customer retention, identifying dissatisfaction, and evaluating customer-retention strategies.

A BA may analyse churn by customer segment, subscription type, region, or length of customer relationship to identify where customers are leaving.

#### **C. Average Order Value (AOV)**

**Meaning:** Average Order Value measures the average amount customers spend per order.

It is commonly used by e-commerce and retail businesses.

**Formula:**

\\[ \text{AOV}= \frac{\text{Total Revenue}}{\text{Number of Orders}} \\]

**Example:**

Suppose an online shopping company generates ₹5,00,000 in revenue from 1,000 orders.

\\[ \text{AOV}=\frac{500000}{1000}=₹500 \\]

₹500 average order value

This means customers spend an average of ₹500 per order.

**Business use:** Evaluating purchasing behaviour, promotional campaigns, and revenue opportunities.

A BA may analyse AOV by product category, customer segment, region, or marketing channel.

#### **D. On-Time Delivery Percentage**

**Meaning:** On-time delivery percentage measures the proportion of orders delivered within the promised delivery timeframe.

It is particularly important for e-commerce, logistics, supply chain, and transportation businesses.

**Formula:**

\\[ \text{On-Time Delivery \\%}= \frac{\text{Orders Delivered on Time}} {\text{Total Orders Delivered}}\times100 \\]

**Example:**

Suppose an e-commerce company delivers 950 orders on time out of 1,000 completed deliveries.

\\[ \text{On-Time Delivery}= \frac{950}{1000}\times100=95\\% \\]

95% on-time delivery

This means 95% of the delivered orders reached customers within the promised timeframe.

**Business use:** Monitoring logistics performance, delivery-partner efficiency, and customer experience.

A BA may investigate late deliveries by region, delivery partner, warehouse, or product category to identify operational problems.

#### **E. Ticket Resolution Time**

**Meaning:** Ticket resolution time measures how long it takes a customer-support team to resolve a customer complaint or service request.

It is commonly used in customer service, IT support, and service-management operations.

**Formula:**

\\[ \text{Average Resolution Time}= \frac{\text{Total Resolution Time of Resolved Tickets}} {\text{Number of Resolved Tickets}} \\]

**Example:**

Suppose a customer-support team resolves four tickets in the following time:

| Ticket    | Resolution time |
| --------- | --------------- |
| Ticket 1  | 2 hours         |
| Ticket 2  | 3 hours         |
| Ticket 3  | 4 hours         |
| Ticket 4  | 3 hours         |
| **Total** | **12 hours**    |

\\[ \text{Average}=\frac{12}{4}=3\text{ hours} \\]

The average ticket resolution time is three hours.

**Business use:** Measuring support-team efficiency, identifying service delays, and improving customer satisfaction.

A BA may analyse resolution time by ticket category, support team, priority, or issue type.

### **Q1. What is VLOOKUP and what are its limitations?**

**VLOOKUP (Vertical Lookup)** is an Excel function used to search for a specific value in the first column of a table or data range and return a related value from another column in the same row.

Business Analysts commonly use VLOOKUP to match records across datasets, retrieve customer or employee details, and reconcile reports.

### **1. Syntax**

```
=VLOOKUP(lookup_value, table_array, col_index_num, FALSE)
```

Each argument has a specific purpose:

| Argument        | Meaning                                                     |
| --------------- | ----------------------------------------------------------- |
| `lookup_value`  | The value you want Excel to find                            |
| `table_array`   | The range containing the data                               |
| `col_index_num` | The column number from which Excel should return the result |
| `FALSE`         | Requests an exact match                                     |

The final argument can be `FALSE` or `TRUE`:

- **FALSE:** Finds an exact match. This is generally preferred for IDs, names and transaction references.
- **TRUE:** Finds an approximate match. The first column should be sorted in ascending order for reliable results.

### **2. Practical example**

Suppose you have an employee dataset:

| Employee ID | Employee Name | Department |
| ----------- | ------------- | ---------- |
| E101        | Rahul         | Finance    |
| E102        | Priya         | HR         |
| E103        | Arjun         | IT         |
| E104        | Neha          | Sales      |

You want to find the employee's name using their Employee ID.

Assume the ID to search is entered in cell `E2`.

**Formula to retrieve the employee name:**

```
=VLOOKUP(E2,A2:C5,2,FALSE)
```

Explanation:

- `E2` contains the Employee ID to search for.
- `A2:C5` is the employee data range.
- `2` tells Excel to return the value from the second column.
- `FALSE` ensures an exact match.

If `E2` contains `E102`, the result is **Priya**.

**Formula to retrieve the department:**

```
=VLOOKUP(E2,A2:C5,3,FALSE)
```

The result is **HR**.

The important point is that VLOOKUP searches for the Employee ID in the first column of the selected range and returns information from another column in that row.

### **3. Limitations of VLOOKUP**

There are three main limitations highlighted in your interview guide.

**Limitation 1: It can only look to the right**

VLOOKUP searches in the first column of the selected range and returns values from columns to its right. It cannot directly return a value from a column to the left.

For example, if Employee Name is in column A and Employee ID is in column B, VLOOKUP cannot search for the ID in column B and return the name from column A using the normal table arrangement.

**Limitation 2: It can break when columns are inserted**

VLOOKUP uses a fixed column index number.

For example:

```
=VLOOKUP(E2,A2:C100,3,FALSE)
```

This formula returns the third column in the selected range.

If someone inserts or rearranges columns, the formula may return an unintended column or require modification. This makes it less flexible when working with changing datasets.

**Limitation 3: It returns only the first matching result**

If an Employee ID or customer name appears multiple times, VLOOKUP returns only the first matching result. It does not automatically return every matching row.

### **4. How does a Business Analyst use VLOOKUP?**

Consider a business scenario in which you receive two Excel files:

- File 1 contains customer IDs and order details.
- File 2 contains customer IDs, names and locations.

You need to add customer names and locations to the order dataset.

You can use VLOOKUP to match the Customer ID in both files and retrieve the corresponding customer information.

This is useful for:

- Combining information from different datasets.
- Matching customer or employee records.
- Reconciling sales reports.
- Checking whether records exist in another dataset.
- Preparing data for reporting and analysis.

### **5. How can you handle VLOOKUP's limitations?**

The guide introduces XLOOKUP and INDEX-MATCH as more flexible alternatives.

- **XLOOKUP** can search in either direction, has a built-in option for displaying a result when no match is found, and does not require a hard-coded column index.
- **INDEX-MATCH** provides similar flexibility and is useful in Excel versions that do not support XLOOKUP.

### **Interview-ready answer**

For example, as a Business Analyst, I can use VLOOKUP to search for an Employee ID in an employee dataset and retrieve the employee's name or department.

## **Q2. How is XLOOKUP or INDEX-MATCH better than VLOOKUP?**

**XLOOKUP and INDEX-MATCH are lookup methods in Excel that overcome several limitations of VLOOKUP.** They allow more flexible searches, including looking both left and right, and they are useful when working with large or frequently changing datasets.

Your interview guide covers both alternatives.&#x20;



### **1. XLOOKUP**

XLOOKUP searches for a value in one range and returns the corresponding value from another range.

**Syntax:**

```
=XLOOKUP(lookup_value, lookup_array, return_array, [if_not_found])
```

| Argument       | Meaning                               |
| -------------- | ------------------------------------- |
| `lookup_value` | Value you want to find                |
| `lookup_array` | Range where Excel should search       |
| `return_array` | Range containing the result           |
| `if_not_found` | Optional message if no match is found |

### **2. Practical example**

Suppose you have an employee dataset:

| Employee ID | Employee Name | Department |
| ----------- | ------------- | ---------- |
| E101        | Rahul         | Finance    |
| E102        | Priya         | HR         |
| E103        | Arjun         | IT         |
| E104        | Neha          | Sales      |

You want to retrieve the department for Employee ID `E102`.

Assume the employee ID is entered in cell `A2`.

```
=XLOOKUP(A2,Emp!A:A,Emp!C:C,"Not found")
```

Explanation:

- `A2`: Employee ID you want to search for.
- `Emp!A:A`: Column containing Employee IDs.
- `Emp!C:C`: Column containing departments.
- `"Not found"`: Message displayed if Excel cannot find the ID.

**Result:** HR

Notice that XLOOKUP does not need a column index number. You directly specify the lookup column and the return column.

### **3. Advantages of XLOOKUP over VLOOKUP**

| Feature           | VLOOKUP                      | XLOOKUP                                                                                 |
| ----------------- | ---------------------------- | --------------------------------------------------------------------------------------- |
| Search direction  | Left to right                | Left or right                                                                           |
| Column index      | Required                     | Not required                                                                            |
| Not-found message | Usually handled with IFERROR | Built-in argument                                                                       |
| Column insertion  | Can affect the formula       | More flexible                                                                           |
| Multiple matches  | Returns first match          | Can be configured to search in different directions, but standard use returns one match |

**Example of looking to the left**

Suppose Employee Name is in column A and Employee ID is in column B. You want to search using the ID and return the name.

```
=XLOOKUP(E2,B2:B100,A2:A100,"Not found")
```

This works even though the return column is to the left of the lookup column.

### **4. INDEX-MATCH**

INDEX-MATCH combines two Excel functions:

- **MATCH** finds the position of a particular value in a range.
- **INDEX** returns the value from a specified position in another range.

**Syntax:**

```
=INDEX(return_range,MATCH(lookup_value,lookup_range,0))
```

Here, `0` in MATCH means an exact match.

**Example:**

```
=INDEX(C:C,MATCH(A2,A:A,0))
```

Explanation:

1. `MATCH(A2,A:A,0)` searches for the value in A2 within column A.
2. MATCH returns the position of the matching row.
3. `INDEX(C:C,...)` uses that position to retrieve the corresponding value from column C.

If A2 contains `E102`, the formula returns the department **HR**.

### **5. Why is INDEX-MATCH better than VLOOKUP?**

**It can look in either direction.**

VLOOKUP normally cannot return a value from a column to the left of its lookup column. INDEX-MATCH can.

**It does not depend on a hard-coded column index.**

VLOOKUP uses a column number such as 2 or 3. INDEX-MATCH directly specifies the return range, making the formula more flexible if columns are rearranged.

**It works in older Excel versions.**

XLOOKUP is not available in some older Excel versions. INDEX-MATCH is a useful alternative in those versions.

### **6. VLOOKUP vs XLOOKUP vs INDEX-MATCH**

| Criteria                   | VLOOKUP | XLOOKUP            | INDEX-MATCH                          |
| -------------------------- | ------- | ------------------ | ------------------------------------ |
| Ease of use                | Easy    | Easy               | Requires understanding two functions |
| Looks left                 | No      | Yes                | Yes                                  |
| Looks right                | Yes     | Yes                | Yes                                  |
| Fixed column number        | Yes     | No                 | No                                   |
| Built-in not-found message | No      | Yes                | No, but can use IFERROR              |
| Older Excel compatibility  | Yes     | Depends on version | Yes                                  |

### **7. How would a Business Analyst use these functions?**

Imagine you receive two datasets from different departments.

**Dataset 1 – Orders**

| Order ID | Customer ID | Order Amount |
| -------- | ----------- | ------------ |
| O101     | C101        | 5,000        |
| O102     | C102        | 3,000        |
| O103     | C103        | 7,000        |

**Dataset 2 – Customers**

| Customer ID | Customer Name |
| ----------- | ------------- |
| C101        | Rahul         |
| C102        | Priya         |
| C103        | Arjun         |

You need to add customer names to the Orders dataset.

You can use XLOOKUP:

```
=XLOOKUP(B2,Customers!A:A,Customers!B:B,"Not found")
```

This matches the Customer ID and retrieves the corresponding Customer Name.

As a BA, you can use this to combine information, validate records, prepare reports and reconcile datasets from different departments.

### **Interview-ready answer**

XLOOKUP and INDEX-MATCH are more flexible alternatives to VLOOKUP.

XLOOKUP can search both left and right, does not require a fixed column index number, and provides a built-in option to display a message when a value is not found.

For example, `=XLOOKUP(A2,Emp!A:A,Emp!C:C,"Not found")` searches for an Employee ID in column A and returns the corresponding department from column C.

INDEX-MATCH combines the INDEX and MATCH functions. MATCH finds the position of the lookup value, and INDEX retrieves the corresponding value from the return range.

As a Business Analyst, I would use these functions to match records across datasets, retrieve customer or employee information, and improve data reconciliation. INDEX-MATCH is also useful when working with older Excel versions that do not support XLOOKUP.

## **Q4. Explain IF, Nested IF and IFS functions in Excel.**

The **IF function** is a logical function in Excel that checks whether a condition is TRUE or FALSE and returns one result if TRUE and another result if FALSE.

Business Analysts use IF functions to classify data, check targets, evaluate business rules and identify records that meet specific conditions.

### **1. IF function**

**Syntax:**

```
=IF(logical_test, value_if_true, value_if_false)
```

| Argument         | Meaning                                   |
| ---------------- | ----------------------------------------- |
| `logical_test`   | The condition Excel needs to check        |
| `value_if_true`  | Result returned if the condition is TRUE  |
| `value_if_false` | Result returned if the condition is FALSE |

**Example: Sales target achievement**

Suppose an employee's monthly sales target is ₹50,000. Actual sales are recorded in cell B2.

You want Excel to display:

- "Target Met" if sales are ₹50,000 or more.
- "Below Target" if sales are less than ₹50,000.

Formula:

```
=IF(B2>=50000,"Target Met","Below Target")
```

| Employee | Sales  | Result       |
| -------- | ------ | ------------ |
| Rahul    | 65,000 | Target Met   |
| Priya    | 45,000 | Below Target |
| Arjun    | 50,000 | Target Met   |

**Explanation:** Excel checks whether B2 is greater than or equal to 50,000. If TRUE, it displays "Target Met"; otherwise, it displays "Below Target".

### **2. Nested IF function**

A **Nested IF** means placing one IF function inside another IF function.

It is used when you need to evaluate multiple conditions and return different results.

**Example: Student grading system**

Suppose marks are available in cell B2.

The grading rules are:

| Marks       | Grade |
| ----------- | ----- |
| 90 or above | A     |
| 75–89       | B     |
| Below 75    | C     |

Formula:

```
=IF(B2>=90,"A",IF(B2>=75,"B","C"))
```

How Excel evaluates it:

1. First, it checks whether marks are at least 90.
2. If TRUE, it returns A.
3. If FALSE, it checks whether marks are at least 75.
4. If TRUE, it returns B.
5. If both conditions are FALSE, it returns C.

**Example results:**

| Student | Marks | Grade |
| ------- | ----- | ----- |
| Rahul   | 95    | A     |
| Priya   | 82    | B     |
| Arjun   | 68    | C     |

Nested IF is useful when business rules have multiple levels, such as performance categories, commission slabs or customer classifications.

### **3. IFS function**

The **IFS function** checks multiple conditions and returns the result corresponding to the first TRUE condition.

It is generally easier to read than a long Nested IF formula.

**Syntax:**

```
=IFS(condition1,result1,condition2,result2,...)
```

**Example: Employee performance classification**

Suppose employee performance scores are in B2.

Rules:

| Score       | Category          |
| ----------- | ----------------- |
| 90 or above | Excellent         |
| 75 or above | Good              |
| 60 or above | Average           |
| Below 60    | Needs Improvement |

Formula:

```
=IFS(B2>=90,"Excellent",B2>=75,"Good",B2>=60,"Average",TRUE,"Needs Improvement")
```

The final `TRUE` acts as a default condition. If none of the earlier conditions is met, Excel returns "Needs Improvement".

### **4. Difference between IF, Nested IF and IFS**

| Feature    | IF                | Nested IF                | IFS                      |
| ---------- | ----------------- | ------------------------ | ------------------------ |
| Conditions | One condition     | Multiple conditions      | Multiple conditions      |
| Complexity | Simple            | Can become complex       | Easier to read           |
| Usage      | Basic decisions   | Multiple decision levels | Multiple decision levels |
| Example    | Target met or not | A, B or C grade          | Performance categories   |

### **5. How does a Business Analyst use these functions?**

A Business Analyst may use these functions to apply business rules to operational or financial data.

Examples:

- **Sales analysis:** Identify whether employees achieved their targets.
- **Customer classification:** Categorise customers based on purchase value.
- **Ticket management:** Classify support tickets by priority.
- **Performance reporting:** Assign performance categories based on KPIs.
- **Financial analysis:** Calculate commission categories based on sales.

For example, a BA analysing customer support data might classify tickets as High, Medium or Low priority based on resolution time or severity.

### **Interview-ready answer**

The IF function in Excel is used to check a condition and return one value if the condition is TRUE and another value if it is FALSE.

For example, `=IF(B2>=50000,"Target Met","Below Target")` checks whether an employee's sales are at least 50,000.

A Nested IF is used when we need to evaluate multiple conditions by placing one IF function inside another. For example, we can use it to assign grades such as A, B or C based on marks.

The IFS function also evaluates multiple conditions, but its formula is generally easier to read because we can specify multiple conditions and their corresponding results directly.

As a Business Analyst, I use these functions for data classification, target tracking, performance reporting and applying business rules to datasets.

### **6. Possible follow-up interview questions**

**1. What happens if none of the conditions in an IFS function is TRUE?**

Excel returns `#N/A` if no condition is TRUE and no default condition is provided. You can add `TRUE` as the final condition to specify a default result.

**2. Can IF be used with AND and OR?**

Yes. AND checks whether all conditions are TRUE, while OR checks whether at least one condition is TRUE.

Example using AND:

```
=IF(AND(B2>=50000,C2>=90),"Eligible","Not Eligible")
```

This returns "Eligible" only if sales are at least 50,000 and the performance score is at least 90.

**3. What is the difference between AND and OR?**

- AND: All specified conditions must be TRUE.
- OR: At least one specified condition must be TRUE.

**4. When would you prefer IFS over Nested IF?**

I would prefer IFS when there are several conditions because it makes the formula easier to understand, maintain and modify.

## **Q5. SUMIF vs SUMIFS, COUNTIF vs COUNTIFS**

These are Excel functions used to calculate totals or count records based on specified conditions.

- **SUMIF and SUMIFS** calculate sums based on conditions.
- **COUNTIF and COUNTIFS** count cells or records that meet conditions.

As a Business Analyst, you can use them to calculate sales by region, count closed support tickets, analyse transactions and prepare performance reports.

### **1. SUMIF function**

**SUMIF** adds values that meet one condition.

**Syntax:**

```
=SUMIF(range, criteria, [sum_range])
```

| Argument    | Meaning                                  |
| ----------- | ---------------------------------------- |
| `range`     | Cells Excel checks against the condition |
| `criteria`  | Condition that must be met               |
| `sum_range` | Cells to add if the condition is met     |

**Example: Calculate sales for the North region**

| Region | Sales  |
| ------ | ------ |
| North  | 50,000 |
| South  | 30,000 |
| North  | 40,000 |
| East   | 20,000 |
| South  | 25,000 |

Formula:

```
=SUMIF(A2:A6,"North",B2:B6)
```

Result: **90,000**

Excel checks the Region column and adds the Sales values only for rows where the region is North.

### **2. SUMIFS function**

**SUMIFS** adds values that meet multiple conditions.

**Syntax:**

```
=SUMIFS(sum_range, criteria_range1, criteria1, criteria_range2, criteria2)
```

Unlike SUMIF, SUMIFS places the `sum_range` argument first.

**Example: Calculate January sales for the North region**

| Region | Month | Sales  |
| ------ | ----- | ------ |
| North  | Jan   | 50,000 |
| South  | Jan   | 30,000 |
| North  | Feb   | 40,000 |
| North  | Jan   | 25,000 |
| South  | Feb   | 20,000 |

Formula:

```
=SUMIFS(C2:C6,A2:A6,"North",B2:B6,"Jan")
```

Result: **75,000**

Explanation:

- `C2:C6`: Sales values to add.
- `A2:A6`: Region column.
- `"North"`: First condition.
- `B2:B6`: Month column.
- `"Jan"`: Second condition.

Excel adds sales only when both conditions are satisfied.

### **3. COUNTIF function**

**COUNTIF** counts the number of cells that meet one condition.

**Syntax:**

```
=COUNTIF(range, criteria)
```

**Example: Count closed support tickets**

| Ticket ID | Status  |
| --------- | ------- |
| T101      | Closed  |
| T102      | Open    |
| T103      | Closed  |
| T104      | Pending |
| T105      | Closed  |

Formula:

```
=COUNTIF(B2:B6,"Closed")
```

Result: **3**

Excel counts the number of records where the status is Closed.

### **4. COUNTIFS function**

**COUNTIFS** counts records that satisfy multiple conditions.

**Syntax:**

```
=COUNTIFS(criteria_range1, criteria1, criteria_range2, criteria2)
```

**Example: Count closed tickets in the North region**

| Ticket ID | Region | Status  |
| --------- | ------ | ------- |
| T101      | North  | Closed  |
| T102      | South  | Open    |
| T103      | North  | Closed  |
| T104      | North  | Pending |
| T105      | South  | Closed  |

Formula:

```
=COUNTIFS(B2:B6,"North",C2:C6,"Closed")
```

Result: **2**

Excel counts records where the region is North AND the status is Closed.

### **5. Difference between SUMIF, SUMIFS, COUNTIF and COUNTIFS**

| Function | Purpose        | Conditions |
| -------- | -------------- | ---------- |
| SUMIF    | Adds values    | One        |
| SUMIFS   | Adds values    | Multiple   |
| COUNTIF  | Counts records | One        |
| COUNTIFS | Counts records | Multiple   |

An easy way to remember:

- **SUM** means add numerical values.
- **COUNT** means count matching cells or records.
- **IF** means one condition.
- **IFS** means multiple conditions.

### **6. Using cell references instead of fixed criteria**

You do not always need to type conditions directly into formulas. You can refer to cells containing the criteria.

For example, suppose cell E2 contains `North`.

```
=SUMIF(A2:A100,E2,B2:B100)
```

This calculates sales for whichever region is entered in E2.

For multiple conditions, suppose E2 contains `North` and F2 contains `Jan`:

```
=SUMIFS(C2:C100,A2:A100,E2,B2:B100,F2)
```

This makes reports more flexible because you can change the criteria without editing the formula.


## **Q6. How do you clean messy data in Excel?**

**Data cleaning** is the process of identifying and correcting inaccurate, incomplete, inconsistent or duplicate information in a dataset so that it can be used reliably for analysis and reporting.

As a Business Analyst, data cleaning is important because incorrect or inconsistent data can produce misleading reports, inaccurate KPIs and poor business decisions.

For example, suppose you receive a customer dataset containing duplicate records, extra spaces, inconsistent names, missing values and numbers stored as text. Before analysing it, you need to clean and standardise the data.

### **1. Remove duplicate records**

Duplicate records occur when the same information appears more than once.

Example:

| Customer ID | Customer Name |
| ----------- | ------------- |
| C101        | Rahul         |
| C102        | Priya         |
| C101        | Rahul         |
| C103        | Arjun         |

Here, customer C101 appears twice.

**How to remove duplicates:**

1. Select the dataset.
2. Go to **Data → Remove Duplicates**.
3. Select the columns Excel should check.
4. Click OK.

Excel identifies and removes duplicate rows based on the selected columns.



**Important:** Before removing duplicates, check whether they are genuinely unwanted. For example, two orders from the same customer are not necessarily duplicate transactions.

### **2. Remove extra spaces using TRIM**

Sometimes data contains unnecessary spaces before, after or between words.

Example:

- `" Rahul "`
- `"Priya  Sharma"`
- `" Arjun"`

These spaces can cause lookup and matching problems.

Use:

```
=TRIM(A2)
```

TRIM removes leading and trailing spaces and reduces repeated spaces between words to a single space.

Example:

| Original value  | TRIM result    |
| --------------- | -------------- |
| `Rahul`         | `Rahul`        |
| `Priya  Sharma` | `Priya Sharma` |
| `Arjun`         | `Arjun`        |

This is especially useful before performing VLOOKUP, XLOOKUP or other matching operations.

### **3. Standardise text using PROPER and UPPER**

Sometimes the same information is written in different formats.

Example:

| Original data |
| ------------- |
| rahul         |
| RAHUL         |
| Rahul         |

You can standardise the text using:

**PROPER:** Capitalises the first letter of each word.

```
=PROPER(A2)
```

Result: `Rahul`

**UPPER:** Converts all text into uppercase.

```
=UPPER(A2)
```

Result: `RAHUL`

**LOWER:** Converts all text into lowercase.

```
=LOWER(A2)
```

Result: `rahul`

These functions are useful for standardising customer names, locations, departments and categories.

### **4. Split combined data using Text to Columns or TEXTSPLIT**

Sometimes multiple pieces of information are stored in one cell.

Example:

| Full Name    |
| ------------ |
| Rahul Sharma |
| Priya Reddy  |
| Arjun Kumar  |

Suppose you need separate First Name and Last Name columns.

**Method 1: Text to Columns**

1. Select the column.
2. Go to **Data → Text to Columns**.
3. Select Delimited.
4. Choose Space as the delimiter.
5. Click Finish.

Excel separates the names into different columns.



**Method 2: TEXTSPLIT**

In supported Excel versions:

```
=TEXTSPLIT(A2," ")
```

This splits the text using a space as the delimiter.

### **5. Find and Replace**

Find and Replace helps correct repeated spelling mistakes or inconsistent values throughout a dataset.

Example:

| Original Region |
| --------------- |
| N               |
| North           |
| N               |
| North           |

Suppose the business wants all entries standardised as `North`.

Steps:

1. Press `Ctrl + H`.
2. Enter the value to find, such as `N`.
3. Enter the replacement value, `North`.
4. Click Replace All.

Be careful with short search terms. Replacing `N` indiscriminately could alter other text. Check the selected data and use whole-cell matching where appropriate.

### **6. Fix numbers stored as text**

Sometimes numerical values are stored as text instead of numbers.

For example, a sales amount may look like `50000` but Excel treats it as text.

This can cause problems when calculating totals or averages.

Ways to fix it include:

- Select the affected cells and use the warning icon's **Convert to Number** option.
- Use `VALUE(A2)` to convert numeric text into a number.
- Use Text to Columns where appropriate.
- Check that the cell format and imported data types are correct.

Example:

```
=VALUE(A2)
```

This converts a numeric text value into a number.

### **7. Handle missing values and errors**

Missing values can affect calculations and reporting.

Example:

| Employee | Sales  |
| -------- | ------ |
| Rahul    | 50,000 |
| Priya    |        |
| Arjun    | 40,000 |

You should first determine why the value is missing.

Possible approaches:

- Filter blank cells to identify incomplete records.
- Obtain the missing information from the source.
- Use a suitable business-approved treatment for missing values.

### **Q8. What is the difference between absolute and relative cell references?**

In Excel, a **cell reference** identifies a cell used in a formula. Relative and absolute references determine whether that cell reference changes when you copy the formula to another cell.

There are three main types:

1. Relative reference — `A1`
2. Absolute reference — `$A$1`
3. Mixed reference — `$A1` or `A$1`

### **1. Relative reference —&#x20;**`A1`

A relative reference changes automatically when you copy a formula to another cell.

**Example:**

Suppose you have sales values in column B:

| Cell | Sales  |
| ---- | ------ |
| B2   | 10,000 |
| B3   | 15,000 |
| B4   | 20,000 |

In cell C2, enter:

```
=B2*10%
```

When you copy the formula down, Excel adjusts the reference automatically:

| Cell | Formula   |
| ---- | --------- |
| C2   | `=B2*10%` |
| C3   | `=B3*10%` |
| C4   | `=B4*10%` |

The reference changes from B2 to B3 and B4 because it is relative.

**When to use it:** When applying the same calculation to multiple rows, such as calculating commission for each employee or tax for each transaction.

### **2. Absolute reference —&#x20;**`$A$1`

An absolute reference remains fixed when you copy a formula.

The dollar signs lock both the column and the row:

- `$A` locks column A.
- `$1` locks row 1.

**Example: Calculating tax using a fixed tax rate**

Suppose:

- Sales amounts are in B2:B4.
- The tax rate is stored in E1 as 10%.

In C2, enter:

```
=B2*$E$1
```

Copy the formula down:

| Cell | Formula    |
| ---- | ---------- |
| C2   | `=B2*$E$1` |
| C3   | `=B3*$E$1` |
| C4   | `=B4*$E$1` |

Notice what happens:

- B2 changes to B3 and B4 because it is relative.
- `$E$1` stays fixed because it is absolute.

This ensures that every sales amount is multiplied by the same tax rate.

**When to use it:** When a formula needs to refer repeatedly to a fixed value, such as a tax rate, commission percentage, exchange rate or target value.

### **3. Mixed reference —&#x20;**`$A1`**&#x20;and&#x20;**`A$1`

A mixed reference locks either the column or the row, but not both.

| Reference | What stays fixed?  | What can change? |
| --------- | ------------------ | ---------------- |
| `$A1`     | Column A           | Row number       |
| `A$1`     | Row 1              | Column letter    |
| `$A$1`    | Column A and row 1 | Nothing          |

**Example:**

Suppose you are creating a multiplication table.

- Numbers are listed vertically in column A.
- Numbers are listed horizontally in row 1.
- You want to multiply the corresponding row and column values.

You could use:

```
=$A2*B$1
```

When copied across and down:

- `$A2` always refers to column A, but its row changes.
- `B$1` always refers to row 1, but its column changes.

Mixed references are particularly useful when copying formulas across both rows and columns.

### **How to switch between reference types**

While editing a formula, select the cell reference and press **F4** to cycle through:

```
A1
$A$1
A$1
$A1
```

On some laptops, you may need to press `Fn + F4`.

### **Comparison**

| Feature                     | Relative                | Absolute          | Mixed                           |
| --------------------------- | ----------------------- | ----------------- | ------------------------------- |
| Example                     | `A1`                    | `$A$1`            | `$A1` or `A$1`                  |
| Row changes when copied?    | Yes                     | No                | Depends                         |
| Column changes when copied? | Yes                     | No                | Depends                         |
| Typical use                 | Row-by-row calculations | Fixed assumptions | Formulas copied across and down |

### **How a Business Analyst uses these references**

A Business Analyst frequently works with large datasets and recurring calculations.

For example, imagine you are preparing a monthly sales report:

- Column B contains sales revenue.
- Cell E1 contains the commission rate.
- Column C calculates commission.

You can use:

```
=B2*$E$1
```

This formula allows you to calculate commission for every employee while keeping the commission rate fixed.

Other BA applications include:

- Applying a fixed exchange rate to transaction values.
- Calculating performance against a fixed target.
- Applying a standard tax percentage.
- Calculating bonuses using a fixed commission rate.

### **Interview-ready answer**

A relative cell reference, such as `A1`, changes when a formula is copied to another cell. An absolute reference, such as `$A$1`, keeps both the row and column fixed. A mixed reference, such as `$A1` or `A$1`, locks either the column or the row.

For example, if I have sales values in column B and a fixed tax rate in cell E1, I can use `=B2*$E$1` and copy the formula down. The sales reference changes for each row, but the tax rate remains fixed.

As a Business Analyst, I use absolute references when applying fixed assumptions such as tax rates, commission percentages or exchange rates across large datasets.

### **Q10. What are some other important Excel functions a Business Analyst should know?**

Apart from lookup functions, Pivot Tables and conditional formatting, a Business Analyst should know several Excel functions for handling text, errors, dates, filtering, sorting and calculations.

The guide highlights these functions:

| ------------------------ | --------------------------------------------- |

| `ROUND`                  | Rounds numbers                                |

Let's understand each with examples.

### **1. IFERROR**

**Purpose:** Returns a specified value instead of displaying an error when a formula produces one.

Syntax:

```
=IFERROR(value, value_if_error)
```

Example:

```
=IFERROR(A2/B2, "Invalid")
```

If B2 is zero or the calculation otherwise produces an error, Excel displays `"Invalid"` instead.

**BA application:** Preventing lookup reports and calculation sheets from displaying errors such as `#N/A` or `#DIV/0!`.

### **2. CONCAT and TEXTJOIN**

Both functions combine text from multiple cells.

**CONCAT:**

```
=CONCAT(A2, " ", B2)
```

If A2 contains `Rahul` and B2 contains `Sharma`, the result is:

`Rahul Sharma`

**TEXTJOIN:**

```
=TEXTJOIN(", ", TRUE, A2:A5)
```

This combines values from A2 through A5, separates them with commas and ignores empty cells.

**Difference:** CONCAT joins text directly; TEXTJOIN lets you specify a delimiter and whether to ignore empty cells.

**BA application:** Combining first and last names, creating labels, or joining product codes and descriptions.

### **3. LEFT, RIGHT and MID**

These functions extract specific characters from a text value.

Suppose cell A2 contains:

`HYD20261124`

| Function | Formula        | Result |
| -------- | -------------- | ------ |
| LEFT     | `=LEFT(A2,3)`  | HYD    |
| RIGHT    | `=RIGHT(A2,4)` | 1124   |
| MID      | `=MID(A2,4,4)` | 2026   |

- `LEFT` extracts characters from the beginning.
- `RIGHT` extracts characters from the end.
- `MID` extracts characters from a specified position.

**BA application:** Extracting region codes, product identifiers, customer codes or year information from structured IDs.

### **4. LEN**

**Purpose:** Counts the number of characters in a text value, including spaces.

Syntax:

```
=LEN(A2)
```

If A2 contains `Business Analyst`, the result is 16.

**BA application:** Checking whether customer IDs, phone numbers or reference codes have the expected number of characters.

### **5. TEXT**

**Purpose:** Converts a number or date into text using a specified format.

Syntax:

```
=TEXT(value, format_text)
```

Examples:

```
=TEXT(A2, "dd-mm-yyyy")
```

Formats a date as day-month-year.

```
=TEXT(B2, "₹#,##0.00")
```

Formats a number as Indian currency text.

**BA application:** Preparing consistent dates and financial values for reports and presentations.

Note that TEXT returns text, not a numeric value. Use the original numeric value for calculations.

### **6. DATEDIF**

**Purpose:** Calculates the difference between two dates in years, months or days.

Syntax:

```
=DATEDIF(start_date, end_date, unit)
```

Examples:

```
=DATEDIF(A2,B2,"Y")
```

Returns the number of complete years between the dates.

```
=DATEDIF(A2,B2,"M")
```

Returns the number of complete months.

```
=DATEDIF(A2,B2,"D")
```

Returns the number of days.

**BA application:** Calculating customer tenure, employee service duration or the time between project milestones.

### **7. NETWORKDAYS**

**Purpose:** Calculates the number of working days between two dates, excluding weekends and optionally specified holidays.

Syntax:

```
=NETWORKDAYS(start_date, end_date, [holidays])
```

Example:

```
=NETWORKDAYS(A2,B2)
```

If A2 is a Monday and B2 is the following Friday, the result is 5.

You can also provide a holiday range:

```
=NETWORKDAYS(A2,B2,H2:H10)
```

**BA application:** Measuring business-day turnaround time, checking SLA compliance and tracking project duration.

**Purpose:** Returns distinct values from a range.

| South  |
| North  |
| East   |
| South  |

| Unique regions |
| -------------- |

**BA application:** Creating unique lists of customers, regions, departments or product categories.

### **9. FILTER**

**Purpose:** Returns only records that meet specified conditions.

Syntax:

```
=FILTER(array, include, [if_empty])
```

Example:

```
=FILTER(A2:C100, C2:C100="North", "No records")
```

This returns rows from A2:C100 where column C contains `North`.

**BA application:** Extracting specific customer groups, regional sales or transactions that meet a business condition.

### **10. SORT**

**Purpose:** Sorts a range in ascending or descending order.

Syntax:

```
=SORT(array, [sort_index], [sort_order])
```

Example:

```
=SORT(A2:C100, 2, -1)
```

This sorts the selected range by its second column in descending order.

- `1` means ascending.
- `-1` means descending.

**BA application:** Ranking sales performance, sorting customer records or arranging products by revenue.

### **11. ROUND**

**Purpose:** Rounds a number to a specified number of decimal places.

Syntax:

```
=ROUND(number, num_digits)
```

Examples:

```
=ROUND(1234.567, 2)
```

Result: `1234.57`

Result: `1235`

### **12. AVERAGEIFS**

Example:

This calculates the average of values in column C where:

**BA application:** Calculating average sales for a particular region and order status, or average ticket resolution time for a specific department.

### **Q11. What are What-If Analysis tools in Excel?**

**What-If Analysis** is an Excel feature that helps you understand how changing input values affects a formula or business result.

It is useful when a Business Analyst needs to evaluate different business scenarios, forecast outcomes or determine what inputs are required to achieve a particular target.

Excel provides three main What-If Analysis tools:

1. Goal Seek
2. Scenario Manager

### **1. Goal Seek**

**Purpose:** Goal Seek finds the input value required to achieve a specific result in a formula.

In simple terms, you know the result you want, and Excel calculates the input needed to reach it.

**Example: Sales target**

Suppose a company wants to achieve ₹1,00,000 in revenue.

- Selling price per unit: ₹500
- Units sold: 150
- Current revenue: ₹75,000
- Target revenue: ₹1,00,000

Revenue formula:

```
=B2*B3
```

Where:

- B2 = Selling price
- B3 = Units sold

You want to find how many units must be sold to achieve ₹1,00,000 revenue.

### **Q1. What are the different types of SQL commands?**

SQL (Structured Query Language) is used to communicate with relational databases. It allows us to create database structures, insert and modify records, retrieve information, control access and manage transactions.

SQL commands are commonly classified into five categories:

1. DDL – Data Definition Language
2. DML – Data Manipulation Language
3. DQL – Data Query Language
4. DCL – Data Control Language
5. TCL – Transaction Control Language

### **1. DDL – Data Definition Language**

DDL commands are used to create or modify the structure of database objects, such as tables.

Common commands:

| Command    | Purpose                                                        |
| ---------- | -------------------------------------------------------------- |
| `CREATE`   | Creates a new database or table                                |
| `ALTER`    | Modifies an existing table structure                           |
| `DROP`     | Deletes a database object                                      |
| `TRUNCATE` | Removes all records from a table while retaining its structure |

**Example: CREATE**

```
CREATE TABLE employees (
    emp_id INT,
    name VARCHAR(100),
    salary DECIMAL(10,2)
);
```

This creates an employees table with three columns.

**Example: ALTER**

```
ALTER TABLE employees
ADD department VARCHAR(50);
```

This adds a department column to the existing table.

**Business Analyst application:** Understanding DDL helps when discussing database structures, required fields and changes to data models with developers or database teams.

### **2. DML – Data Manipulation Language**

DML commands are used to insert, modify and delete records stored in database tables.

Common commands:

| Command  | Purpose                   |
| -------- | ------------------------- |
| `INSERT` | Adds new records          |
| `UPDATE` | Modifies existing records |
| `DELETE` | Removes selected records  |

**Example: INSERT**

```
INSERT INTO employees (emp_id, name, salary)
VALUES (101, 'Rahul', 50000);
```

This inserts a new employee record.

**Example: UPDATE**

```
UPDATE employees
SET salary = 55000
WHERE emp_id = 101;
```

This changes the salary of employee 101.

**Example: DELETE**

```
DELETE FROM employees
WHERE emp_id = 101;
```

This removes the record for employee 101.

**Business Analyst application:** DML knowledge helps you understand how business transactions and operational records are maintained in databases.

### **3. DQL – Data Query Language**

DQL is used to retrieve data from database tables.

The primary command is:

`SELECT`

**Example:**

```
SELECT name, salary
FROM employees;
```

This retrieves employee names and salaries.

You can also filter records:

```
SELECT *
FROM employees
WHERE salary > 50000;
```

This returns employees whose salary is greater than 50,000.

**Business Analyst application:** DQL is especially important because Business Analysts frequently use SQL to retrieve data, analyse business performance, validate reports and answer stakeholder questions.

### **4. DCL – Data Control Language**

DCL commands are used to manage database access permissions.

Common commands:

| Command  | Purpose                                |
| -------- | -------------------------------------- |
| `GRANT`  | Gives users specific permissions       |
| `REVOKE` | Removes previously granted permissions |

**Example: GRANT**

```
GRANT SELECT ON employees TO analyst_user;
```

This gives `analyst_user` permission to retrieve data from the employees table, subject to the database's permission model.

**Example: REVOKE**

```
REVOKE SELECT ON employees FROM analyst_user;
```

This removes that permission.

**Business Analyst application:** Understanding DCL is useful when working with sensitive business information and discussing access requirements with database administrators.

### **5. TCL – Transaction Control Language**

TCL commands manage database transactions. A transaction is a group of database operations treated as a logical unit of work.

Common commands:

| Command    | Purpose                                  |
| ---------- | ---------------------------------------- |
| `COMMIT`   | Saves transaction changes                |
| `ROLLBACK` | Reverses uncommitted transaction changes |

**Example: COMMIT**

```
BEGIN TRANSACTION;

UPDATE employees
SET salary = 60000
WHERE emp_id = 101;

COMMIT;
```

The change is committed to the database.

**Example: ROLLBACK**

```
BEGIN TRANSACTION;

UPDATE employees
SET salary = 60000
WHERE emp_id = 101;

ROLLBACK;
```

The uncommitted update is reversed.

**Business Analyst application:** TCL is useful to understand when analysing database transactions and discussing data consistency with technical teams.

*Note:* Exact transaction syntax and behaviour can vary between database systems.

### **6. Comparison of all five SQL command types**

| Category | Full form                    | Main purpose               | Commands                      |
| -------- | ---------------------------- | -------------------------- | ----------------------------- |
| DDL      | Data Definition Language     | Defines database structure | CREATE, ALTER, DROP, TRUNCATE |
| DML      | Data Manipulation Language   | Changes table records      | INSERT, UPDATE, DELETE        |
| DQL      | Data Query Language          | Retrieves data             | SELECT                        |
| DCL      | Data Control Language        | Controls permissions       | GRANT, REVOKE                 |
| TCL      | Transaction Control Language | Manages transactions       | COMMIT, ROLLBACK              |

1. DDL stands for Data Definition Language. It is used to create or modify database structures. Examples include CREATE, ALTER, DROP and TRUNCATE.
2. DML stands for Data Manipulation Language. It is used to insert, update and delete records. Examples include INSERT, UPDATE and DELETE.
3. DQL stands for Data Query Language. It is used to retrieve information from tables using SELECT.
4. DCL stands for Data Control Language. It manages database permissions using GRANT and REVOKE.
5. TCL stands for Transaction Control Language. It manages transactions using COMMIT and ROLLBACK.

As a Business Analyst, I primarily use SELECT queries to retrieve and analyse business data, but understanding all five categories helps me communicate effectively with database and development teams.

### **Q2. Explain the different types of SQL JOINs.**

A **JOIN** in SQL is used to combine records from two or more tables based on a related column between them.

In a relational database, information is often stored in separate tables to avoid unnecessary duplication. JOINs allow us to retrieve related information from these tables together.

There are five important JOIN types:

1. INNER JOIN
2. LEFT JOIN
3. RIGHT JOIN
4. FULL OUTER JOIN
5. SELF JOIN

Let's understand each using a practical example.

### **1. Understanding the sample tables**

Suppose a company has two tables.

**Employees table**

| emp_id | name  | dept_id |
| ------ | ----- | ------- |
| 101    | Rahul | 10      |
| 102    | Priya | 20      |
| 103    | Amit  | 30      |
| 104    | Neha  | 10      |

**Departments table**

| dept_id | dept_name |
| ------- | --------- |
| 10      | HR        |
| 20      | Finance   |
| 40      | Marketing |

Notice:

- Rahul and Neha belong to HR.
- Priya belongs to Finance.
- Amit has department ID 30, which does not exist in the Departments table.
- Marketing has no employee assigned to it.

We will use these tables to understand the JOIN types.

### **2. INNER JOIN**

**Definition:** INNER JOIN returns only the records that have matching values in both tables.

It excludes records that do not have a match.

**SQL query:**

```
SELECT
    e.name,
    d.dept_name
FROM employees e
INNER JOIN departments d
    ON e.dept_id = d.dept_id;
```

**Output:**

| name  | dept_name |
| ----- | --------- |
| Rahul | HR        |
| Priya | Finance   |
| Neha  | HR        |

Amit is excluded because department ID 30 does not exist in the Departments table. Marketing is excluded because no employee belongs to department 40.

**When to use it:**

- When you only need employees with valid department records.
- When combining orders with matching customer records.
- When analysing transactions linked to existing products.

### **3. LEFT JOIN (LEFT OUTER JOIN)**

**Definition:** LEFT JOIN returns all records from the left table and matching records from the right table.

If there is no match in the right table, SQL returns NULL for the right table's columns.

**SQL query:**

```
SELECT
    e.name,
    d.dept_name
FROM employees e
LEFT JOIN departments d
    ON e.dept_id = d.dept_id;
```

**Output:**

| name  | dept_name |
| ----- | --------- |
| Rahul | HR        |
| Priya | Finance   |
| Amit  | NULL      |
| Neha  | HR        |

All employees are included, even if their department does not have a matching record.

**When to use it:**

- Finding employees who do not have a valid department mapping.
- Listing all customers, including those who have never placed an order.
- Identifying products that have no sales.

**Important:** LEFT JOIN is particularly useful for identifying missing relationships between datasets.

### **4. RIGHT JOIN (RIGHT OUTER JOIN)**

**Definition:** RIGHT JOIN returns all records from the right table and matching records from the left table.

If there is no match in the left table, SQL returns NULL for the left table's columns.

**SQL query:**

```
SELECT
    e.name,
    d.dept_name
FROM employees e
RIGHT JOIN departments d
    ON e.dept_id = d.dept_id;
```

**Output:**

| name  | dept_name |
| ----- | --------- |
| Rahul | HR        |
| Neha  | HR        |
| Priya | Finance   |
| NULL  | Marketing |

Marketing appears even though no employee belongs to it.

**When to use it:**

- When you need every department, including departments without employees.
- When analysing all products, including products without sales.
- When checking whether all records in a reference table are being used.

A RIGHT JOIN can often be rewritten as a LEFT JOIN by reversing the order of the tables.

### **5. FULL OUTER JOIN**

**Definition:** FULL OUTER JOIN returns all records from both tables.

It includes:

- Matching records from both tables.
- Unmatched records from the left table.
- Unmatched records from the right table.

For unmatched records, the missing side contains NULL values.

**SQL query:**

```
SELECT
    e.name,
    d.dept_name
FROM employees e
FULL OUTER JOIN departments d
    ON e.dept_id = d.dept_id;
```

**Output:**

| name  | dept_name |
| ----- | --------- |
| Rahul | HR        |
| Priya | Finance   |
| Amit  | NULL      |
| Neha  | HR        |
| NULL  | Marketing |

**When to use it:**

- Reconciling two datasets.
- Comparing records between two systems.
- Identifying records missing from either dataset.
- Checking whether two reports contain the same underlying records.

### **6. SELF JOIN**

**Definition:** A SELF JOIN is when a table is joined with itself.

It is useful when records in the same table are related to other records in that table.

**Example: Employee and Manager relationship**

Suppose the employees table contains:

| emp_id | name  | manager_id |
| ------ | ----- | ---------- |
| 101    | Rahul | NULL       |
| 102    | Priya | 101        |
| 103    | Amit  | 101        |
| 104    | Neha  | 102        |

Here, `manager_id` refers to another employee's `emp_id`.

We can use a SELF JOIN to display employee names alongside their manager names.

**SQL query:**

```
SELECT
    e.name AS employee_name,
    m.name AS manager_name
FROM employees e
LEFT JOIN employees m
    ON e.manager_id = m.emp_id;
```

**Output:**

| employee_name | manager_name |
| ------------- | ------------ |
| Rahul         | NULL         |
| Priya         | Rahul        |
| Amit          | Rahul        |
| Neha          | Priya        |

Here:

- `e` represents the employee.
- `m` represents the manager.
- Both aliases refer to the same employees table.

**When to use it:**

- Employee-manager relationships.
- Parent-child relationships.
- Organisational hierarchies.
- Comparing records within the same table.

### **7. Comparison of all JOIN types**

| JOIN type       | What it returns                                         |
| --------------- | ------------------------------------------------------- |
| INNER JOIN      | Only matching records from both tables                  |
| LEFT JOIN       | All left-table records and matching right-table records |
| RIGHT JOIN      | All right-table records and matching left-table records |
| FULL OUTER JOIN | All records from both tables                            |
| SELF JOIN       | Records joined with other records from the same table   |

### **Q3. What is the difference between WHERE and HAVING in SQL?**

Both `WHERE` and `HAVING` are used to filter data in SQL, but they operate at different stages of query execution.

- **WHERE** filters individual rows before grouping and aggregation.
- **HAVING** filters groups after `GROUP BY` has been applied. It is commonly used with aggregate functions such as `SUM()`, `COUNT()` and `AVG()`.

This distinction is important when analysing business data because you may need to filter individual transactions first and then filter the resulting groups.

### **1. Understanding WHERE**

`WHERE` filters individual records before SQL performs grouping or aggregate calculations.

**Example:** Suppose you have an employees table:

| emp_id | name  | dept_id | salary | hire_date  |
| ------ | ----- | ------- | ------ | ---------- |
| 101    | Rahul | 10      | 60,000 | 2022-05-10 |
| 102    | Priya | 20      | 75,000 | 2023-02-15 |
| 103    | Amit  | 10      | 45,000 | 2023-06-20 |
| 104    | Neha  | 20      | 80,000 | 2024-01-10 |

You want to retrieve employees hired after 1 January 2023.

```
SELECT *
FROM employees
WHERE hire_date >= '2023-01-01';
```

**Output:**

| emp_id | name  | dept_id | salary |
| ------ | ----- | ------- | ------ |
| 102    | Priya | 20      | 75,000 |
| 103    | Amit  | 10      | 45,000 |
| 104    | Neha  | 20      | 80,000 |

The WHERE condition filters the individual employee records before any grouping takes place.

Other examples:

```
-- Employees earning more than 50,000
SELECT *
FROM employees
WHERE salary > 50000;
```

```
-- Employees belonging to department 10
SELECT *
FROM employees
WHERE dept_id = 10;
```

### **2. Understanding HAVING**

`HAVING` filters groups after SQL performs grouping using `GROUP BY`.

It is commonly used with aggregate functions such as:

- `SUM()` – calculates totals.
- `COUNT()` – counts records.
- `AVG()` – calculates averages.
- `MAX()` – finds the maximum.
- `MIN()` – finds the minimum.

**Example:** Suppose management wants to identify departments with an average salary greater than ₹50,000.

```
SELECT
    dept_id,
    AVG(salary) AS avg_sal
FROM employees
GROUP BY dept_id
HAVING AVG(salary) > 50000;
```

**How SQL processes this query:**

Step 1: `FROM employees` selects the source table.

Step 2: `GROUP BY dept_id` groups employees by department.

Step 3: `AVG(salary)` calculates the average salary for each department.

Step 4: `HAVING AVG(salary) > 50000` keeps only departments whose average salary exceeds ₹50,000.

**Output:**

| dept_id | avg_sal |
| ------- | ------- |
| 10      | 52,500  |
| 20      | 77,500  |

Department 10 has an average salary of ₹52,500, while department 20 has an average salary of ₹77,500. Both satisfy the condition.

### **3. Can WHERE and HAVING be used together?**

Yes. This is a very common SQL interview question.

Suppose management wants to find departments where:

- Employees were hired on or after 1 January 2023.
- The average salary of those employees is greater than ₹50,000.

You can use both clauses:

```
SELECT
    dept_id,
    AVG(salary) AS avg_sal
FROM employees
WHERE hire_date >= '2023-01-01'
GROUP BY dept_id
HAVING AVG(salary) > 50000;
```

Here is what happens:

1. `WHERE` filters employees hired from 2023 onwards.
2. `GROUP BY` groups the remaining employees by department.
3. `AVG()` calculates the average salary for each department.
4. `HAVING` retains only departments with an average salary greater than ₹50,000.

The important distinction is that the average is calculated using only the employees who passed the WHERE condition.

### **4. Key differences**

| Feature                | WHERE                                              | HAVING                       |
| ---------------------- | -------------------------------------------------- | ---------------------------- |
| Purpose                | Filters individual rows                            | Filters grouped results      |
| Execution              | Before GROUP BY                                    | After GROUP BY               |
| Aggregate functions    | Generally cannot be used directly in the condition | Commonly used                |
| Used without GROUP BY? | Yes                                                | Yes, depending on the query  |
| Example                | `WHERE salary > 50000`                             | `HAVING AVG(salary) > 50000` |

### **5. Practical Business Analyst applications**

Imagine you are analysing an e-commerce company's order data.

A stakeholder asks:

"Show me customers who placed orders in 2025 and whose total purchase value exceeds ₹1 lakh."

You can write:

```
SELECT
    customer_id,
    SUM(amount) AS total_sales
FROM orders
WHERE order_date >= '2025-01-01'
  AND order_date < '2026-01-01'
GROUP BY customer_id
HAVING SUM(amount) > 100000;
```

In this query:

- WHERE selects orders placed during 2025.
- GROUP BY combines orders by customer.
- SUM calculates each customer's total purchase value.
- HAVING returns customers whose total exceeds ₹1 lakh.

This is useful for customer segmentation, revenue analysis and identifying high-value customers.

### **Interview-ready answer**

WHERE and HAVING are both used to filter data in SQL, but they work at different stages.

WHERE filters individual rows before grouping, whereas HAVING filters groups after GROUP BY and is commonly used with aggregate functions such as SUM, COUNT and AVG.

For example, if I want to calculate the average salary by department for employees hired after 2023, I use WHERE to filter employees based on their hire date and HAVING to filter departments based on their average salary.

As a Business Analyst, I use WHERE for filtering transaction-level data and HAVING for analysing aggregated results, such as departments with high average salaries or customers whose total purchases exceed a certain amount.

### **Possible follow-up interview questions**

**1. Can we use WHERE without GROUP BY?**

Yes. WHERE can be used independently to filter individual records.

**2. Can we use HAVING without GROUP BY?**

Yes. Some SQL databases allow HAVING without an explicit GROUP BY. The query treats the result as one group, so it is often used with aggregate conditions.

**3. Can we use SUM() inside WHERE?**

Generally, no. Aggregate functions are used to filter grouped results with HAVING, rather than filtering individual rows with WHERE.

**4. Which executes first: WHERE or HAVING?**

WHERE executes before GROUP BY and HAVING. HAVING filters the groups after aggregation.


### **Q4. How do you find the second-highest salary in SQL?**

To find the second-highest salary, first clarify whether the requirement means the **second-highest distinct salary** or the salary of the **second employee when employees are sorted by salary**. These can produce different results when multiple employees earn the same amount.

The interview question usually expects the second-highest distinct salary.

**Method 1: Using&#x20;**`MAX()`**&#x20;with a subquery**

```
SELECT MAX(salary) AS second_highest_salary
FROM employees
WHERE salary < (
    SELECT MAX(salary)
    FROM employees
);
```

**How it works:**

1. The inner query finds the highest salary in the employees table.
2. The outer query filters out that highest salary using `WHERE salary < ...`.
3. `MAX(salary)` then returns the highest salary among the remaining records.

For example, consider these salaries:

| Employee | Salary  |
| -------- | ------- |
| A        | 100,000 |
| B        | 90,000  |
| C        | 90,000  |
| D        | 80,000  |
| E        | 70,000  |

The highest salary is 100,000. After excluding it, the highest remaining salary is 90,000.

Result: 90,000

This method automatically handles duplicate salaries because it finds the maximum salary below the highest salary.

**Method 2: Using&#x20;**`DENSE_RANK()`

The guide also provides a window-function approach, which is useful when finding the Nth-highest salary.

```
SELECT DISTINCT salary
FROM (
    SELECT salary,
           DENSE_RANK() OVER (
               ORDER BY salary DESC
           ) AS rnk
    FROM employees
) t
WHERE rnk = 2;
```

**How it works:**

- `ORDER BY salary DESC` sorts salaries from highest to lowest.
- `DENSE_RANK()` assigns a rank to each distinct salary level.
- Employees with the same salary receive the same rank.
- `WHERE rnk = 2` selects the second-highest salary level.

For the example above:

| Salary  | DENSE_RANK |
| ------- | ---------- |
| 100,000 | 1          |
| 90,000  | 2          |
| 90,000  | 2          |
| 80,000  | 3          |
| 70,000  | 4          |

If the interviewer asks for the names of all employees earning the second-highest salary, use:

```
SELECT name, salary
FROM (
    SELECT name,
           salary,
           DENSE_RANK() OVER (
               ORDER BY salary DESC
           ) AS rnk
    FROM employees
) t
WHERE rnk = 2;
```

This returns every employee tied at the second-highest salary.

**Important edge cases**

- If the table contains only one distinct salary, there is no second-highest salary, so the first query returns `NULL`.
- If multiple employees share the second-highest salary, the `DENSE_RANK()` approach can return all of them.
- If the question asks for the second employee in a sorted list rather than the second distinct salary, `ROW_NUMBER()` may be more appropriate.

**Business Analyst application**

A Business Analyst might use this query to:

- Identify the second-highest compensation level in an organization.
- Compare salary levels across departments.
- Analyze compensation distribution.
- Support salary benchmarking and management reporting.

The key is to clarify the business definition before writing the query. A second-highest distinct salary is not necessarily the same as the salary of the second employee in a sorted list.

**Interview-ready answer**

To find the second-highest salary in SQL, I can use a subquery with the `MAX()` function.

First, the inner query finds the highest salary. Then the outer query filters out that salary and returns the maximum salary among the remaining records.

```
SELECT MAX(salary)
FROM employees
WHERE salary < (
    SELECT MAX(salary)
    FROM employees
);
```

Alternatively, I can use `DENSE_RANK()` to rank distinct salary levels in descending order and select rank 2.

I prefer `DENSE_RANK()` when I need to retrieve all employees earning the second-highest salary, especially when multiple employees have the same salary.

**Possible follow-up interview questions**

**1. What is the difference between finding the second-highest salary and finding the second employee by salary?**

- **Second-highest salary:** Finds the second-highest *distinct salary value*. Employees with the same salary count as one salary level.
- **Second employee by salary:** Finds the employee in the second position after sorting employees by salary in descending order. Duplicate salaries can occupy separate positions.

Example:

| Employee | Salary  |
| -------- | ------- |
| A        | 100,000 |
| B        | 90,000  |
| C        | 90,000  |
| D        | 80,000  |

- Second-highest distinct salary = **90,000**.
- Second employee by salary = **B or C**, depending on how ties are ordered.

**2. What happens if two employees have the same second-highest salary?**

Both employees are considered tied at the second-highest salary.

For example, if two employees earn 90,000 and the highest salary is 100,000, both employees receive rank 2 when using `DENSE_RANK()`.

To retrieve both employees:

```
SELECT name, salary
FROM (
    SELECT name,
           salary,
           DENSE_RANK() OVER (
               ORDER BY salary DESC
           ) AS rnk
    FROM employees
) t
WHERE rnk = 2;
```

This returns all employees earning the second-highest distinct salary.

**3. How would you find the third-highest salary?**

Use `DENSE_RANK()` and filter for rank 3:

```
SELECT DISTINCT salary
FROM (
    SELECT salary,
           DENSE_RANK() OVER (
               ORDER BY salary DESC
           ) AS rnk
    FROM employees
) t
WHERE rnk = 3;
```

Alternatively, use nested `MAX()` queries:

```
SELECT MAX(salary) AS third_highest_salary
FROM employees
WHERE salary < (
    SELECT MAX(salary)
    FROM employees
    WHERE salary < (
        SELECT MAX(salary)
        FROM employees
    )
);
```

The `DENSE_RANK()` approach is generally easier to extend to the fourth-, fifth-, or Nth-highest salary.

**4. Why would you use&#x20;**`DENSE_RANK()`**&#x20;instead of&#x20;**`ROW_NUMBER()`**?**

`DENSE_RANK()` is useful when multiple employees have the same salary because it assigns the same rank to tied salaries. It also does not skip rank numbers.

`ROW_NUMBER()` assigns a unique sequential number to every row, even when salaries are tied.

Example:

| Salary  | ROW_NUMBER | DENSE_RANK |
| ------- | ---------- | ---------- |
| 100,000 | 1          | 1          |
| 90,000  | 2          | 2          |
| 90,000  | 3          | 2          |
| 80,000  | 4          | 3          |

Therefore:

- Use `DENSE_RANK()` to find the second- or third-highest **distinct salary**, including all employees tied at that level.
- Use `ROW_NUMBER()` when you need a specific row position, such as the second employee after sorting.

**Interview point:** If the question asks for the second-highest salary, `DENSE_RANK()` is usually the safer choice when duplicate salaries are possible.



### **Q5. How do you find duplicate records in SQL?**

Duplicate records occur when the same value or combination of values appears more than once in a table.

To identify duplicates, SQL commonly uses `GROUP BY` with the aggregate function `COUNT()` and filters the results using `HAVING`.

**Query from the interview guide:**

```
SELECT name, COUNT(*) AS cnt
FROM customers
GROUP BY name
HAVING COUNT(*) > 1;
```

**How it works:**

1. `SELECT name` retrieves the customer name.
2. `COUNT(*)` counts how many times each name appears.
3. `GROUP BY name` groups records with the same name.
4. `HAVING COUNT(*) > 1` returns only names that appear more than once.

**Example**

Consider the following `customers` table:

| customer_id | name  | city      |
| ----------- | ----- | --------- |
| 101         | Rahul | Hyderabad |
| 102         | Priya | Mumbai    |
| 103         | Rahul | Hyderabad |
| 104         | Amit  | Delhi     |
| 105         | Priya | Pune      |

The query produces:

| name  | cnt |
| ----- | --- |
| Rahul | 2   |
| Priya | 2   |

Both names appear twice, so they are identified as duplicates.

**How to find duplicates using multiple columns**

Sometimes, duplicate records must be identified using a combination of columns rather than just one column.

For example, if you want to find customers with the same name and city:

```
SELECT name, city, COUNT(*) AS cnt
FROM customers
GROUP BY name, city
HAVING COUNT(*) > 1;
```

Here, a record is considered a duplicate only when both the `name` and `city` values match.

**How to retrieve the complete duplicate records**

The original query identifies duplicate names and their counts. To retrieve the full customer records, you can use a subquery:

```
SELECT *
FROM customers
WHERE name IN (
    SELECT name
    FROM customers
    GROUP BY name
    HAVING COUNT(*) > 1
);
```

This returns all rows whose names occur more than once.

**Business Analyst application**

Identifying duplicate records is important for data quality because duplicates can affect:

- Customer counts and customer segmentation.
- Sales and revenue reports.
- Marketing campaign results.
- Business dashboards and management decisions.
- Data migration and system integration.

For example, if the same customer is registered twice, a report may overstate the total number of customers. A Business Analyst should identify the duplicates, understand why they occurred, and coordinate with the relevant team to resolve the issue.

**Interview-ready answer**

To find duplicate records in SQL, I use `GROUP BY`, `COUNT()` and `HAVING`.

For example:

```
SELECT name, COUNT(*) AS cnt
FROM customers
GROUP BY name
HAVING COUNT(*) > 1;
```

The `GROUP BY` clause groups records with the same name, `COUNT()` counts the occurrences, and `HAVING COUNT(*) > 1` filters records that appear more than once.

If duplicates need to be identified using multiple columns, I can include those columns in the `GROUP BY` clause.

As a Business Analyst, I would use this technique to identify data-quality issues and prevent duplicate records from affecting reports, customer counts and business decisions.


### **Q6. How do you calculate total sales per customer and find the top 5 customers?**

To calculate total sales per customer, we combine the `customers` and `orders` tables using a `JOIN`, calculate the total purchase amount using `SUM()`, group the results by customer, and sort them in descending order.

**SQL query from the interview guide:**

```
SELECT c.name,
       SUM(o.amount) AS total_sales
FROM customers c
JOIN orders o
    ON c.customer_id = o.customer_id
GROUP BY c.name
ORDER BY total_sales DESC
LIMIT 5;
```

For SQL Server, use `SELECT TOP 5` instead of `LIMIT 5`.

**Understanding the query step by step**

1. **FROM customers c** — Selects the customers table and assigns it the alias `c`.
2. **JOIN orders o** — Connects the orders table using the alias `o`.
3. **ON c.customer_id = o.customer_id** — Matches each order to the customer who placed it.
4. **SUM(o.amount)** — Adds all order amounts for each customer.
5. **GROUP BY c.name** — Groups orders by customer name so the total can be calculated for each customer.
6. **ORDER BY total_sales DESC** — Sorts customers from highest total sales to lowest.
7. **LIMIT 5** — Returns only the top five customers.

**Example**

Customers table:

| customer_id | name  |
| ----------- | ----- |
| 101         | Rahul |
| 102         | Priya |
| 103         | Amit  |
| 104         | Neha  |
| 105         | John  |

| order_id | customer_id | amount |
| -------- | ----------- | ------ |
| 1        | 101         | 5,000  |
| 2        | 102         | 8,000  |
| 3        | 101         | 3,000  |
| 4        | 103         | 10,000 |
| 5        | 104         | 4,000  |
| 6        | 102         | 2,000  |
| 7        | 105         | 7,000  |

After joining and calculating total sales:

| Customer | Total sales |
| -------- | ----------- |
| Amit     | 10,000      |
| Rahul    | 8,000       |
| Priya    | 10,000      |
| John     | 7,000       |
| Neha     | 4,000       |

The query sorts these totals in descending order and returns the top five customers.

**Why are JOIN, GROUP BY, SUM and ORDER BY important here?**

- `JOIN` combines related information from different tables.
- `GROUP BY` creates one group for each customer.
- `SUM()` calculates the total sales for each group.
- `ORDER BY` ranks customers based on their sales.

**Business Analyst application**

A Business Analyst can use this query to identify the highest-revenue customers and support business decisions such as:

For example, management may want to identify the top five customers contributing the most revenue so that the sales team can prioritize customer relationships.

**Interview-ready answer**

To find the top five customers based on total sales, I would join the customers and orders tables using the customer ID.

I would use the `SUM()` function to calculate the total order amount for each customer, `GROUP BY` to group the records by customer, and `ORDER BY DESC` to sort customers from highest sales to lowest.

This analysis is useful for identifying high-value customers, understanding sales performance and supporting business decisions.


### **Q7. How do you find customers who have never placed an order in SQL?**

To identify customers who have never placed an order, use a **LEFT JOIN** between the `customers` and `orders` tables, followed by a `WHERE` condition that checks for `NULL` values in the orders table.

**SQL query from the interview guide:**

```
SELECT c.name
FROM customers c
LEFT JOIN orders o
    ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL;
```

**How it works:**

1. `FROM customers c` — Starts with the customers table.
2. `LEFT JOIN orders o` — Includes every customer, whether or not they have placed an order.
3. `ON c.customer_id = o.customer_id` — Matches customers with their orders using customer ID.
4. `WHERE o.order_id IS NULL` — Selects customers for whom no matching order exists.

**Example**

Customers table:

| customer_id | name  |
| ----------- | ----- |
| 101         | Rahul |
| 102         | Priya |
| 103         | Amit  |
| 104         | Neha  |

| order_id | customer_id | amount |
| -------- | ----------- | ------ |
| 1        | 101         | 5,000  |
| 2        | 102         | 8,000  |
| 3        | 101         | 3,000  |

After performing the `LEFT JOIN`:

| Customer | Order ID |
| -------- | -------- |
| Rahul    | 1        |
| Rahul    | 3        |
| Priya    | 2        |
| Amit     | NULL     |
| Neha     | NULL     |

The final `WHERE` condition selects only records where `order_id` is `NULL`.

Result

| Customer |
| -------- |

These customers have never placed an order.

**Why use LEFT JOIN instead of INNER JOIN?**

- `INNER JOIN` returns only customers who have matching orders.

**Interview-ready answer**

To find customers who have never placed an order, I would use a `LEFT JOIN` between the customers and orders tables based on customer ID.

A LEFT JOIN returns all customers, including those who do not have matching orders. I would then use `WHERE o.order_id IS NULL` to identify customers without any orders.

### **Q8. How do you calculate monthly sales trends in SQL?**

A monthly sales trend shows how sales change over time. It helps a Business Analyst identify sales growth, declining performance, seasonal patterns and changes in customer demand.

To calculate monthly sales, use the `YEAR()`, `MONTH()` and `SUM()` functions along with `GROUP BY` and `ORDER BY`.

**SQL query from the interview guide:**

```
SELECT
    YEAR(order_date) AS yr,
    MONTH(order_date) AS mth,
    SUM(amount) AS sales
FROM orders
GROUP BY
    YEAR(order_date),
    MONTH(order_date)
ORDER BY
    yr,
    mth;
```

**How it works:**

1. `YEAR(order_date)` extracts the year from the order date.
2. `MONTH(order_date)` extracts the month.
3. `SUM(amount)` calculates the total sales for that month.
4. `GROUP BY` groups orders by year and month.
5. `ORDER BY yr, mth` sorts the results chronologically.

**Example**

Consider this orders table:

| order_id | order_date | amount |
| -------- | ---------- | ------ |
| 1        | 2026-01-05 | 5,000  |
| 2        | 2026-01-15 | 3,000  |
| 3        | 2026-02-10 | 7,000  |
| 4        | 2026-02-20 | 4,000  |
| 5        | 2026-03-12 | 6,000  |
| 6        | 2026-03-25 | 2,000  |

The query produces:

| Year | Month | Total sales |
| ---- | ----- | ----------- |
| 2026 | 1     | 8,000       |
| 2026 | 2     | 11,000      |
| 2026 | 3     | 8,000       |

This shows that February had the highest sales among these three months.

**Why do we group by both year and month?**

Grouping only by month can combine January sales from different years.

For example, January 2025 and January 2026 would be treated as one group if we grouped only by `MONTH(order_date)`.

Grouping by both year and month keeps each month's sales separate across different years.

**Business Analyst application**

Monthly sales trend analysis helps businesses:

- Compare sales performance across months.
- Identify seasonal demand patterns.
- Monitor revenue growth or decline.
- Evaluate marketing campaign performance.
- Support monthly and quarterly business reviews.
- Forecast future sales using historical trends.

For example, if sales decline for two consecutive months, a Business Analyst can investigate whether the decline is related to fewer orders, lower order values, specific regions or particular products.

**Interview-ready answer**

To calculate monthly sales trends in SQL, I would use the `YEAR()`, `MONTH()` and `SUM()` functions.

I would extract the year and month from the order date, calculate total sales using `SUM(amount)`, and group the records by year and month.

Finally, I would use `ORDER BY` to arrange the results chronologically.

This analysis helps a Business Analyst monitor monthly revenue, identify seasonal trends, compare performance and support business forecasting.


### **Q9. How do you find the highest-paid employee in each department using SQL?**

To find the highest-paid employee in each department, use the `RANK()`**&#x20;window function** with `PARTITION BY` and `ORDER BY`.

The `RANK()` function ranks employees according to their salary within each department.

**SQL query from the interview guide:**

```
SELECT *
FROM (
    SELECT
        name,
        dept_id,
        salary,
        RANK() OVER (
            PARTITION BY dept_id
            ORDER BY salary DESC
        ) AS rnk
    FROM employees
) t
WHERE rnk = 1;
```

**How it works:**

1. `PARTITION BY dept_id` divides employees into separate department groups.
2. `ORDER BY salary DESC` sorts employees within each department from highest salary to lowest.
3. `RANK()` assigns a ranking to each employee within their department.
4. `AS rnk` gives the ranking column an alias.
5. The outer query uses `WHERE rnk = 1` to retrieve the highest-paid employee or employees in each department.

**Example**

Consider this employees table:

| Name  | Department ID | Salary |
| ----- | ------------- | ------ |
| Rahul | 10            | 80,000 |
| Priya | 10            | 60,000 |
| Amit  | 10            | 80,000 |
| Neha  | 20            | 90,000 |
| John  | 20            | 70,000 |
| Sara  | 30            | 55,000 |

After applying `RANK()`:

| Name  | Department ID | Salary | Rank |
| ----- | ------------- | ------ | ---- |
| Rahul | 10            | 80,000 | 1    |
| Amit  | 10            | 80,000 | 1    |
| Priya | 10            | 60,000 | 3    |
| Neha  | 20            | 90,000 | 1    |
| John  | 20            | 70,000 | 2    |
| Sara  | 30            | 55,000 | 1    |

The final output is:

| Name  | Department ID | Salary |
| ----- | ------------- | ------ |
| Rahul | 10            | 80,000 |
| Amit  | 10            | 80,000 |
| Neha  | 20            | 90,000 |
| Sara  | 30            | 55,000 |

Notice that both Rahul and Amit are returned because they share the highest salary in Department 10.

**What is the purpose of PARTITION BY?**

`PARTITION BY` divides the data into groups before applying a window function.

In this query, it ensures that salaries are ranked separately within each department rather than across the entire company.

Without `PARTITION BY dept_id`, SQL would rank all employees together, returning only the highest-paid employee across the whole organization.

**Business Analyst application**

A Business Analyst can use this query to:

- Compare compensation across departments.
- Identify top-performing employees by department.
- Prepare department-level salary reports.
- Support compensation benchmarking.
- Analyse salary distribution for management reporting.

**Interview-ready answer**

To find the highest-paid employee in each department, I would use the `RANK()` window function.

I would use `PARTITION BY dept_id` to rank employees separately within each department and `ORDER BY salary DESC` to rank salaries from highest to lowest.

Then I would use an outer query with `WHERE rnk = 1` to retrieve the highest-paid employee or employees in each department.

I prefer this approach because it also handles situations where multiple employees share the highest salary in a department.
